## Two Pointer Pattern Approach

### 🎯 Learning Goal

By the end of this note, I will understand what the Two Pointer technique is, when to use it (arrays, linked lists, sorted data), and how to apply it to solve pair-sum, triplet-sum, remove-duplicates, reverse, merge, and cycle-detection problems — with working code for each.

---

### 🤔 What is it?

The **Two Pointer** technique uses two index variables ("pointers") that move through a data structure — usually from opposite ends, or at different speeds — instead of using nested loops to compare every pair of elements.

> 🧸 Think of two people looking for each other in a single-file line: one starts from the front, one from the back, and they both walk toward the middle. That's way faster than everyone in the line individually checking against everyone else.

---

### ❓ Why do we need it?

- A naive approach to "find a pair that sums to X" uses two nested loops → **O(n²)**.
- The two-pointer technique solves the same problem in a **single pass** → **O(n)** (once the data is sorted).
- It avoids extra memory too — most two-pointer solutions run in **O(1) space**, unlike hashmap-based approaches.

---

### 🧠 Key Idea

- Two pointers can move **toward each other** (from both ends — used for pair/triplet sum on sorted arrays) or **in the same direction** (used for remove-duplicates, or fast/slow pointers for cycle detection).
- Works best on **arrays and linked lists**.
- Usually requires the data to be **sorted** (for sum-finding problems) — though some variants (like removing duplicates, or fast/slow cycle detection) don't need sorting at all.
- Common problem types: **pair find, triplet find, quadruplet find, merge, remove, rearrange, detect cycle**.

---

### 📚 Important Terms

| Term | Simple Meaning | Example |
|------|----------------|----------|
| Left pointer | Index starting from the beginning | `left = 0` |
| Right pointer | Index starting from the end | `right = len(arr) - 1` |
| Slow pointer | Moves one step at a time | Used in cycle detection |
| Fast pointer | Moves two steps at a time | Used in cycle detection |
| Converging pointers | Pointers moving toward each other | Pair sum in sorted array |
| Same-direction pointers | Both pointers move forward, at different paces | Remove duplicates |

---

### 🔄 How it Works

```mermaid
flowchart LR
Start["left = 0, right = n-1"] --> Check{"arr[left] + arr[right] vs target"}
Check -->|"sum == target"| Found["Pair found!"]
Check -->|"sum < target"| MoveLeft["left += 1 (need a bigger sum)"]
Check -->|"sum > target"| MoveRight["right -= 1 (need a smaller sum)"]
MoveLeft --> Check
MoveRight --> Check
```

---

### 🌍 Real-Life Example

Imagine you're matching dance partners by height, and everyone is already lined up shortest to tallest. Instead of checking every possible pair (which takes forever), you put the shortest and tallest person together, check if they fit your rule, and adjust from there — moving inward until everyone's matched. That's the two-pointer approach.

---

### ✅ When Should I Use Two Pointers?

| Situation | Use Two Pointers? |
|-----------|---------------------|
| Data structure is an **array or linked list** | ✅ Yes |
| Data structure is a hashmap/tree/graph | ❌ No — different techniques apply |
| Array is **sorted** (or can be sorted) and you need pair/triplet/quadruplet sums | ✅ Yes — converging pointers |
| You need to **remove/rearrange elements in-place** | ✅ Yes — same-direction pointers |
| You need to **merge two sorted arrays/lists** | ✅ Yes — one pointer per list |
| You need to **detect a cycle in a linked list** | ✅ Yes — fast/slow pointers (Floyd's algorithm) |

### 💻 Technical Examples

#### 1️⃣ Pair Sum in a Sorted Array | LEETCODE - 167 (Two Sum II)

**Problem:** Given a sorted array, find two numbers that add up to a target value.

**Approach:** Start `left` at index 0 and `right` at the last index. If the sum is too small, move `left` forward (to increase the sum). If too big, move `right` backward (to decrease it). If equal, we found our pair.

````python
from typing import List

def pairSum(nums: List[int], target: int) -> List[int]:
    left, right = 0, len(nums) - 1
    while left < right:
        current_sum = nums[left] + nums[right]
        if current_sum == target:
            return [left, right]
        elif current_sum < target:
            left += 1
        else:
            right -= 1
    return []

nums = [2, 7, 11, 15]
target = 9
print(pairSum(nums, target))
````

#### 2️⃣ Triplet Sum (3Sum) | LEETCODE - 15

**Problem:** Find all unique triplets in the array that sum to zero.

**Approach:** Sort the array first. Fix one number with an outer loop, then use two pointers (converging) on the rest of the array to find pairs that complete the triplet — this turns an O(n³) brute force into O(n²).

````python
def tripletSum(nums: List[int]) -> List[List[int]]:
    nums.sort()
    result = []
    for i in range(len(nums) - 2):
        if i > 0 and nums[i] == nums[i - 1]:
            continue  # skip duplicate first numbers
        left, right = i + 1, len(nums) - 1
        while left < right:
            total = nums[i] + nums[left] + nums[right]
            if total == 0:
                result.append([nums[i], nums[left], nums[right]])
                left += 1
                right -= 1
                while left < right and nums[left] == nums[left - 1]:
                    left += 1  # skip duplicate pairs
            elif total < 0:
                left += 1
            else:
                right -= 1
    return result

nums = [-1, 0, 1, 2, -1, -4]
print(tripletSum(nums))
````

#### 3️⃣ Remove Duplicates from Sorted Array | LEETCODE - 26

**Problem:** Remove duplicates in-place from a sorted array, so each element appears once, and return the new length.

**Approach:** This uses **same-direction pointers**. `slow` marks the last position of a unique element; `fast` scans ahead looking for the next new value. When `fast` finds a new value, we copy it right after `slow`.

````python
def removeDuplicates(nums: List[int]) -> int:
    if not nums:
        return 0
    slow = 0
    for fast in range(1, len(nums)):
        if nums[fast] != nums[slow]:
            slow += 1
            nums[slow] = nums[fast]
    return slow + 1

nums = [0, 0, 1, 1, 1, 2, 2, 3, 3, 4]
new_length = removeDuplicates(nums)
print("New length:", new_length)
print("Array after dedup:", nums[:new_length])
````

#### 4️⃣ Reverse a String In-Place | LEETCODE - 344

**Problem:** Reverse an array of characters in-place.

**Approach:** Converging pointers again, but this time we **swap** elements instead of comparing sums.

````python
def reverseString(s: List[str]) -> None:
    left, right = 0, len(s) - 1
    while left < right:
        s[left], s[right] = s[right], s[left]
        left += 1
        right -= 1

chars = ["h", "e", "l", "l", "o"]
reverseString(chars)
print(chars)
````

#### 5️⃣ Merge Two Sorted Arrays | LEETCODE - 88

**Problem:** Merge two sorted arrays into one sorted array.

**Approach:** Use **one pointer per array**. Compare the current elements of both, pick the smaller one, and advance that pointer. This is the same idea used in Merge Sort's "merge" step.

````python
def mergedSort(arr1: list[int], arr2: list[int]) -> list[int]:

    i = 0
    j = 0

    n = len(arr1)
    m = len(arr2)

    new_arr = []

    while (i < n) and (j < m):
        if (arr1[i] <= arr2[j]):
            new_arr.append(arr1[i])
            i += 1
        else:
            new_arr.append(arr2[j])
            j += 1

    while (i < n):
        new_arr.append(arr1[i])
        i += 1

    while (j < m):
            new_arr.append(arr2[j])
            j += 1

    return new_arr



arr1 = [1, 3, 5, 7]
arr2 = [2, 4, 6, 8, 10]
print(mergedSort(arr1, arr2)) 

````

````python
def mergeSorted(arr1: List[int], arr2: List[int]) -> List[int]:
    i, j = 0, 0
    merged = []
    while i < len(arr1) and j < len(arr2):
        if arr1[i] <= arr2[j]:
            merged.append(arr1[i])
            i += 1
        else:
            merged.append(arr2[j])
            j += 1
    merged.extend(arr1[i:])  # leftover elements, if any
    merged.extend(arr2[j:])
    return merged

arr1 = [1, 3, 5, 7]
arr2 = [2, 4, 6, 8, 10]
print(mergeSorted(arr1, arr2))
````

#### 5️⃣ Squares of a Sorted Array | LEETCODE - 977

**Problem:** Given an integer array nums sorted in non-decreasing order, return an array of the squares of each number sorted in non-decreasing order.

**FollowUp:** Follow up: Squaring each element and sorting the new array is very trivial, could you find an O(n) solution using a different approach?

````python
def squareSortedArray(nums: list[int]) -> list[int]:

    siz = len(nums)
    neg = []
    pos = []

    # seperate negative and postive values
    for num in nums:
        if num < 0:
            neg.append(num)
        else:
            pos.append(num)

    
    # Case 1: No negative numbers
    if len(neg) == 0:
        return [x * x for x in pos]

    # Case 2: No Positive numbers
    if len(pos) == 0:
        res = [x * x for x in neg]
        res.reverse()
        return res

    # case 3: Both Exist
    neg = [ x * x for x in neg][::-1] # use this for reverse
    pos = [ x * x for x in pos]
    n, m = len(neg), len(pos)
    res = []

    i = j = 0
    while i < n and j < m:
        if neg[i] <= pos[j]:
            res.append(neg[i])
            i += 1
        else:
            res.append(pos[j])
            j += 1

    while i < n:
        res.append(neg[i])
        i += 1

    while j < m:
        res.append(pos[j])
        j += 1

    return res

nums = [-10, -4, -1, 0, 3, 9]
print(squareSortedArray(nums))
````

#### 6️⃣ Detect Cycle in a Linked List (Floyd's Algorithm) | LEETCODE - 141

**Problem:** Given a linked list, determine if it has a cycle (a node points back to an earlier node, creating a loop).

**Approach:** Use a **slow pointer** (moves 1 step at a time) and a **fast pointer** (moves 2 steps at a time). If there's a cycle, the fast pointer will eventually "lap" the slow pointer and they'll meet. If there's no cycle, the fast pointer reaches the end (`None`) first.

> 🧸 Think of two runners on a circular track — the faster runner will eventually catch up to the slower one and lap them. But if the track is a straight line (no loop), the faster runner just finishes first and there's no "catching up."

````python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def hasCycle(head: ListNode) -> bool:
    slow, fast = head, head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
        if slow == fast:
            return True
    return False

# Build a list WITHOUT a cycle: 1 -> 2 -> 3 -> None
n3 = ListNode(3)
n2 = ListNode(2, n3)
n1 = ListNode(1, n2)
print("No cycle list:", hasCycle(n1))

# Build a list WITH a cycle: 1 -> 2 -> 3 -> back to 2
n3.next = n2
print("Cyclic list:", hasCycle(n1))
````

### ⚖ Comparison: Two Pointer Variants

| Variant | Pointer Movement | Used For |
|---------|-------------------|----------|
| Converging | Start at both ends, move toward center | Pair sum, triplet sum, reverse |
| Same-direction | Both start near the beginning, move forward at different paces | Remove duplicates, partitioning |
| Fast/slow | One moves 1 step, other moves 2 steps | Cycle detection, finding the middle of a list |
| One-per-structure | Each pointer walks a separate array/list | Merging two sorted arrays |

---

### ⚠ Common Misconceptions

❌ Two pointers only work on sorted arrays.
✅ Sorting is required for sum-finding problems (pair/triplet sum), but fast/slow cycle detection and remove-duplicates-style problems don't need sorted data — just the right pointer strategy.

❌ Two pointers is the same as using two nested loops.
✅ Nested loops re-check pairs from scratch (O(n²)); two pointers make a single coordinated pass (O(n)) because each pointer only moves forward (or toward the other), never backtracking over ground already covered.

❌ Fast/slow pointers are only for cycle detection.
✅ They're also used to find the middle of a linked list, detect palindromes, and more — anywhere you need one pointer to "see ahead" of the other.

### 📝 Quick Revision

- Two pointers replace nested loops with a coordinated single pass — usually O(n) time and O(1) space.
- **Converging pointers** (start/end, move inward) solve pair-sum, triplet-sum, and in-place reversal.
- **Same-direction pointers** (slow/fast at different paces) solve remove-duplicates and partitioning problems.
- **Fast/slow pointers** (1 step vs 2 steps) detect cycles in linked lists (Floyd's algorithm).
- **One-pointer-per-structure** merges two sorted arrays or lists.
- Sum-finding two-pointer problems require the array to be sorted first.
- Works on arrays and linked lists — not directly applicable to hashmaps, trees, or graphs.

| Pattern | One-Line Meaning |
|----------|------------------|

| Remove Duplicates | Slow pointer marks unique position, fast pointer scans ahead |
| Reverse In-Place | Converging pointers, swap instead of compare |

1. Two pointers turn many O(n²) nested-loop problems into O(n) single-pass solutions, usually with O(1) extra space.
2. **Converging pointers** (opposite ends, moving inward) are the go-to for pair/triplet sum problems on sorted arrays.
3. **Same-direction pointers** (slow marks position, fast scans ahead) handle in-place removal and rearrangement.
