## 🤔 What is it?

A **dictionary** is a collection of **key-value pairs** — instead of looking things up by position (like a list), you look them up by a meaningful **key**.

> 🧸 Think of a dictionary exactly like a real paper dictionary: you don't read page by page to find a word's meaning — you jump straight to the word (the key) and read its definition (the value). A list is more like a photo album where you flip through page by page (index by index) until you find what you want.

### Core Properties

- **Key-Value Pairs** — every item is a `key: value` mapping, not a single value.
- **Keys must be unique** — assigning to an existing key overwrites its value, it never creates a duplicate.
- **Keys must be hashable (immutable)** — numbers, strings, and tuples can be keys; lists and other dicts cannot.
- **Ordered** (Python 3.7+) — dictionaries remember insertion order, though you should still look items up by key, not position.
- **Mutable** — you can add, change, or remove key-value pairs after creation.

---

## ❓ Why do we need it?

- Looking something up by a meaningful name (`student["roll_no"]`) is far more readable than remembering "that value is at index 7."
- Dictionaries give near-instant lookup (**O(1) average**) no matter how large they get, unlike searching through a list one item at a time (**O(n)**).
- Many real-world problems are naturally "map this thing to that thing" — a phone book (name → number), an inventory (item → quantity), a frequency count (character → count).

---

## 🧠 Key Idea

- `dict[key]` raises a `KeyError` if the key doesn't exist; `dict.get(key)` returns `None` instead (or a default you choose) — these are NOT interchangeable.
- Any **immutable** type can be a key: `int`, `float`, `str`, `bool`, or `tuple` (as long as the tuple itself only contains immutable items). **Mutable** types (`list`, `dict`, `set`) can never be keys.
- `dict1 = dict2` creates an alias (same object); `dict1.copy()` creates a shallow copy — the exact same reference-vs-copy distinction covered for lists and tuples.
- Dictionaries achieve their speed through **hashing** — covered in its own section below.

## 🔑 Accessing Values: `[]` vs `.get()`

> ❌ Misconception: `d[key]` and `d.get(key)` behave the same way.
> ✅ Correct: `d[key]` raises a `KeyError` if the key doesn't exist. `d.get(key)` returns `None` instead (or a fallback value you specify) — no crash. Use `.get()` whenever a missing key is a normal, expected possibility.

````python
rollDict = {1: "Abhinav", 2: "Pratham", 3: "Rahul", 4: "Sita"}

print(rollDict[2])          # works -- key 2 exists
print(rollDict.get(2))       # same result, via get()
print(rollDict.get(99))      # key 99 doesn't exist -- returns None, no crash
print(rollDict.get(99, "Not Found"))  # get() can take a custom fallback value
````

````python
rollDict = {1: "Abhinav", 2: "Pratham", 3: "Rahul", 4: "Sita"}

rollDict[99]   # KeyError: 99 -- this is exactly why .get() exists for "might not be there" lookups
````

## ➕ Adding and Updating Entries

Assigning to a NEW key adds an entry; assigning to an EXISTING key overwrites its value. `update()` can add/overwrite several keys at once.

````python
rollDict = {1: "Abhinav", 2: "Pratham", 3: "Rahul", 4: "Sita"}

rollDict[5] = "Yash"          # new key -> adds an entry
print(rollDict)

rollDict[2] = "Aman"          # existing key -> overwrites the value
print(rollDict)

rollDict.update({1: "Mayank", 10: "Yuvraj"})   # update multiple keys at once
print(rollDict)
````

## ➖ Removing Entries: `pop()`, `popitem()`, `del`, `clear()`

| Method | What it Does |
|--------|----------------|
| `d.pop(key)` | Removes `key` and returns its value; raises `KeyError` if missing |
| `d.popitem()` | Removes and returns the LAST inserted key-value pair (as a tuple) |
| `del d[key]` | Removes `key`; raises `KeyError` if missing (no return value) |
| `del d` | Deletes the ENTIRE dictionary variable, not just one entry |
| `d.clear()` | Removes every entry, leaving an empty dict `{}` |

````python
rollDict = {1: "Abhinav", 2: "Pratham", 3: "Rahul", 4: "Sita"}

rollDict.pop(1)
print(rollDict[2])
print(rollDict.get(2))
print(rollDict)
````

````python
my_dict = {"1": "JavaScript", "2": "Python", "3": "Machine Learning"}
my_dict["2"] = "Junk Food"
my_dict["3"] = "Pizza Food"
del my_dict["2"]   # remove key "2"
print(my_dict)
````

````python
my_dict = {"harry": "Code", "1": "JavaScript", "2": "Python", "3": "Machine Learning"}

my_dict.popitem()    # removes the LAST inserted item (guaranteed order since Python 3.7)
print(my_dict)
````

````python
my_dict = {"harry": "Code", "1": "JavaScript", "2": "Python", "3": "Machine Learning"}

my_dict.clear()
print(my_dict)
````

`del my_dict` (without a key) removes the ENTIRE variable — trying to use it afterward raises a `NameError`, not a `KeyError`. This is a genuinely different kind of error, worth telling apart.

````python
my_dict = {"harry": "Code", "1": "JavaScript"}

del my_dict          # deletes the WHOLE dictionary, not one key
print(my_dict)        # NameError: name 'my_dict' is not defined -- the variable itself is gone
````

## 🔗 Reference vs Shallow Copy

Exactly the same distinction covered for lists ([09_List.ipynb](09_List.ipynb)) — `dict2 = dict1` is an ALIAS (same object), `dict1.copy()` is a genuinely new (but shallow) dictionary.

````python
dict1 = {1: "Abhinav", 2: "Pratham", 3: "Rahul", 4: "Sita"}
dict2 = dict1          # alias -- NOT a copy, same object as dict1
dict3 = dict1.copy()    # shallow copy -- a genuinely separate dict object

dict1.pop(2)

print(dict2)   # changed too! dict2 IS dict1
print(dict3)   # unaffected -- dict3 is a separate object

dict3.clear()
print(dict3)
````

## 📋 Membership: `in`

`in` checks against the KEYS by default, not the values.

````python
my_dict = {"harry": "Code", "1": "JavaScript", "2": "Python", "3": "Machine Learning"}

if "harry" in my_dict:
    print("Yes, 'harry' is one of the keys in my_dict")

print("Code" in my_dict)          # False -- "Code" is a VALUE, not a key
print("Code" in my_dict.values())  # True -- explicitly check the values instead
````

## 🧩 `setdefault()` — Get a Value, or Insert a Default if Missing

`d.setdefault(key, default)` returns the existing value if `key` is present. If it's NOT present, it inserts `key` with `default` and returns that default — all in one call, instead of writing an `if key not in d:` check yourself.

````python
freq = {}

freq.setdefault("a", 0)   # "a" doesn't exist yet -> inserts it with value 0, returns 0
print(freq)

freq["a"] += 1
freq.setdefault("a", 0)   # "a" already exists -> returns its CURRENT value, does NOT reset it to 0
print(freq)
````

> ⚠ Gotcha: if the default value is MUTABLE (like a list), every key that falls back to that default shares the SAME object — exactly the shallow-copy trap seen earlier.

````python
groups = {}

# Using setdefault to build up lists of values per key
pairs = [("fruit", "apple"), ("veg", "carrot"), ("fruit", "banana")]
for category, item in pairs:
    groups.setdefault(category, []).append(item)

print(groups)
````

## 🏗 `fromkeys()` — Build a Dictionary from a List of Keys

`dict.fromkeys(iterable, value)` creates a new dictionary where every key from `iterable` maps to the SAME `value`.

> ⚠ Gotcha: if `value` is mutable (like a list), EVERY key ends up pointing to the exact same list object — mutating one "key's list" mutates them all.

````python
keys = ["a", "b", "c"]
new_dict = dict.fromkeys(keys, 0)
print(new_dict)
````

````python
# The mutable-default gotcha
shared_default = dict.fromkeys(["a", "b"], [])
shared_default["a"].append(1)
print(shared_default)   # both "a" and "b" changed -- they share the SAME list object!
````

Just like a list comprehension, a dictionary comprehension builds a new dict in one line: `{key_expr: value_expr for item in iterable}`.

````python
numbers = [1, 2, 3, 4, 5]
squares = {n: n**2 for n in numbers}
print(squares)
````

## ⏱ Time and Space Complexity of Dictionary Operations

> 💡 This is WHY dictionaries feel "instant" — see the Hashing section right below for exactly how that O(1) average is achieved.

| Operation | Code | Time (Average) | Time (Worst Case) | Why |
|-----------|------|------------------|---------------------|-----|
| Access by key | `d[key]` | O(1) | O(n) | Hash lookup; worst case only with many hash collisions |
| Insert / Update | `d[key] = v` | O(1) | O(n) | Same hashing mechanism as access |
| Delete | `del d[key]`, `d.pop(key)` | O(1) | O(n) | Same hashing mechanism |
| Membership | `key in d` | O(1) | O(n) | Checks the hash table directly, not a linear scan |
| `get()` / `setdefault()` | `d.get(key)` | O(1) | O(n) | Same hashing mechanism as access |
| Iteration | `for k in d`, `.items()` | O(n) | O(n) | Must visit every entry once |
| Copy (shallow) | `d.copy()` | O(n) | O(n) | Must copy every key-value reference |
| `update(other)` | `d.update(other)` | O(len(other)) | O(len(other)) | One insert per entry in `other` |
| `fromkeys()` | `dict.fromkeys(keys)` | O(n) | O(n) | One insert per key |
| `clear()` | `d.clear()` | O(n) | O(n) | Must release every entry |

## 🔐 Hashing in Python

### 🤔 What is it?

**Hashing** is a technique that converts a key into a fixed-size number (a **hash**) using a **hash function**. That number tells Python exactly which "slot" in memory to store or look up the value — no searching required.

> 🧸 Think of a hash like a locker number system at a gym. Instead of checking every locker one by one for your bag, a formula instantly tells you "your bag is in locker #42." Hashing does the same thing for dictionary keys — `hash(key)` instantly points to the right slot.

### ❓ Why do we need it?

Without hashing, looking up a key would mean scanning every single entry until a match is found — exactly like searching an unsorted list, which is O(n). Hashing is what makes dictionary (and set) lookups **O(1) on average**, regardless of how many entries exist.

### 🧠 Key Idea

- `hash(key)` computes a key's hash — try it yourself: `hash("hello")`, `hash(42)`.
- This is EXACTLY why dictionary keys must be **hashable (immutable)** — a mutable object's hash could change after it's stored, which would corrupt the lookup table. This is the same reason lists can't be dict keys or set members (see [10_Tupple.ipynb](10_Tupple.ipynb)).
- Two different keys CAN produce the same hash (a **collision**) — Python handles this internally, but heavy collisions are what push the worst case from O(1) to O(n).

````python
print(hash(42))
print(hash("hello"))
print(hash((1, 2, 3)))   # tuples are hashable

# hash([1, 2, 3])   # TypeError: unhashable type: 'list' -- lists have no stable hash, since they can change
````

### Two Sum | LEETCODE - 1

**Problem:** Given an array of integers and a target, return the indices of the two numbers that add up to the target.

**Example:** `nums = [2,7,11,15]`, `target = 9` → `[0,1]`

**Approach:** Walk through the array once. For each number, check if `target - number` was already SEEN (stored as a dict key mapped to its index). If so, we've found our pair instantly — no nested loop needed.

**Complexity:** O(n) time, O(n) space — this is exactly the O(n²) → O(n) speed-up that hashing enables.

````python
from typing import List

class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        n = len(nums)
        seen = {}   # maps: number we've already seen -> its index

        for i in range(n):
            rem = target - nums[i]
            if rem in seen:
                return [seen[rem], i]
            seen[nums[i]] = i
        return []


list1 = [2, 7, 11, 15]
obj = Solution()
print("Target 9:", obj.twoSum(list1, 9))
````

### Two Sum II - Input Array Is Sorted | LEETCODE - 167

**Problem:** Same idea, but the array is already SORTED, and you must use only constant extra space (so a hashmap is off the table this time).

**Approach:** Since the array is sorted, a **converging Two Pointer** (see [08_DsaPattern_part1.ipynb](08_DsaPattern_part1.ipynb)) works in O(1) extra space — no dictionary needed at all here. It's included in this notebook to contrast with the O(n)-space hashmap approach above: sorted input unlocks a cheaper technique.

**Complexity:** O(n) time, O(1) extra space.

> ⚠ Bug fixed: the original code defined a class named `solution` (lowercase) but never used it — it then called `Solution().twoSum(...)` (the PREVIOUS problem's method!) instead of the intended `twoSumII(...)`. Because the earlier cell's `Solution` class was still in memory, this silently ran the WRONG algorithm and returned `[0, 1]` (the 0-indexed Two Sum I answer) instead of the correct 1-indexed Two Sum II answer, `[1, 2]`.

````python
from typing import List

class Solution:
    def twoSumII(self, numbers: List[int], target: int) -> List[int]:
        left = 0
        right = len(numbers) - 1

        while left < right:
            total = numbers[left] + numbers[right]
            if total == target:
                return [left + 1, right + 1]   # 1-indexed, as the problem requires
            elif total > target:
                right -= 1
            else:
                left += 1
        return []

numbers = [2, 7, 11, 15]
obj = Solution()
print("Target 9:", obj.twoSumII(numbers, 9))
````

### Intersection of Two Arrays | LEETCODE - 349

**Problem:** Return the unique elements common to both arrays.

**Approach:** Convert `nums2` to a `set` for O(1) membership checks, then walk `nums1` once collecting matches.

**Complexity:** O(n + m) time, O(n + m) space.

````python
from typing import List

class Solution:
    def findIntersections(self, nums1: List[int], nums2: List[int]) -> List[int]:
        result = []
        set2 = set(nums2)

        for i in nums1:
            if i in set2:
                result.append(i)
        return list(set(result))   # dedupe in case nums1 has repeats


nums1 = [1, 2, 2, 1]
nums2 = [2, 2]

obj = Solution()
print(obj.findIntersections(nums1, nums2))
````

### First Unique Character in a String | LEETCODE - 387

**Problem:** Find the index of the first character that does NOT repeat anywhere else in the string. Return `-1` if none exists.

**Example:** `s = "loveleetcode"` → `2` (the character `'v'` at index 2 never repeats)

**Approach:** First pass builds a frequency count per character. Second pass finds the first character whose count is exactly `1`.

**Complexity:** O(n) time, O(1) space (at most 26 lowercase letters, a constant).

> ⚠ Bug fixed: the original code checked `freq[s[i]] == i` — comparing a character's COUNT to its INDEX, which is not a meaningful comparison at all. It happened to compile and run without crashing, but silently returned the wrong answer (`-1` instead of `2`) for the example above. The fix is `freq[s[i]] == 1` — check the count against ONE, meaning "this character appears exactly once."

````python
class Solution:
    def firstUniqChar(self, s: str) -> int:
        freq = {}
        for ch in s:
            freq[ch] = freq.get(ch, 0) + 1

        for i in range(len(s)):
            if freq[s[i]] == 1:   # fixed: compare the COUNT to 1, not to the index i
                return i

        return -1


s = "loveleetcode"
obj = Solution()
print(obj.firstUniqChar(s))
````

### Valid Anagram | LEETCODE - 242

**Problem:** Return `True` if `t` is an anagram of `s` (same letters, same counts, any order).

**Approach:** Count every character in `s`, then walk `t` decrementing those counts. If any character in `t` is missing or over-used, they're not anagrams.

**Complexity:** O(n) time, O(1) space (bounded alphabet).

````python
class Solution:
    def checkAnagram(self, s: str, t: str) -> bool:
        if len(s) != len(t):
            return False

        freq = {}
        for ch in s:
            freq[ch] = freq.get(ch, 0) + 1

        for ch in t:
            if ch not in freq or freq[ch] == 0:
                return False
            freq[ch] -= 1

        return True


s = "anagram"
t = "nagaram"
obj = Solution()
print(obj.checkAnagram(s, t))
````

### Group Anagrams | LEETCODE - 49

**Problem:** Group a list of strings so that all anagrams of each other end up in the same group.

**Approach:** Anagrams share the same SORTED form (`"eat"` and `"tea"` both sort to `"aet"`). Use that sorted string as a dictionary key, and collect the originals in a list per key.

**Complexity:** O(n · k log k) time (n strings, each of length up to k, each needing a sort), O(n · k) space.

````python
from typing import List

class Solution:
    def sortString(self, s):
        return "".join(sorted(s))

    def groupAnagram(self, strs: List[str]) -> List[List[str]]:
        groups = {}

        for s in strs:
            key = self.sortString(s)
            groups.setdefault(key, []).append(s)   # setdefault avoids a manual if/else here

        return list(groups.values())


strs = ["eat", "tea", "tan", "ate", "nat", "bat"]
obj = Solution()
print(obj.groupAnagram(strs))
````

### Substrings of Size Three with Distinct Characters | LEETCODE - 1876 (Sliding Window)

**Problem:** Count how many length-3 substrings have all distinct characters.

**Approach:** A **sliding window** of fixed size 3 — slide across the string, and for each window just check the 3 characters pairwise.

**Complexity:** O(n) time, O(1) space.

````python
class Solution:
    def countGoodSubstrings(self, s: str) -> int:
        n = len(s)
        count = 0

        for i in range(0, n - 2):
            if s[i] != s[i + 1] and s[i + 1] != s[i + 2] and s[i] != s[i + 2]:
                count += 1
        return count


s = "aababcabc"
obj = Solution()
print(obj.countGoodSubstrings(s))
````

### Longest Substring Without Repeating Characters | LEETCODE - 3 (Sliding Window + Hashmap)

**Problem:** Find the length of the longest substring with no repeating characters.

**Example:** `s = "abcabcbb"` → `3` (the substring `"abc"`)

**Approach:** A **variable-size sliding window** (`left`/`right`), tracking the most recent index each character was seen at in a dictionary. When a repeat is found INSIDE the current window, jump `left` forward to just past that repeat's earlier position.

**Complexity:** O(n) time, O(min(n, alphabet size)) space.

> ⚠ Bug fixed: the original function had NO body at all — just the signature followed by a blank line — which raises `IndentationError: expected an indented block` the moment Python tries to parse it. The function is implemented below.

````python
class Solution:
    def lengthOfLongestSubstring(self, s: str) -> int:
        last_seen = {}   # character -> most recent index it was seen at
        left = 0
        max_len = 0

        for right in range(len(s)):
            ch = s[right]
            if ch in last_seen and last_seen[ch] >= left:
                left = last_seen[ch] + 1   # jump left past the earlier occurrence
            last_seen[ch] = right
            max_len = max(max_len, right - left + 1)

        return max_len


for s in ["abcabcbb", "bbbbb", "pwwkew"]:
    obj = Solution()
    print(f"{s!r} -> {obj.lengthOfLongestSubstring(s)}")
````

## ⚠ Common Misconceptions

❌ `d[key]` and `d.get(key)` behave the same way.
✅ `d[key]` raises `KeyError` on a missing key; `d.get(key)` returns `None` (or a fallback) instead.

❌ Any value can be a dictionary key.
✅ Only HASHABLE (immutable) values can be keys — `int`, `str`, `float`, `bool`, `tuple`. Lists, dicts, and sets cannot.

❌ `dict2 = dict1` makes a copy.
✅ It creates another name for the SAME dictionary object — mutating one mutates the other. Use `.copy()` for a genuine (shallow) copy.

❌ Looping over a dictionary gives you `(key, value)` pairs by default.
✅ `for x in d:` gives you the KEYS only — use `.items()` explicitly for pairs.

❌ `dict.fromkeys(keys, [])` gives every key its OWN empty list.
✅ Every key shares the exact SAME list object — mutating one key's list affects all of them.

❌ Dictionaries preserve keys in sorted or arbitrary order.

## 📝 Quick Revision

- A dictionary stores key-value pairs; keys must be unique and hashable (immutable).
- `d[key]` raises `KeyError` if missing; `d.get(key)` returns `None` (or a default) instead.
- `pop()`/`popitem()`/`del`/`clear()` all remove entries, but differ in what they target and return.
- `dict2 = dict1` is an alias (same object); `.copy()` makes a genuine shallow copy.
- Looping directly over a dict iterates its KEYS; use `.items()` for key-value pairs.
- `setdefault()` and `fromkeys()` are two convenient shortcuts for common patterns — but watch the mutable-default sharing gotcha.
- Hashing converts a key into a number that points directly to its storage slot — this is why lookups are O(1) on average instead of O(n).
- Dictionaries are the classic tool for frequency counting and "have I seen this before?" problems (Two Sum, anagrams, unique characters).

## ✅ Key Takeaways

1. A dictionary maps unique, hashable keys to values — `d[key]` raises `KeyError` on a miss; `d.get(key)` doesn't.
2. Only immutable types (`int`, `str`, `float`, `tuple`) can be dictionary keys, because hashing requires a value that never changes.
3. `dict2 = dict1` is an alias, not a copy — exactly like lists and tuples, use `.copy()` for a genuine (shallow) copy.
4. Hashing is what makes dictionary lookups O(1) on average, turning many O(n²) brute-force problems (like Two Sum) into O(n).
5. Frequency-counting with a dictionary (`freq[ch] = freq.get(ch, 0) + 1`) is the single most common pattern behind anagram, unique-character, and grouping problems.

````python
## Store Frequency in Dictionary

def freqDict(nums: list[int]) -> dict:
    freq_map = {}

    for i in range(0, len(nums)):
        if nums[i] in freq_map:
            freq_map[nums[i]] += 1
        else:
            freq_map[nums[i]] = 1
    return freq_map


nums = [5, 6, 7, 7, 1, 9, 12, 1, 1, 5, 1, 1, 6, 7]
print(freqDict(nums))

# Time: O(n) -- one pass over nums, and each dict lookup/insert (`in`, `[]`) is O(1) average.
# Space: O(k) -- k is the number of distinct values in nums (one entry per unique key).
````

````python
def freqDictOptimise(nums: list[int]) -> dict:

    freq_map = {}
    n = len(nums)

    for i in range(0, n):
        freq_map[nums[i]] = freq_map.get(nums[i], 0) + 1
        # freq_map[nums[0]] = freq_map.get(nums[0], 0) + 1
        #                   = freq_map.get(5, 0) + 1
        #       freq_map[5] = 0 + 1 = 1
    return freq_map

nums = [5, 6, 7, 7, 1, 9, 12, 1, 1, 5, 1, 1, 6, 7]
print(freqDict(nums))
````

## Hashing in Python
- PreStoring values into same datastructure like list/dictionary/set and then fetching it

````python
# Constraints
# 1. 1 <= num1[i] <= 10
# 2. num1 can have 10^8 elements
# 3. num2 can have 10^8 elements


def hashList(nums1: list[int], nums2: list[int]) -> dict:
    hash_list = [0] * 11   # index 0 unused; indices 1-10 hold counts -- values are bounded, so a fixed-size array works as a direct-address table (no hashing/dict needed)

    for i in nums1:
        hash_list[i] += 1

    result = {}   # dict style: map each queried value -> its count, instead of a plain positional list
    for i in nums2:
        if i < 1 or i > 10:
            result[i] = 0
        else:
            result[i] = hash_list[i]
    return result


num1 = [5, 3, 2, 2, 1, 5, 5, 7, 5, 10]
num2 = [10, 11, 1, 9, 5, 67, 2]
print(hashList(num1, num2))

# Time: O(n + m) -- one pass to build hash_list (n = len(nums1)), one pass to answer queries (m = len(nums2)).
# Space: O(1) for hash_list (always size 11); O(m) for the result dict (one entry per unique query).
````

````python
# Same problem, but with a dict + .get() instead of a fixed-size array.
# No need to bounds-check nums2 -- freq_map.get(i, 0) already returns 0 for any key that was never counted.

def hashDict(nums1: list[int], nums2: list[int]) -> dict:
    freq_map = {}

    for i in nums1:
        freq_map[i] = freq_map.get(i, 0) + 1

    result = {}   # dict style: map each queried value -> its count, instead of a plain positional list
    for i in nums2:
        result[i] = freq_map.get(i, 0)
    return result


num1 = [5, 3, 2, 2, 1, 5, 5, 7, 5, 10]
num2 = [10, 11, 1, 9, 5, 67, 2]
print(hashDict(num1, num2))

# Time: O(n + m) -- one pass to build freq_map, one pass to answer queries; dict get()/[] are O(1) average.
# Space: O(k) for freq_map (k = distinct values in nums1, at most 10 here); O(m) for the result dict.
````

````python
## character Hashing

def charHash(s: str, q: list[str]) -> dict:
    hash_list = [0] * 26   # 26 lowercase letters -- 'a' maps to index 0, 'z' maps to index 25

    for ch in s:
        index = ord(ch) - 97
        hash_list[index] += 1

    result = {}   # dict style: map each queried character -> its count, instead of a plain positional list
    for ch in q:
        index = ord(ch) - 97   # fixed: recompute from ch, not the leftover ascii_val from the s-loop
        result[ch] = hash_list[index]
    return result


s = "azyxyyzaaaa"
q = ['d', 'a', 'y', 'x']
print(charHash(s, q))

# Time: O(n + m) -- one pass to build hash_list (n = len(s)), one pass to answer queries (m = len(q)).
# Space: O(1) for hash_list (always size 26); O(m) for the result dict.
````

````python
## character Hashing -- mixed uppercase, lowercase, and symbols
#
# The array trick above only works because lowercase letters are a small, KNOWN, contiguous
# range: ord('a')..ord('z') maps cleanly to indices 0..25 via `ord(ch) - 97`.
# Uppercase and symbols break that math:
#   ord('A') - 97 = -32   -- negative index, wraps around / crashes
#   ord('!') - 97 = -64   -- same problem
# You could size an array to cover the whole ASCII range (`[0] * 128`, indexed by `ord(ch)` directly,
# no offset), but that wastes space on the huge gaps you'll never use, and still can't handle
# arbitrary unicode. A dict sidesteps all of this: it's keyed by the character itself, no index
# math, no bounds to get wrong, and it works for ANY character.

def charHashMixed(s: str, q: list[str]) -> dict:
    freq_map = {}
    for ch in s:
        freq_map[ch] = freq_map.get(ch, 0) + 1

    result = {}   # dict style: map each queried character -> its count
    for ch in q:
        result[ch] = freq_map.get(ch, 0)
    return result


s = "AzY!xy?yz#aaAA"
q = ['d', 'a', 'y', 'A', '!', 'Z']
print(charHashMixed(s, q))

# Time: O(n + m) -- one pass to build freq_map (n = len(s)), one pass to answer queries (m = len(q)).
# Space: O(k) -- k is the number of distinct characters in s (bounded by however many unique chars actually appear).
````
