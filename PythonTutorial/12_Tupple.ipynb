By the end of this note, I will understand what a tuple is, how it differs from a list, every common tuple operation (with proper examples), why tuples exist at all when lists can do more, the single-element-tuple gotcha, tuple packing/unpacking, and why tuples can be used as dictionary keys when lists can't.

---

## 🤔 What is it?

A **tuple** is an ordered collection of items, just like a list — except once created, it **cannot be changed**. No adding, removing, or reassigning items.

> 🧸 Think of a tuple like a printed receipt. Once it's printed, you can read every line on it, but you can't edit it — if you need different values, you print a brand-new receipt. A list, on the other hand, is like a whiteboard shopping list — you can cross items off and add new ones anytime.

### Core Properties

- **Ordered** — items keep the order you put them in, just like a list.
- **Immutable** — items CANNOT be changed, added, or removed after creation.
- **Allow duplicates** — the same value can appear more than once.

---

## ❓ Why do we need it?

If lists can already do everything a tuple can (and more), why would we ever choose a tuple?

- **Safety** — if data should never change (like a date of birth, or GPS coordinates), a tuple guarantees no part of your code can accidentally modify it.
- **Hashability** — tuples can be used as **dictionary keys** or stored in a **set**, because they never change. Lists can't be used this way (see the example further down).
- **Slight performance edge** — tuples are a bit faster to create and iterate over than lists, since Python doesn't need to reserve extra room for growth.

---

## 🧠 Key Idea

- Every list operation that **reads** data (indexing, slicing, `len()`, `count()`, `index()`, looping) also works on tuples.
- Every list operation that **changes** data (`append()`, `insert()`, `remove()`, `pop()`, `sort()`, `clear()`, item assignment) does NOT exist on tuples — Python will raise an error.
- A single-element tuple needs a trailing comma: `(1,)` — without it, `(1)` is just the number `1` in parentheses.
- Tuples are what make Python's famous one-line variable swap (`a, b = b, a`) possible.

| Term | Simple Meaning | Example |

## 🌍 Real-Life Example

Think of **GPS coordinates** — `(latitude, longitude)`. These two values belong together as a single fixed point; it wouldn't make sense to "append" a third value or "sort" a coordinate pair. That's exactly the kind of fixed, related data a tuple is built for — compare that to a **shopping list**, which you constantly add to, remove from, and reorder (a perfect fit for a mutable list).

---

## 💻 Technical Example: Mutable List vs Immutable Tuple

Both start out looking the same, but only the list allows changing an item after creation.

````python
# Tuple is immutable
tup = (1, 2, 3, 4, 5, 6)
print(tup)
print(type(tup))

# List is mutable
list1 = [1, 2, 3, 4, 5, 6]
list1[0] = 3
print(list1)
````

Trying the same item-assignment on a tuple raises a `TypeError` — this is the whole point of a tuple, not a bug.

````python
tup = (1, 2, 3, 4, 5, 6)
tup[1] = 4  # TypeError: 'tuple' object does not support item assignment
print(tup)
````

It's not just item assignment — tuples also have **none** of the list's mutating methods. Calling `.append()` on a tuple raises a different error: `AttributeError`, because the method doesn't exist at all.

````python
tup = (1, 2, 3)
tup.append(4)  # AttributeError: 'tuple' object has no attribute 'append'
````

# Tuple Operations

Since tuples are immutable, only **read-only** operations are supported — nothing that adds, removes, or reorders items in place.

### ⏱ Time Complexity of Tuple Operations (Worst Case)

Tuples share the same underlying array-like storage as lists, so read operations have identical complexity — there just aren't any mutating operations to worry about.

| Operation | Code | Time (Worst Case) | Why |
|-----------|------|---------------------|-----|
| Access by index | `tup[i]` | O(1) | Direct memory offset |
| Length | `len(tup)` | O(1) | Length is tracked internally |
| Slicing | `tup[a:b]` | O(b - a) | Creates a new tuple with those elements |
| Concatenation | `tup1 + tup2` | O(n + m) | Builds a brand-new tuple holding both |
| Repetition | `tup * k` | O(n × k) | Builds a new tuple repeating the contents |
| Search (`index()`, `in`) | `tup.index(x)`, `x in tup` | O(n) | Linear scan, no shortcuts |
| Count occurrences | `tup.count(x)` | O(n) | Must scan every element |
| `min()` / `max()` / `sum()` | `min(tup)` | O(n) | Must check every element once |
| `sorted(tup)` | `sorted(tup)` | O(n log n) | Returns a NEW **list** — tuples have no `.sort()` |

### `len()` and Basic Indexing

````python
tup = (1, 2, 3, 4, 5, 6)
print(len(tup))
print(tup[0])       # first item
print(tup[-1])      # last item, via negative indexing
````

### Negative Indexing

````python
number = (12, 3, 90, 2, 4, 5, 6, 7)
print(number[-2])  # second-to-last item

thistuple = ("apple", "banana", "cherry", "orange", "kiwi")
print(thistuple[-3:-1])
````

### Slicing

Works exactly like list slicing — `tup[start:stop:step]` — but always returns a NEW tuple.

````python
tup = (1, 2, 3, 4, 5, 6)
print(tup)
print(type(tup))
print(tup[1:3])
print(tup[::-1])  # reversed copy
````

### Concatenation (`+`) and Repetition (`*`)

Both create a **brand-new** tuple — the originals are never modified, since they can't be.

````python
tup1 = (1, 2, 3)
tup2 = (4, 5, 6)

print(tup1 + tup2)   # concatenation -> new tuple
print(tup1 * 3)      # repetition -> new tuple with contents repeated
````

### Membership (`in`) and Iteration

````python
tup = (10, 20, 30, 40)

print(20 in tup)
print(99 in tup)

for item in tup:
    print(item, end=" ")
print()
````

### `count()` — Count Occurrences of a Value

````python
tup = (1, 2, 5, 6, 1, 2)
print("Count of 2 in tup:", tup.count(2))
````

### `index()` — Find the Position of a Value

Just like a list, this returns the index of the **first** match, and raises `ValueError` if the value isn't found.

````python
tup = (1, 2, 5, 6, 1, 2)
print("Index of 5 in tup:", tup.index(5))
````

### `min()`, `max()`, and `sum()`

````python
tup = (12, 3, 9, 2, 4, 5, 6, 7)
print(min(tup))
print(max(tup))
print(sum(tup))
````

### Sorting a Tuple — `sorted()`, NOT `.sort()`

Tuples have **no `.sort()` method** — sorting would mean rearranging items in place, which immutability forbids. Instead, the built-in `sorted()` function returns a **new list** (not a tuple!) with the sorted values.

> ❌ Misconception: `sorted(tup)` returns a sorted tuple.
> ✅ Correct: `sorted()` ALWAYS returns a `list`, regardless of what you pass in. Wrap it in `tuple(...)` if you specifically need a tuple back.

````python
tup = (5, 2, 9, 1, 7)

result = sorted(tup)
print(result, type(result))       # a list!

result_as_tuple = tuple(sorted(tup))
print(result_as_tuple, type(result_as_tuple))
````

## ⚠ The Single-Element Tuple Gotcha

Parentheses alone don't make a tuple — Python needs to see a **comma**. `(1)` is just the integer `1` wrapped in parentheses (parentheses are also used for grouping math expressions, like `(2 + 3)`). To make a genuine one-item tuple, you need a trailing comma: `(1,)`.

````python
tp = (1)
print(tp, type(tp))          # just an int — the parentheses did nothing

tupF = (1.0)
print(tupF, type(tupF))      # just a float

tupS = ('Hello')
print(tupS, type(tupS))      # just a string

tup = (1,)
print(tup, type(tup))        # NOW it's a real tuple — the comma is what matters, not the parentheses
````

## 📦 Tuple Packing and Unpacking

**Packing** groups multiple values into one tuple. **Unpacking** does the reverse — splitting a tuple's values into separate variables in one line. This is exactly what makes Python's famous one-line swap trick work — no temporary variable needed.

````python
# The traditional way to swap two variables (needs a temp variable)
a = 22
b = 33

temp = a
a = b
b = temp
print(a, b)
````

````python
# The Pythonic way, using tuple packing + unpacking
a = "Two"
b = "One"

a, b = b, a   # the right side (b, a) is PACKED into a tuple, then UNPACKED into a, b
print(a, b)
````

````python
# Packing and unpacking with more than 2 values
point = 3, 4, 5          # packing -- parentheses are optional here
x, y, z = point           # unpacking
print(point, type(point))
print(x, y, z)
````

## 🔄 Converting Between Tuple and List

If you need to modify a tuple's contents, the standard trick is to convert it to a list, make the changes, and (optionally) convert it back.

````python
tuple1 = (3, 4, 35, 65, 34, 5)

# tuple1[0] = 4   # would raise TypeError -- tuples can't be modified directly

list1 = list(tuple1)   # convert to a mutable list
list1[0] = 90            # now we CAN modify it
print(list1)

tuple2 = tuple(list1)   # convert back to a tuple, if needed
print(tuple2)
````

## 🔑 Why Tuples Can Be Dictionary Keys (and Lists Can't)

Because tuples never change, Python can safely compute a fixed "fingerprint" (hash) for them — this is what makes tuples **hashable**. Lists are mutable, so their contents (and hash) could change after being used as a key, which would break the dictionary. Python simply refuses to let you use a list as a key at all.

> 🧸 Think of a hash like a label printed on a sealed box. Since a tuple can never be opened and refilled, the label always accurately describes what's inside. A list is like an open box you keep changing — the label would go stale immediately, so Python won't even let you print one.

````python
# Tuples work fine as dictionary keys...
locations = {
    (28.6139, 77.2090): "New Delhi",
    (40.7128, -74.0060): "New York",
}
print(locations[(28.6139, 77.2090)])

# ...and as set members
visited = {(28.6139, 77.2090), (40.7128, -74.0060)}
print((28.6139, 77.2090) in visited)
````

````python
# But a list raises an error -- lists are unhashable because they're mutable
bad_dict = {[1, 2]: "this will fail"}
````

## 📚 Tuple Methods

Tuples have exactly **two** built-in methods — no more — precisely because immutability rules out every method that would modify the tuple (`append`, `insert`, `remove`, `pop`, `sort`, `reverse`, `clear`, `extend` all simply don't exist on a tuple).

| Method | What it Does |
|--------|----------------|
| `count(x)` | Returns the number of times `x` appears in the tuple |
| `index(x)` | Returns the position of the first occurrence of `x` (raises `ValueError` if not found) |

## ⚠ Common Misconceptions

❌ Tuples support the same "clear the contents" style operations as lists (like `clear()`).
✅ Tuples have **no** mutating methods at all — `clear()`, `append()`, `remove()`, `pop()`, `sort()`, `reverse()`, and `extend()` don't exist on a tuple and raise `AttributeError` if called.

❌ `(1)` is a tuple containing the number 1.
✅ `(1)` is just the integer `1` — parentheses alone don't create a tuple. You need a trailing comma: `(1,)`.

❌ `sorted(my_tuple)` returns a sorted tuple.
✅ `sorted()` always returns a **list**, no matter what you pass in. Use `tuple(sorted(my_tuple))` if you need a tuple back.

❌ `del my_tuple[0]` removes the first item, just like with a list.
✅ You can't delete an individual item from a tuple — that's mutation. `del my_tuple` deletes the *entire* variable, not one element.

❌ Tuples can't hold mutable objects like lists.
✅ A tuple's immutability only applies to its own slots — `([1, 2], "hi")` is a valid tuple, and you can still mutate the inner list (`t[0].append(3)`). The tuple itself just can't be reassigned or resized.

## 📝 Quick Revision

- A tuple is ordered and allows duplicates, just like a list — but it's immutable.
- Every *read* operation works the same as on a list: indexing, negative indexing, slicing, `len()`, `count()`, `index()`, `min()`/`max()`/`sum()`, membership (`in`), and iteration.
- No *write* operation exists: no `append`, `insert`, `remove`, `pop`, `sort`, `reverse`, `clear`, or item assignment.
- `(1)` is just an int; `(1,)` is a real one-element tuple — the comma matters, not the parentheses.
- Tuple packing/unpacking (`a, b = b, a`) is what makes Python's one-line swap possible.
- Tuples are hashable (since they never change), so they can be dictionary keys or set members — lists cannot.
- `sorted(tup)` returns a new **list**, never a tuple.
- To modify tuple data, convert to a list first (`list(tup)`), edit it, then convert back if needed (`tuple(...)`).
