## 🤔 What is it?

A **string** is an ordered, **immutable** sequence of characters — text, essentially. Like a tuple, once created, a string's characters can't be changed in place; every "modifying" method (`.upper()`, `.replace()`, ...) actually returns a **brand-new** string.

> 🧸 Think of a string like a printed sentence on a page. You can read any letter, copy the whole sentence, or make a NEW page with a word replaced — but you can't reach into the original page and swap out one letter.

---

## ❓ Why do we need it?

Almost every program deals with text somewhere — names, messages, file paths, user input. Strings give us a rich toolbox (searching, cleaning, splitting, reformatting) instead of having to manipulate raw characters one by one.

---

## 🧠 Key Idea

- Strings are **immutable** — just like tuples, every string method returns a new string rather than modifying the original.
- Strings support the same indexing/slicing rules as lists and tuples: `s[start:stop:step]`.
- String methods generally fall into a few families: **case conversion**, **searching**, **validating** (the `is...()` methods), **cleaning** (`strip`), and **splitting/joining**.
- `in` checks for a **substring**, not a whole word — `"cat" in "concatenate"` is `True`.

---

## 📚 Important Terms

| Term | Simple Meaning | Example |
|------|----------------|----------|

## ✂️ Slicing a String

`s[start:stop:step]` — same rules as list/tuple slicing: `start` is inclusive, `stop` is exclusive, and a negative `step` reverses direction.

### Basic Slicing

````python
name = "Python for Beginners"
print(name)
print(name[5])        # a single character at index 5

print(name[0:5])      # up to (not including) index 5
print(name[:5])       # same as above -- start defaults to 0
print(name[0:])       # from index 0 to the end -- same as the whole string
print(name[7:10])
print(len(name))
````

### Slicing with Negative Indices

Negative indices count from the end, and you can mix positive and negative bounds in the same slice.

````python
name = "Python for Beginners"
print(name[0:-3])              # everything except the last 3 characters
print(name[0:len(name) - 3])   # the same thing, written explicitly

name = "Pratham"
print(name[1:])     # from index 1 to the end
print(name[1:-1])   # everything except the first and last character
print(name[-3:-1])
print(name[-4:-2])
````

### Slicing with a Step

````python
name = "Python for Beginners"
print(name[0:5:2])   # every 2nd character, within index 0-5
print(name[0::2])    # every 2nd character, across the whole string
````

### Reverse Slicing

A negative step walks backward — `[::-1]` is the classic "reverse the whole string" trick.

````python
name = "Python for Beginners"
print(name[::-1])     # the whole string, reversed
print(name[:-2])       # everything except the last 2 characters
print(name[1::-2])     # starts at index 1, walks BACKWARD 2 at a time (toward index 0)
````

## 🔠 Case Conversion

````python
name = "physics Wallah"

print(name.lower())
print(name.upper())
print(name.capitalize())   # capitalizes only the first letter, lowercases the rest
````

````python
heading = "introduction tO js"
print(heading.capitalize())  # -> "Introduction to js" -- note: only the FIRST letter is affected

mystr = "pratham is a good boy"
print(mystr.title())          # capitalizes the first letter of EVERY word
print(mystr.title().istitle())  # True -- istitle() checks if a string is ALREADY title-cased
````

## 🔍 Searching Inside a String

`find()` and `index()` do the same job — return the position of the first match — but behave differently when nothing is found.

> ❌ Misconception: `find()` and `index()` are interchangeable.
> ✅ Correct: `find()` returns `-1` when the value isn't found. `index()` raises a `ValueError` instead. Use `find()` when "not found" is a normal outcome you want to check for; use `index()` when you're confident the value exists.

````python
name = "Pratham is a Good Boy"
print(name.find("Good"))   # found -> returns the starting index
print(name.find("iss"))    # not found -> returns -1 (no error)

# name.index("iss")        # would raise ValueError: substring not found
````

````python
s = "hello world"
print(s.count("l"))
print(s.index("o"))       # index of the FIRST match
print(s.index("llo"))     # works on substrings too, not just single characters
````

`startswith()` and `endswith()` also accept optional `start`/`end` bounds, so you can check just a slice of the string without slicing it manually first.

````python
s = "hello world"
print(s.startswith("h"))
print(s.startswith("hi"))
print(s.endswith("d"))

name = "Pratham is a Good Boy"
print(name.endswith("h", 2, 5))  # checks name[2:5] == "ath" -- does THAT end with "h"? Yes.
````

### The `in` Operator Checks for a Substring — NOT a Whole Word

This is a common trip-up: `in` will match ANY occurrence, even in the middle of another word.

````python
mystr = "Pratham is a Good Boy"
print("Pratham" in mystr)   # True -- it's a substring
print("Prat" in mystr)      # ALSO True -- "Prat" is just part of "Pratham", not a separate word
print("..." in mystr)       # False -- genuinely not present
````

## ✅ Validation Methods (the `is...()` family)

These all return `True`/`False` and never modify the string — they just answer a yes/no question about its contents.

````python
alpha = "PrathamKumar"
alpha1 = "prathamkumar"
mystr = "pratham00"
mynum = "123"

print(alpha.isalnum())    # letters AND/OR digits, no spaces/symbols
print(alpha1.isalpha())   # letters ONLY
print(mystr.isalnum())    # "pratham00" -- letters + digits -> True
print(mystr.isalpha())    # contains digits -> False
print(mynum.isalnum())    # digits only -> True
print(mynum.isalpha())    # no letters at all -> False
````

````python
name = "pratham"
print(name.islower())
print(name.isupper())

mystr = "Welcometotheconsole\n"
print(mystr.isprintable())  # False -- \n (newline) is NOT considered printable

space = "  "
print(space.isspace())      # True -- only whitespace characters
````

````python
title = "Pratham"
print(title.istitle())  # True -- already title-cased (one word, capitalized)
````

## 🧹 Cleaning: `strip()`, `rstrip()`, and `replace()`

> ❌ Misconception: `strip("abc")` removes the substring `"abc"` from the ends.
> ✅ Correct: `strip(chars)` removes ANY of the individual characters in `chars`, repeatedly, from both ends — not the string as a whole. See the example below where `"abc"` strips a mix of a's, b's, and c's, not just the exact sequence `"abc"`.

````python
st = "        pratham kumar       "
hiphen = "----------pratham kumar--------"
print(repr(st.strip(" ")))       # removes leading/trailing spaces only
print(repr(hiphen.strip("-")))   # removes leading/trailing "-" characters

# strip(chars) removes ANY of those characters from the ends, not the literal substring
messy = "aabbccXaabbcc"
print(messy.strip("abc"))   # strips every a/b/c from both ends until it hits "X" -> "X"
````

````python
name2 = "Pratham !!!!!!!!"
print(name2.rstrip("!"))                      # removes trailing "!" characters only
print(name2.upper().rstrip("!"))              # methods can be chained
print(name2.replace("Pratham", "Kumar").rstrip("!"))
````

````python
mystr = "Pratham is a Good Boy"
print(mystr.upper())
print(mystr.lower())
print(mystr.replace("Good", "Bad"))
````

### 📋➡️✂️ Joining a List Back into a String

`separator.join(iterable)` puts `separator` **between** each item — NOT before or after every single one.

> ❌ Misconception: `"hello".join(list1)` puts `"hello"` before/around every item.
> ✅ Correct: it's inserted BETWEEN consecutive items only — so a 4-item list gets exactly 3 separators, not 4.

````python
list1 = ['a', 'b', 'c', 'd']
list2 = ['hello', 'my', 'name', 'is', 'pratham']

s = "hello".join(list1)   # "hello" goes BETWEEN each letter, not around all of them
st = " ".join(list2)        # the classic way to turn a word-list back into a sentence
print(s)
print(st)
````

## 🔄 Itertools

`itertools` isn't string-specific — it works on any iterable — but it's a natural next step once you're comfortable looping over sequences like strings and lists.

### `itertools.count(start, step)` — An Infinite Counter

Generates numbers forever, starting at `start` and increasing by `step` each time. Since it never stops on its own, you MUST include a `break` condition.

````python
import itertools

for i in itertools.count(5, 5):
    if i == 35:
        break
    else:
        print(i, end=" ")
````

### `itertools.cycle(iterable)` — Loop Forever Through a Sequence

Repeats the given iterable endlessly. Again, you need your own stopping condition — here we count how many items we've printed.

````python
import itertools

count = 0
for i in itertools.cycle('AB'):
    if count > 7:
        break
    else:
        print(i, end=" ")
        count += 1
````

````python
import itertools

l = ['Geeks', 'for', 'Geeks']
iterators = itertools.cycle(l)  # an iterator object -- call next() to advance it manually

for i in range(6):
    print(next(iterators), end=" ")
````

# String Practice Problems

### Defanging an IP Address | LEETCODE - 1108

**Problem:** Given a valid IPv4 address, replace every `.` with `[.]`.

**Example:** `"1.1.1.1"` → `"1[.]1[.]1[.]1"`

**Approach:** Walk through the string one character at a time, appending each character normally, except when it's a `.` — then append `"[.]"` instead. (Python's built-in `address.replace(".", "[.]")` does the exact same thing in one line — the manual version below is for practicing character-by-character string building.)

**Complexity:** O(n) time, O(n) space (for the new string).

````python
class Solution:
    def defangIPaddr(self, address: str) -> str:
        # One-liner alternative: return address.replace(".", "[.]")

        ans = ""
        for ch in address:
            if ch != ".":
                ans += ch
            else:
                ans += "[.]"
        return ans

address = "255.100.50.0"
solution = Solution()

print(solution.defangIPaddr(address))
````

### Valid Palindrome | LEETCODE - 125

**Problem:** Given a string, ignoring case and non-alphanumeric characters, check if it reads the same forward and backward.

**Example:** `"A man, a plan, a canal: Panama"` → `True`

**Approach:** This is the **converging Two Pointer** pattern (see [08_DsaPattern_part1.ipynb](08_DsaPattern_part1.ipynb)) — `i` starts at the front, `j` at the back. Skip over non-alphanumeric characters on either side; once both point at real characters, compare them and move inward.

**Complexity:** O(n) time, O(1) extra space.

````python
class Solution:
    def isAlphanumeric(self, ch):
        x = ord(ch)
        if 97 <= x <= 122 or 65 <= x <= 90 or 48 <= x <= 57:
            return True
        return False

    def palin(self, s: str) -> bool:
        s = s.lower()
        i = 0
        j = len(s) - 1

        while i < j:
            if not self.isAlphanumeric(s[i]):
                i += 1
            elif not self.isAlphanumeric(s[j]):
                j -= 1
            elif s[i] == s[j]:
                i += 1
                j -= 1
            else:
                return False

        return True


test_str = "A man, a plan, a canal: Panama"
solution = Solution()
print(solution.palin(test_str))
````

### Reverse String | LEETCODE - 344

**Problem:** Reverse an array of characters in-place, using O(1) extra memory.

**Approach:** Converging two pointers, swapping as they move inward — the exact same technique already covered for a plain list in [08_DsaPattern_part1.ipynb](08_DsaPattern_part1.ipynb), applied here to a character array.

**Complexity:** O(n) time, O(1) extra space.

````python
from typing import List

class Solution:
    def reverseString(self, s: List[str]) -> None:
        i = 0
        j = len(s) - 1

        while i < j:
            s[i], s[j] = s[j], s[i]
            i += 1
            j -= 1

s = ["h", "e", "l", "l", "o"]
print("Input :", s)
solution = Solution()
solution.reverseString(s)
print("Output:", s)
````

### Reverse Words in a String | LEETCODE - 151

**Problem:** Reverse the ORDER of the words in a sentence, collapsing any extra whitespace, with no leading/trailing spaces in the result.

**Example:** `"the sky is blue"` → `"blue is sky the"`

**Approach:** `s.split()` (with no argument) is smarter than `split(" ")` here — it automatically collapses multiple spaces and ignores leading/trailing whitespace. Then reverse the resulting word list with two pointers, and `" ".join(...)` it back together.

**Complexity:** O(n) time, O(n) space (for the word list).

````python
class Solution:
    def reverseWordString(self, s: str) -> str:
        words = s.strip().split()  # split() with no args also collapses multiple spaces

        i = 0
        j = len(words) - 1
        while i < j:
            words[i], words[j] = words[j], words[i]
            i += 1
            j -= 1

        return " ".join(words)


s = "the sky is blue"
print("Input :", s)
obj = Solution()
res = obj.reverseWordString(s)
print("Output:", res)
````

### Length of Last Word | LEETCODE - 58

**Problem:** Given a string of words and spaces, return the length of the LAST word.

**Example:** `"   fly me   to   the moon  "` → `4` (the last word is `"moon"`)

**Approach shown:** walk backward from the end, counting characters until hitting a space (or the start of the string).

**Simpler alternative:** `len(s.strip().split()[-1])` — split on whitespace, take the last word, measure its length. The manual version below is worth knowing for interviews that ask you to avoid built-in split logic.

**Complexity:** O(n) time, O(1) extra space (for the manual version).

````python
class Solution:
    def lengthOfLastWord(self, s: str) -> int:
        s = s.strip()
        n = len(s)

        i = -1
        while i >= (-1 * n) and s[i] != " ":
            i -= 1

        i += 1
        i *= -1

        return i


s = "Hello world"
print("Input: ", s)
obj = Solution()
res = obj.lengthOfLastWord(s)
print("Output:", res)

# Simpler alternative, for comparison:
print("Alternative:", len(s.strip().split()[-1]))
````

## ⚠ Common Misconceptions

❌ `strip("abc")` removes the exact substring `"abc"` from the ends.
✅ It removes ANY of the individual characters `a`, `b`, `c` — repeatedly — from both ends, not the literal sequence.

❌ `"hello".join(["a","b","c"])` wraps `"hello"` around every item.
✅ `join()` inserts the separator only BETWEEN consecutive items — a 3-item list gets exactly 2 separators.

❌ `"Prat" in "Pratham is here"` means `"Prat"` is a separate word in the sentence.
✅ `in` checks for a **substring** anywhere, including in the middle of another word — not a whole-word match.

❌ `find()` and `index()` behave the same way.
✅ `find()` returns `-1` when nothing is found; `index()` raises `ValueError` instead.

❌ String methods like `.upper()` or `.replace()` modify the string in place.
✅ Strings are immutable — every method returns a brand-new string; the original is untouched unless you reassign the variable.
