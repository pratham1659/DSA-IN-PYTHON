```python
## Largest Element in an Array

def largestElement(nums: list[int]):

    large = float("-inf") ## - infinity value

    n = len(nums)

    for i in range(0, n):
        if nums[i] > large:
            large = nums[i]
    return large


nums = [55, 32, -97, 99, 3, 67]
negativeNums = [-55, -32, -97, -99, -3, -67]
print(largestElement(negativeNums))
```

```python
## Smallest Element in an Array

def SmallestElement(nums: list[int]):

    large = float("+inf") ## + infinity value

    n = len(nums)

    for i in range(0, n):
        if nums[i] < large:
            large = nums[i]
    return large


nums = [55, 32, -97, 99, 3, 67]
negativeNums = [-55, -32, -97, -99, -3, -67]
print(SmallestElement(negativeNums))
```

```python
## Find the 2nd Largest Element in an Array without sorting

def secondLargest(nums: list[int]):
    large = float("-inf")
    s_large = float("-inf")
    n = len(nums)

    for i in range(0, n):
        if(nums[i] > large):
            s_large = large
            large = nums[i]
        elif(nums[i]>s_large and nums[i] != large):
            s_large = nums[i]
    return s_large

nums = [55, 32, -97, 99, 3, 67]
negativeNums = [-55, -32, -97, -99, -3, -67]
print(secondLargest(nums))
```

```python
## Check is the Array or List is sorted or not

def checkSorted(nums: list[int]) -> bool:
    
    n = len(nums)
    for i in range(0, n-1):
        if nums[i] > nums[i+1]:
            return False
    return True

nums = [3, 5, 6, 7, 8, 9, 20]
nums1 = [5, 7, 9, 3, 6, 8, 4, 6, 9, 4, -43]
print(checkSorted(nums1))
```

```python
## Remove Duplicated from the sorter Array in place | Leetcode - 26 | Normal Approach | Brute Force Approach

def removeDuplicate(nums: list[int]):
    n = len(nums)
    freq_map = {}
    for i in range(0, n):
        freq_map[nums[i]] = 0

    j = 0

    for k in freq_map:
        nums[j] = k
        j += 1
    return j

nums = [1,1,1,2,3,4,4,7,9,9,9,10]
print(removeDuplicate(nums))
```

```python
## Remove Duplicated from the sorter Array in place | Leetcode - 26 | Optimal Approach

def removeDuplicate(nums: list[int]):

    n = len(nums)

    if n == 1:
        return 1

    i = 0
    j = i+1

    while j<n:
        if nums[j] != nums[i]:
            i += 1
            nums[i], nums[j] = nums[j], nums[i]
        j += 1
    return i + 1

nums = [1,1,1,2,3,4,4,7,9,9,9,10]
print(removeDuplicate(nums))
```
