# Iterations and Looping in Python

## 🎯 Learning Goal

By the end of this note, I will understand what loops are and why we need them, the difference between `for` and `while` loops, how `range()` works, what `break`/`continue`/`while-else` do, and how loops are used to solve real coding problems and print patterns.

---

## 🤔 What is it?

A **loop** is a way to tell the computer "repeat this action multiple times" instead of writing the same line of code over and over.

> 🧸 Think of a loop like a washing machine on a repeat cycle. You don't manually rinse each piece of clothing one by one — you set it once, and it repeats the "wash" step for every item until it's done.

Python has two main types of loops:
- **`for` loop** — repeats a fixed number of times, or once for each item in a collection (list, string, range).
- **`while` loop** — repeats *as long as* a condition stays `True`. You don't always know in advance how many times it will run.

---

## ❓ Why do we need it?

- Without loops, printing numbers 1 to 1000 would need 1000 separate `print()` lines!
- Loops let us process collections (lists, strings) item by item without knowing their size in advance.
- Many real-world problems (searching, summing digits, building patterns, checking conditions repeatedly) are naturally repetitive — loops are how code handles repetition cleanly.

---

## 🧠 Key Idea

- `for` loops are best when you know *what* you're iterating over (a list, a string, a range of numbers).
- `while` loops are best when you're repeating *until a condition changes* (e.g., "keep asking the user until they enter a valid number").
- `break` stops a loop completely and jumps out of it.
- `continue` skips the rest of the current iteration and moves to the next one.
- `while...else` runs the `else` block only if the loop finished normally (was never `break`-ed).

---

## 📚 Important Terms

| Term | Simple Meaning | Example |
|------|----------------|----------|
| `for` loop | Repeats once for each item in a sequence | `for x in [1,2,3]:` |
| `while` loop | Repeats while a condition is `True` | `while count <= 10:` |
| `range()` | Generates a sequence of numbers | `range(1, 6)` → 1,2,3,4,5 |
| `break` | Immediately exits the loop | Stop searching once found |
| `continue` | Skips to the next iteration | Skip banana, print the rest |
| `while...else` | `else` runs only if loop wasn't broken | Confirm the loop completed fully |
| Nested loop | A loop inside another loop | Used for grids, patterns, tables |
| Infinite loop | A loop with a condition that never becomes `False` | `while True:` with no `break` |

Each time through the loop is called an **iteration**. Before every iteration, Python checks the condition (or checks if there are more items left in a `for` loop). The loop keeps going until that check fails.

Think of a **teacher taking attendance**:

- If the teacher is looking for one specific student and finds them, they stop checking the rest of the list (`break`).
- "Keep calling names *while* there are still students left on the list" is the `while` loop version of the same task.

## ⚠ What is "Time Limit Exceeded" (TLE)?

### 🤔 What is it?

When you submit code on a coding platform (LeetCode, HackerRank, Codeforces, etc.), the judge doesn't just check if your answer is correct — it also checks if your code finishes running within a **set time limit** (usually 1-2 seconds). If your loop takes too long — either because it's stuck forever, or because it's just too slow for the given input size — the judge kills it early and shows **"Time Limit Exceeded (TLE)."**

> 🧸 Think of TLE like a cooking competition. You might cook the *correct* dish, but if you take 3 hours when the limit is 30 minutes, you still fail the round — being right isn't enough, you also have to be fast enough.

### ❓ Why does it happen?

- **Infinite loops** — the loop's stopping condition never becomes `False` (classic bug: forgetting to update the loop variable).
- **Inefficient algorithms** — an O(n²) solution might finish instantly for `n = 100` but take forever for `n = 1,000,000`, while an O(n) or O(log n) solution would pass easily. See [Time and Space Complexity](06_TimeAndSpace_Complexity.ipynb).
- **Repeated expensive work inside a loop** — e.g., searching through a whole list again and again inside another loop, instead of doing it once.

### 🖼 Visual Representation

```
Buggy version:            Fixed version:
i = 0                     i = 0
while i < 10:             while i < 10:
    print(i)                  print(i)
    # i never changes             i += 1   <- prevents infinite loop
    v
i stays 0 forever
v
Judge kills it -> TLE (X)
```

### 💡 Easy Trick to Remember

> 📌 **Every loop needs a "finish line."** Before writing a `while` loop, always ask: *"What changes each time, and when does it stop?"* If you can't answer that, you're about to write an infinite loop.

### ⚠ Common Misconceptions

❌ TLE means my code's logic/answer is wrong.
✅ TLE means the code is *too slow* (or stuck) — the logic could be perfectly correct, just not efficient enough, or missing a variable update.

❌ TLE only happens with `while` loops.
✅ `for` loops can also cause TLE if they're deeply nested or process huge inputs inefficiently (e.g. an O(n³) triple loop over a huge list).

---

````python
# BUGGY VERSION (do not uncomment/run — this is an infinite loop that would TLE):
# i = 0
# while i < 10:
#     print(i)
#     # forgot to update i here — i stays 0 forever, condition never becomes False

# FIXED VERSION — the loop variable is updated every iteration, so it correctly stops
i = 0
while i < 10:
    print(i)
    i += 1   # this line is what prevents an infinite loop / TLE

# Time Complexity: buggy version = O(infinity) (never stops); fixed version = O(n), n = 10
````

## 🔁 `range()` in Loops

`range(start, stop, step)` generates numbers from `start` up to (but NOT including) `stop`, moving by `step` each time. If `step` is negative, it counts down instead of up.

| Call | Meaning | Output |
|------|---------|--------|
| `range(1, 6)` | 1 to 5 | 1 2 3 4 5 |
| `range(5, 11)` | 5 to 10 | 5 6 7 8 9 10 |
| `range(20, 0, -1)` | 20 down to 1 | 20 19 ... 1 |
| `range(20, 0, -2)` | 20 down to 2, step 2 | 20 18 ... 2 |
| `range(1, 10, 2)` | 1 to 9, step 2 (odd numbers) | 1 3 5 7 9 |

````python
for i in range(1, 6):
    print(i, end=" ")
print()

for i in range(5, 11):
    print(i, end=" ")
print()

for i in range(20, 0, -1):
    print(i, end=" ")
print()

for i in range(20, 0, -2):
    print(i, end=" ")
print()

for i in range(1, 10, 2):
    print(i, end=" ")
print()

# Time Complexity: O(n) -- 5 separate (non-nested) loops, so their costs ADD, not multiply
````

### While Loop in Python

A `while` loop repeats *as long as* its condition stays `True`. Be careful — if the condition never becomes `False`, you get an **infinite loop** (this is exactly how a "Time Limit Exceeded" error happens above).

````python
#while loop
count = int(input("Enter Number for Loop: "))
while count <= 10:
  print(count)
  count += 1
  
print("Done With The Loop")

# Time Complexity: O(n) -- n = (11 - count)
````

````python
i = int(input("Enter the number: "))
while(i<20):
    i = int(input("Enter Valid Number: "))
    print(i)

print("Done with Loop")

# Time Complexity: O(k) -- k = number of attempts until input >= 20 (unbounded)
````

### `while...else` — a Python-only trick

The `else` block after a `while` loop runs **only if the loop finished on its own** (condition became `False`) — it is SKIPPED if the loop was stopped early with `break`.

> ❌ Misconception: "else means the loop failed."
> ✅ Correct: "else means the loop completed normally, without being interrupted."

````python
count = 5
while(count > 0):
    print(count)
    count -= 1
else:
    print("I am Inside Else")

# Time Complexity: O(n) -- n = count (5 iterations)
````

````python
#break statement
fruits = ["apple", "banana", "cherry"]
for x in fruits:
  print(x)
  if x == "banana":
    break #for loops break from here

# Time Complexity: O(n) worst case, O(1) best case -- break can end the loop early
````

````python
n = 3  # Number of rows

for i in range(1, n + 1):
    for j in range(1, i + 1):
        print(j, end="")
    print()

# Time Complexity: O(n^2) -- nested loop
````

````python
# Triangle number pattern using nested loops
n = int(input("Enter number of lines: "))
for i in range(1, n + 1):
    for c in range(1, i + 1):
        print(c, end=" ")
    print()

# Time Complexity: O(n^2) -- nested loop
````

# Practice Problems in Loops and Control Statements

````python
# FIZZBUZZ Problem | LEETCODE - 412

from typing import List


def fizzbuzz(n: int) -> List[str]:
    ans = []
    for i in range(1, n + 1):
        if i % 3 == 0 and i % 5 == 0:
            print("FizzBuzz")
        elif i % 3 == 0:
            print("Fizz")
        elif i % 5 == 0:
            print("Buzz")
        else:
            ans.append(str(i))
    return ans
            
                   
print(fizzbuzz(10), end=" ") 

# Time Complexity: O(n) time, O(n) space
````

````python
## Count Odd Numbers in an Interval Range | LEETCODE - 1523
low = 3
high = 7

def countOdds(low: int, high: int) -> int:
    return ((high + 1) // 2) - (low // 2)


print(countOdds(3, 7))

# Time Complexity: O(1) -- pure arithmetic, no loop
````

````python
## How Many Numbers Are Smaller Than the Current Number | LEETCODE - 1365

from typing import List

def smallerNumberCount(nums: List[int]) -> List[int]:
    ans = []
    for i in nums:
        count = 0
        for j in nums:
            if j < i:
                count += 1
        ans.append(count)
    return ans
    
    
nums = [8, 1, 2, 2, 3]    
print(smallerNumberCount(nums))

# Time Complexity: O(n^2) time, O(n) space -- nested loop
````

````python
## Find the count of number using while loop

num = 5183
count = 0
sum = 0
while num>0:
    r = num%10
    num //= 10
    sum = r + sum
    count+=1
    

print("Count of Digits", count)
print("Sum of digits", sum)

# Time Complexity: O(log n) -- iterations = number of digits in num
````

````python
## Count the Digits That Divide a Number | LEETCODE - 2520

def countDigits(num: int) -> int:
    temp = num
    ans = 0
    while temp > 0:
        r = temp % 10
        if num % r == 0:
            ans+=1
        temp //= 10
    return ans
        
num = 121 
print(countDigits(num))

# Time Complexity: O(log n) -- one iteration per digit of num
````

````python
## Palindrome Number | LEETCODE - 9

def checkPalindrome(num: int) -> bool:
    ans = 0
    temp = num
    while temp > 0:
        r = temp % 10
        ans = ans * 10 + r
        temp //=10
    
    return num == ans
        

num = 212
print(checkPalindrome(num))

# Time Complexity: O(log n) -- one iteration per digit of num
````

````python
## Subtract the Product and Sum of Digits of an Integer | LEETCODE - 1281

def subtractProductAndSum(num: int) -> int:
    temp = num
    prod = 1
    add = 0
    
    while temp > 0:
        r = temp % 10
        temp //= 10
        add+= r
        prod*= r
    
    return prod - add
        
    
num = 234
print(subtractProductAndSum(num))

# Time Complexity: O(log n) -- one iteration per digit of num
````

````python
## Kids With the Greatest Number of Candies | LEETCODE - 1431
from typing import List

def kidsWithCandies(candies: List[int], extraCandies: int) -> int:
    maxCandies = max(candies)
    ans = []
    
    for i in candies:
        if (i+extraCandies) >= maxCandies:
            ans.append(True)
        else:
            ans.append(False) 
    return ans   

candies = [2,3,5,1,3]
extraCandies = 3   
print(kidsWithCandies(candies, extraCandies))

# Time Complexity: O(n) time, O(n) space
````

# 🖼 Pattern Problems (Nested Loops in Action)

Star/number patterns are a classic way to practice nested loops: the **outer loop** controls the row, and the **inner loop** controls what gets printed in that row.

````python
n = 3  # Number of rows

for i in range(1, n + 1):
    spaces = n - i
    stars = i
    print(" " * spaces + "*" * stars)

# Time Complexity: O(n^2) -- string built per row grows with n
````

## ⚖ Comparison: `for` vs `while`

| `for` loop | `while` loop |
|-----------|--------------|
| Use when you know the number of iterations (or have a sequence to walk through) | Use when you repeat until a condition changes, and don't know the count upfront |
| `for i in range(5):` | `while count <= 10:` |
| Automatically stops when the sequence ends | You must manually update the condition, or risk an infinite loop |

## ⚠ Common Misconceptions

❌ `range(1, 9)` includes 9.
✅ `range(start, stop)` is exclusive of `stop` — `range(1, 9)` goes 1 to 8.

❌ `continue` exits the loop.
✅ `continue` only skips the *current* iteration — the loop keeps going.

❌ The `else` after a `while` loop means "if the condition was False from the start."
✅ It means "the loop finished without hitting a `break`" — it can run even if the loop body executed many times.

❌ Nested loops are just "two loops" — same cost as one loop.
✅ Nested loops multiply the number of operations (n × n), which is why patterns and grids can get slow for large `n` (see [Time and Space Complexity](06_TimeAndSpace_Complexity.ipynb)).

## 📝 Quick Revision

- A loop repeats a block of code multiple times instead of writing it out manually.
- `for` loops iterate over a known sequence; `while` loops repeat until a condition becomes false.
- `range(start, stop, step)` generates numbers — `stop` is always excluded.
- `break` exits the loop entirely; `continue` skips to the next iteration.
- `while...else` runs its `else` only if the loop was never `break`-ed.
- Nested loops run the inner loop fully for every single iteration of the outer loop — this is why they multiply, not add, complexity.
- Forgetting to update a `while` loop's condition variable causes an infinite loop → Time Limit Exceeded on judges.
- Patterns (triangles, pyramids) are a great way to practice nested loop logic.

## Extraction of Digits
- Count Digits
- Reverse a number
- check palindrome
- armstrong number

````python
def countDigits(num: int) -> int:

    count = 0
    while(num > 0):
        count += 1
        num //= 10
    return count

nums = 58737
print(countDigits(nums))

# Time Complexity: O(log n) -- one iteration per digit of num
````

````python
def palindromeNumber(num: int) -> bool:

    real_num = num
    result = 0
    while(num > 0):
        last_digit = num % 10
        result = (result * 10) + last_digit
        num //= 10
    return result == real_num


nums = 1221
print(palindromeNumber(nums))

# Time Complexity: O(log n) -- one iteration per digit of num
````

````python
def armStrong(num: int) -> bool:
    n = num
    total = 0
    nod = len(str(n))

    while num > 0:
        digit = num % 10
        total = total + (digit ** nod)
        num //= 10
    return total == n

nums = 153
print(armStrong(nums))

# Time Complexity: O(log n) -- one iteration per digit of num
````

````python
## print the factors and divisors for a given number
# 10 -> {1, 2, 5, 10}
# 15 -> {1, 3, 5, 15}
# 20 -> {1, 2, 4, 5, 10, 20}
# 25 -> {1, 5, 25}
# 7 -> {1, 7}

from typing import List

def factorNum(num: int) -> List[int]:
    result = []

    for i in range(1, num // 2 + 1):   # fixed: "+1" so num//2 itself isn't skipped (was missing 50 for num=100)
        if num % i == 0:
            result.append(i)
    result.append(num)
    return result

nums = 100
print(factorNum(nums))

# Time Complexity: O(n) -- checks every candidate divisor up to n/2 (n = num)
````

````python
## Optimal Solution
# Checks only up to the square root of num -- every larger divisor is found as the PAIR (num // i).

from math import sqrt
from typing import List

def factorNumOptimise(num: int) -> List[int]:
    result = []

    for i in range(1, int(sqrt(num)) + 1):
        if num % i == 0:
            result.append(i)
            if num // i != i:
                result.append(num // i)
    return result


nums = 100
print(factorNumOptimise(nums))

# Time Complexity: much faster than O(n) above -- iterations grow with sqrt(num), not num itself
````
