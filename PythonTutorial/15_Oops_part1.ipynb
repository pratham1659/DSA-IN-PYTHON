## 🤔 What is a Class? (Real-life analogy)

Think of a **cookie cutter** 🍪. The cutter itself is not a cookie — it's just the *shape*, the blueprint. Every time you press it into dough, you get a new cookie (an **object**). All cookies made from the same cutter share the same shape (the class's attributes and methods), but each individual cookie is its own thing — you can decorate one with chocolate and another with sprinkles without affecting the rest.

- **Class** = the cookie cutter (blueprint, defined once)
- **Object** = an actual cookie (an instance, created many times)
- **`self`** = "this specific cookie" — inside a method, `self` refers to the one object the method is currently working on.

<a id="key-idea"></a>
### 🧠 Key Idea

- A **class** is defined with the `class` keyword; it groups related data (**attributes**) and behavior (**methods**) together.
- `__init__` is the **constructor** — Python calls it automatically every time you create a new object, to set up that object's starting attributes.
- **Class attributes** (defined directly under the class, outside `__init__`) are shared by every object of that class. **Instance attributes** (set via `self.x = ...` inside `__init__`) belong to one specific object.
- **Abstraction** hides *how* something works and exposes only *what* it does (e.g. `car.start()` — you don't need to know about pistons and fuel injection).
- **Encapsulation** bundles data + the methods that operate on it into one unit (the class/object), so related logic lives in one place instead of being scattered.

## 1. 🧱 Classes and Objects

- A **class** is a blueprint for creating objects.
- An **object** is a specific instance created from that blueprint, with its own data.

```python
class Students:
    college_name = "ABC University"
    name = "Karan" # class Attribute
    def __init__(self, name, marks):
       self.name = name # Object Attribute
       self.marks = marks
       print("Adding in Databases")
       
    def welcome(self):
        print("welcome student,", self.name)
        
    def get_marks(self):
        return self.marks
        
    
s1 = Students("Pratham", 234)
print(s1.name, s1.marks)

s2 = Students("Rahul", 345)
print(s1.name, s1.marks)

print(s2.college_name)

s1.welcome()

print(s1.get_marks())
```

**Note:** `college_name` and `name = "Karan"` are **class attributes** — defined directly under `class Students:`, outside `__init__`. `s2.college_name` works even though we never set it on `s2` — Python looks it up on the class when it can't find it on the instance.

**Gotcha:** The class attribute `name = "Karan"` is immediately shadowed for every object, because `__init__` always sets `self.name = name` (an *instance* attribute) with the same name. So `Students.name` is `"Karan"`, but `s1.name` and `s2.name` are whatever was passed in — instance attributes take priority over class attributes of the same name.

### Let's Practice - Class Objects and Constructors

**Why `self` here:** `set_details` and `display_details` both need `self` as their first parameter so Python knows **which** `Student` object to read from or write to. Without it, `self.roll_no = roll_no` would have no object to attach `roll_no` to — `self` is what turns a plain local variable into a piece of data that sticks to one specific object.

**Why this approach is used:** this "setter" style shows that attributes *can* be added after an object already exists, not only inside `__init__` — handy when the real data (e.g. from a form, or a file) only becomes available later. The trade-off is that nothing forces `set_details()` to actually be called first — see Example 2 for exactly what breaks when it isn't.

```python
## Example 1: User have to set the detials like intialise the object data before calling

class Student:

    def set_details(self, roll_no: int, name: str, gender: str, age: int) -> None:
        self.roll_no = roll_no
        self.name = name
        self.gender = gender
        self.age = age

    def display_details(self) -> None:
        print(f"Roll No: {self.roll_no}")
        print(f"Name: {self.name}")
        print(f"Gender: {self.gender}")
        print(f"Age: {self.age}")


student1 = Student()

student1.set_details(1, "Pratham", "Male", 99)
student1.display_details()
```

```python
## Example 2: what happens if set_details() forgets an attribute

class Student:

    def set_details(self, roll_no: int, name: str, gender: str, age: int) -> None:
        self.age = age   # bug: only age is ever saved -- roll_no, name, and gender are silently dropped

    def display_details(self) -> None:
        # print(f"Roll No: {self.roll_no}")   # AttributeError -- self.roll_no was never set
        # print(f"Name: {self.name}") # AttributeError -- self.roll_no was never set
        # print(f"Gender: {self.gender}"). # AttributeError -- self.roll_no was never set
        print(f"Age: {self.age}")


student1 = Student()

student1.set_details(1, "Pratham", "Male", 99)
student1.display_details()
```

**What happened:** `set_details()` was called with all 4 arguments (`roll_no`, `name`, `gender`, `age`), but its body only saved `self.age = age` — the other three parameters were received and then silently ignored. When `display_details()` later tries to read `self.roll_no`, Python raises `AttributeError: 'Student' object has no attribute 'roll_no'`, because that attribute was never actually created on this object.

> ❌ Misconception: passing an argument to a method automatically creates a matching attribute.
> ✅ Correct: a parameter is just a local variable inside that method call — it only becomes a real, reusable attribute if you explicitly write `self.something = something`. This is exactly why the setter pattern from Example 1 is riskier than `__init__`: there's no compiler check forcing every attribute to be set, so a typo or a forgotten line creates a half-built object that only fails later, when something tries to read the missing piece.

### 🧷 Instance Variables vs Class Variables

We already saw a hint of this earlier — `college_name` on `Students` was a **class variable**, and it stayed accessible on `s2` even though we never set it there directly. Let's go deeper and see exactly how the two interact (and where it gets tricky).

**Instance Variable**
- Belongs to ONE specific object.
- Usually created with `self.x = ...` inside `__init__` (or any method).
- Every object keeps its own separate copy — changing one object's copy never affects another's.

**Class Variable**
- Belongs to the **class itself**, not to any one object.
- Defined directly inside the class body, OUTSIDE any method.
- Shared by every object of that class — perfect for values that should be the same everywhere (a company name, a fixed tax rate) or a running total (e.g. "how many objects have been created so far").

| | Instance Variable | Class Variable |
|---|---|---|
| Belongs to | One object | The class (shared by all objects) |
| Defined | `self.x = ...`, usually in `__init__` | Directly under `class X:`, outside any method |
| Changing it via one object | Only affects that object | ⚠️ Depends how you change it — see below |
| Typical use | Data that's different per object (`name`, `marks`) | Data shared by all objects (`company_name`, a counter) |

```python
## Example: Instance Variable and Class Variable

class Employee:
    company_name = "TechNova"   # class variable -- shared by every Employee
    total_employees = 0          # class variable used as a running counter

    def __init__(self, name, salary):
        self.name = name          # instance variable -- unique per Employee
        self.salary = salary      # instance variable -- unique per Employee
        Employee.total_employees += 1   # update the class variable via the CLASS

    def display_details(self):
        print(f"Name: {self.name}, Salary: {self.salary}, Company: {self.company_name}")


emp1 = Employee("Asha", 50000)
emp2 = Employee("Ravi", 60000)

emp1.display_details()
emp2.display_details()

print("Total Employees:", Employee.total_employees)   # 2 -- counts how many Employees exist so far
```

**Note:** `name` and `salary` are different for every `Employee` (instance variables), but `company_name` is identical for both — because it's ONE shared class variable, not a separate copy per object. `display_details()` reads `self.company_name` just like any other attribute — it doesn't need to know or care whether that value came from the instance or the class. `Employee.total_employees` increments inside `__init__`, so it works as an automatic "how many objects have I created?" counter — a classic, genuinely useful pattern for class variables.

### ⚠️ The Gotcha: Changing a Class Variable Through an Instance

This is the part that trips almost everyone up at least once.

```python
emp1.company_name = "StartUp Inc"   # looks like we're updating the class variable... but we're NOT

emp1.display_details()   # Company: StartUp Inc -- emp1 now has its OWN instance variable
emp2.display_details()   # Company: TechNova    -- emp2 is untouched, still using the class variable
print("Employee.company_name:", Employee.company_name)   # TechNova -- the CLASS variable itself never changed at all
```

> ❌ Misconception: `emp1.company_name = "StartUp Inc"` updates the shared class variable.
> ✅ Correct: assigning to `obj.x` ALWAYS creates (or overwrites) an **instance** variable on that specific object — it never reaches into the class. `emp1` now has its own private `company_name` that permanently shadows the class variable of the same name, exactly like the `name`/`college_name` shadowing we saw earlier with `Students`.

**So how do you actually change it for everyone?** Update it through the CLASS, not through any one object:

```python
Employee.company_name = "Global Corp"   # update it on the CLASS itself -- this is how you change it for EVERYONE

emp2.display_details()   # Company: Global Corp -- emp2 has no instance variable, so it falls back to the (now-changed) class
emp1.display_details()   # Company: StartUp Inc -- emp1 still has its OWN instance variable, unaffected by the class change
```

**Rule of thumb:** to change a class variable for **every** object, update it via `ClassName.variable = ...`. Updating it via `some_object.variable = ...` only ever affects that one object, by shadowing — the class variable underneath is left completely alone.

### 🏗️ Constructor (`__init__`)

The setter pattern from Examples 1 and 2 has a real weakness: Python never forces you to call `set_details()` before `display_details()` — you can create a `Student` and immediately try to use it while it's still empty. The fix is to move that same logic into `__init__`, so the object is **never allowed to exist half-built** in the first place.

**Why this is the optimization:**
- `__init__` runs **automatically** the moment you write `Student(1, "Pratham", "Male", 99)` — there's no separate step to forget.
- If you leave out a required argument, Python raises `TypeError` immediately, at CREATION time — not an `AttributeError` later, somewhere else in your code, when it's harder to trace back to the real cause.
- Fewer lines overall: no separate `set_details()` method needed at all.

```python
## Example 3: the same Student, refactored to use a constructor

class Student:

    def __init__(self, roll_no: int, name: str, gender: str, age: int) -> None:
        self.roll_no = roll_no
        self.name = name
        self.gender = gender
        self.age = age

    def display_details(self) -> None:
        print(f"Roll No: {self.roll_no}")
        print(f"Name: {self.name}")
        print(f"Gender: {self.gender}")
        print(f"Age: {self.age}")


# One line instead of two -- the object is fully built the instant it's created
student1 = Student(1, "Pratham", "Male", 99)
student1.display_details()

# Try creating one without all the required details -- Python catches it immediately:
try:
    broken_student = Student(2, "Rahul")   # missing gender and age
except TypeError as e:
    print(f"TypeError caught right away: {e}")
```

**Why `self` here:** exactly the same reason as Example 1 — `self.roll_no = roll_no` inside `__init__` needs to know *which* new `Student` object it's attaching `roll_no` to. Without `self`, `roll_no` would just be a local variable that vanishes the instant `__init__` finishes running, instead of becoming a permanent, reusable piece of that object.

**Why this one is used:** this constructor version is the idiomatic, preferred way to build a `Student` — it guarantees the object is fully filled in during creation, in a single step, instead of relying on a separate call that's easy to forget (as Example 2 demonstrated). That's why `__init__` is almost always favored over a manual setter for data every object *must* have.

**Note:** Compare this `TypeError: __init__() missing 2 required positional arguments: 'gender' and 'age'` to Example 2's `AttributeError`. The constructor version fails **immediately and clearly**, at the exact line where the mistake happened. The setter version failed **later**, somewhere else entirely (`display_details()`), which is much harder to debug in a larger program — this is the core reason `__init__` is almost always preferred over a manual setter for required data.

```python
## Example 3:

class Student:

    def __init__(self, name: str, age: int, marks: list[int]) -> None:
        self.name: str = name
        self.age: int = age
        self.marks: list[int] = marks

    def total(self):
        return sum(self.marks)

    def average(self):
        # return sum(self.marks) / len(self.marks)
        return self.total() / len(self.marks)

    def grade(self):
        # avg = sum(self.marks) / len(self.marks)
        avg = self.average()
        if avg >= 90:
            print("Grade is A")
        elif avg >= 50 and avg <= 89:
            print("Grade is B")
        else:
            print("Grade is C")

student1 = Student("Pratham", 67, [78, 89,54, 19, 63])

total = student1.total()
print(f"Total Marks: {total}")

avg = student1.average()
print(f"Total Average of Student Marks: {avg}")

student1.grade()

```

```python
## Lets Practice

class Student:
    
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
        
    def get_avg(self):
        total = 0
        for val in self.marks:
            total += val
            
        print("Hi", self.name, "You avg score is: ", total/len(self.marks))
        
    @staticmethod ## without self this is used wth the def methods
    def hello_print():
        print("Hello Pratham")
        
        
s1 = Student("Tony Stark", [33, 45, 67])

s1.get_avg()

s1.name = "Iron Man"

s1.get_avg()


s1.hello_print()
```

It happened to look correct here because this example has exactly 3 marks — but it would silently give the wrong average for any student with a different number of marks (e.g. `[10, 20, 30, 40]` would wrongly report `33.33` instead of the correct `25.0`). Also renamed the loop variable from `sum` to `total`, since `sum` shadows Python's built-in `sum()` function.

**Note:** `@staticmethod` marks `hello_print` as **not needing `self`** — it doesn't touch any object's data, it's just a function that happens to live inside the class for organizational purposes. You call it the same way (`s1.hello_print()`), but it receives no automatic first argument.

### Each Object Has Its Own Independent State

A common beginner worry: "if I change one object's attribute, does it affect the others?" Let's prove it doesn't — every object gets its own separate copy of instance attributes, even though they're built from the same class.

```python
class Book:
    library_name = "City Central Library"   # class attribute -- shared by every Book

    def __init__(self, title, pages):
        self.title = title                   # instance attribute -- unique per Book
        self.pages = pages

book1 = Book("Python 101", 250)
book2 = Book("Data Structures", 400)

print(book1.title, book1.pages)
print(book2.title, book2.pages)

# Changing book1's pages does NOT affect book2 -- they're separate objects
book1.pages = 260
print(book1.pages, book2.pages)

# But both still share the SAME class attribute
print(book1.library_name, book2.library_name)
```

### Making `print(object)` Readable with `__str__`

By default, printing an object gives you an unhelpful memory address like `<__main__.Book object at 0x...>`. Defining a `__str__` method lets you control exactly what shows up instead.

```python
class BookV2:
    def __init__(self, title, pages):
        self.title = title
        self.pages = pages

    def __str__(self):
        return f"'{self.title}' ({self.pages} pages)"

plain = Book("Python 101", 250)
readable = BookV2("Python 101", 250)

print(plain)      # ugly default -- no __str__ defined
print(readable)   # uses our __str__ automatically
```

### `@classmethod` — An Alternative Constructor

`@classmethod` receives the class itself (conventionally named `cls`, just like `self` refers to the object) instead of an instance. The most common real-world use is building an **alternative constructor** — a second way to create an object, parsed from a different input shape.

```python
class BookV3:
    def __init__(self, title, pages):
        self.title = title
        self.pages = pages

    def __str__(self):
        return f"'{self.title}' ({self.pages} pages)"

    @classmethod
    def from_string(cls, book_str):
        # Parses "Title-Pages", e.g. "Python 101-250", into a new BookV3
        title, pages = book_str.split("-")
        return cls(title, int(pages))   # cls(...) is the same as calling BookV3(...)

# Normal way to create a book
b1 = BookV3("Python 101", 250)
print(b1)

# Alternative way, via the classmethod -- useful when your data arrives as a single string
b2 = BookV3.from_string("Data Structures-400")
print(b2)
```

## 2. 🎭 Abstraction

- Hiding the implementation details of a class and only showing the essential features to the user.

**Real-life analogy:** Driving a car 🚗 — you press the accelerator and the car moves. You don't need to know about fuel injection, spark plugs, or the combustion cycle. The car's `start()` and `stop()` **abstract away** all that internal complexity.

```python
class Car:
    
    def __init__(self):
        self.acc = False
        self.brk = False
        self.clutch = True
    
    def start(self):
        self.clutch = False
        self.acc = True
        print("Car Started...")
        
    
    def stop(self):
        self.brk = True
        self.clutch = True
        print("Car Stopped...")
    
toyata = Car()

toyata.start()
toyata.stop()
    
```

**Note:** From outside the class, you only ever call `toyata.start()` / `toyata.stop()` — you never need to touch `self.acc`, `self.brk`, or `self.clutch` directly. That's abstraction: the internal flags exist, but the *interface* you interact with is just two simple methods.

### 🌍 More Real-Life Abstraction Examples

- **TV remote** — pressing "Volume Up" hides the actual circuitry that changes the speaker signal.
- **ATM machine** — you select "Withdraw" and enter an amount; you never see the internal ledger updates, bank server calls, or cash-dispensing motor logic.
- **Coffee machine** — one "Brew" button hides grinding, heating water to the right temperature, and timing the extraction.
- **Online payment** — you click "Pay Now"; you don't see (or need to see) how the payment gateway talks to your bank.

In each case, the *user* only interacts with a small, simple interface — everything complicated happens behind it, invisibly.

### 🆚 How Python Treats Abstraction Differently

This is a common interview question, and the honest answer is: **Python doesn't enforce abstraction at the language level the way Java, C++, or C# do.**

| | Java / C++ / C# | Python |
|---|---|---|
| Abstract classes | A dedicated `abstract`/`virtual` keyword, enforced by the **compiler** | No dedicated keyword — achieved via the `abc` module (`ABC`, `@abstractmethod`), enforced at **runtime** |
| Interfaces | A separate `interface` construct; a class *must* implement every method to compile | No true interfaces — Python relies on **duck typing**: "if it has the right methods, it works," regardless of declared type |
| Enforcement timing | Caught at **compile time** — the program won't even build | Caught at **object-creation time** (when you try to instantiate), since Python has no compile step |

> 🧸 **Duck typing**, Python's real philosophy: *"If it walks like a duck and quacks like a duck, it's a duck."* Python doesn't care whether an object formally implements some interface — it only cares whether the object has the methods you're about to call. This is far more flexible than Java's strict interface contracts, but it also means Python catches "you forgot to implement this method" mistakes much later — only when that code path actually runs.

The `abc` module is Python's OPT-IN way to get closer to Java-style enforcement, when you actually want it:

```python
from abc import ABC, abstractmethod

class PaymentMethod(ABC):        # ABC = "Abstract Base Class" -- can't be instantiated directly
    @abstractmethod
    def pay(self, amount):        # every subclass MUST override this, or it can't be instantiated either
        pass

class CreditCardPayment(PaymentMethod):
    def pay(self, amount):
        print(f"Paid {amount} using Credit Card")

class UpiPayment(PaymentMethod):
    def pay(self, amount):
        print(f"Paid {amount} using UPI")

cc = CreditCardPayment()
cc.pay(500)

upi = UpiPayment()
upi.pay(300)

# Trying to create the abstract class directly fails immediately -- Python enforces this at runtime:
try:
    pm = PaymentMethod()
except TypeError as e:
    print(f"TypeError: {e}")
```

**Note:** `PaymentMethod` defines the essential interface (`pay()`) without saying HOW payment happens — that's abstraction. `CreditCardPayment` and `UpiPayment` each hide their own implementation details behind that same `pay()` method. And because `PaymentMethod` inherits from `ABC` and has an `@abstractmethod`, Python actively **stops** you from creating a bare `PaymentMethod()` — this is the closest Python gets to Java's compiler-enforced abstract classes, except the check happens when the object is created, not when the file is "compiled" (Python has no separate compile step).

## 3. 🔐 Encapsulation

- Wrapping data and functions into a single unit (an object).

**Real-life analogy:** A capsule pill 💊 — the medicine (data) is sealed inside a shell (the class), and you interact with it only through the shell, not by touching the raw powder directly.

### Let's practice
- Create an `Accounts` class with 2 attributes — `balance` and `account_no`.
- Create methods for `debit`, `credit`, and printing the balance.

```python
class Accounts:
    
    def __init__(self, bal, acc):
        self.bal = bal
        self.account_no = acc
        
    def debit_method(self, amt):
        self.bal -= amt
        print("Rs. ", amt, "was debited")   
        print("Total Balance = ", self.get_balance()) 
        
    def credit_method(self, amt):    
        self.bal += amt
        print("Rs. ", amt, "was credited") 
        print("Total Balance = ", self.get_balance())    
        
    def get_balance(self):
        return self.bal 
    
    
    
    
acc1 = Accounts(10000, "Pratham")
acc1.debit_method(1250)
acc1.credit_method(250)
acc1.credit_method(40000)
acc1.debit_method(250)  
```

**Note:** `debit_method` and `credit_method` don't just mutate `self.bal` — they also call `self.get_balance()` to print the updated total. This is encapsulation in action: the balance-changing logic and the balance-reading logic both live inside the same class, so nothing outside `Accounts` needs to know how the balance is stored internally.

**Gotcha:** `debit_method` doesn't check whether `amt` is larger than the current balance — it will happily let `self.bal` go negative. A real bank account class would validate this before debiting (see the "Interview Questions" section below for this exact question).

### 🌍 More Real-Life Encapsulation Examples

- **ATM PIN** — your PIN and account balance are sealed inside the bank's system; you interact only through approved actions (withdraw, check balance), never by directly editing the ledger.
- **Car dashboard** — the wiring, sensors, and engine control unit are bundled away inside the car; you only see clean gauges and warning lights.
- **Medicine capsule** 💊 (already covered above) — the shell bundles the actual medicine and controls exactly how/when it's released into your body.
- **A sealed water bottle** — you interact with it via the cap (open/close), never by reaching inside to touch the water directly.

### 🆚 How Python Treats Encapsulation Differently

Encapsulation is where Python's philosophy diverges from Java/C++ the most.

| | Java / C++ / C# | Python |
|---|---|---|
| Access control | `private`, `protected`, `public` keywords — the **compiler enforces** these; outside code truly cannot access a `private` field | No enforced keywords at all — only **naming conventions**, which are advisory, not blocked |
| "Protected" | `protected` keyword — compiler-checked | `_variable` (single leading underscore) — a hint meaning "internal use, please don't touch," but nothing stops you |
| "Private" | `private` keyword — compiler-checked, truly inaccessible from outside | `__variable` (double leading underscore) — triggers **name mangling** (renamed to `_ClassName__variable` internally), which makes ACCIDENTAL access unlikely, but it's still reachable if you know the mangled name |

> 🧸 Python's philosophy is often summarized as **"we're all consenting adults here."** Instead of the language *forcing* you to respect boundaries (like Java's compiler does), Python *trusts* you to respect the `_`/`__` naming conventions as a signal from the original author: "this is internal — read it if you must, but don't rely on it staying the same." This is a genuine, real difference in language philosophy, not just syntax.

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner            # public -- freely accessible, no protection at all
        self._pin = "1234"             # protected BY CONVENTION -- the "_" is a polite request, not a lock
        self.__balance = balance       # "private" -- gets name-mangled to _BankAccount__balance internally

    @property
    def balance(self):
        return self.__balance          # controlled READ access -- looks like a plain attribute from outside

    def deposit(self, amount):
        if amount <= 0:
            print("Deposit amount must be positive")
            return
        self.__balance += amount


acc = BankAccount("Pratham", 1000)

print(acc.owner)      # public -- works fine, no restrictions
print(acc._pin)        # protected -- ALSO works! Python never actually blocks this, it's just a signal
print(acc.balance)     # accessed via the @property getter -- clean syntax, but actually a controlled method call

acc.deposit(500)
print(acc.balance)
```

Now let's prove `__balance` isn't *truly* private — just harder to reach by accident.

```python
# The plain name doesn't exist from outside the class -- Python renamed it during name mangling
try:
    print(acc.__balance)
except AttributeError as e:
    print(f"AttributeError: {e}")

# But the MANGLED name still works -- proving __balance was never truly, unbreakably private
print(acc._BankAccount__balance)
```

**Note:** `acc.__balance` fails with `AttributeError` because Python silently renamed the attribute to `_BankAccount__balance` the moment it was written as `self.__balance` inside the class. It's NOT gone — `acc._BankAccount__balance` reaches it just fine. Compare this to Java's `private`, where the field is genuinely, permanently inaccessible from outside the class — no matter what name you try. This is the real, practical difference: Python's "private" is a **speed bump**, Java's `private` is a **wall**.

<a id="del-keyword"></a>
## 4. 🗑️ The `del` Keyword

- `del` deletes an object's properties, or the object reference itself, from memory.

```python
class Student:
    
    def __init__(self, name):
        self.name = name
        

s1 = Student("Pratham")
print(s1.name)

del s1

print(s1)

```

**Note:** After `del s1`, the name `s1` no longer exists in this scope, so `print(s1)` correctly raises `NameError: name 's1' is not defined`. This is *expected, intentional* behavior for this cell — it's demonstrating exactly what `del` does, not a bug.

**Gotcha:** `del s1` removes the *reference* `s1`, not necessarily the object itself. If another variable also pointed to the same `Student` object (e.g. `s2 = s1` was run before `del s1`), the object would still be alive and accessible through `s2` — Python only destroys the object once *no* references to it remain.

<a id="misconceptions"></a>
## ⚠ Common Misconceptions

❌ Class attributes and instance attributes are the same thing.
✅ Class attributes are shared across all objects (defined at class level); instance attributes are set per-object inside `__init__` via `self.x = ...`. An instance attribute with the same name always **shadows** the class attribute for that object.

❌ `self` is a special Python keyword.
✅ `self` is just a **convention** — you could name it anything — but it must always be the first parameter of an instance method, and Python passes the calling object into it automatically.

❌ Abstraction and Encapsulation are the same concept.
✅ Encapsulation is about **bundling** data + methods together (and optionally restricting direct access). Abstraction is about **hiding complexity** and exposing only what's necessary — encapsulation is often the *mechanism* used to achieve abstraction.

❌ `del obj` guarantees the object is immediately destroyed.
✅ It only removes that one reference. The object is only garbage-collected once its reference count drops to zero.

<a id="key-takeaways"></a>
## 🎯 Key Takeaways

1. A class is a blueprint; an object is a concrete instance built from it — each object gets its own copy of instance attributes.
2. `self` refers to "this object" inside a method; Python passes it in automatically when you call `obj.method()`.
3. Instance attributes (set via `self.x = ...`) shadow class attributes of the same name.
4. Abstraction hides *how* something works; encapsulation bundles data + behavior into one unit — they work together, not as competing ideas.
5. `@staticmethod` methods don't need `self`; `@classmethod` methods receive the class itself (`cls`) and are often used as alternative constructors.
6. `del` removes a reference, not necessarily the underlying object — and using a deleted name afterward raises `NameError`.
7. Always double-check hardcoded numbers (like a fixed divisor) inside methods that should scale with the input — they're an easy source of silent bugs.
