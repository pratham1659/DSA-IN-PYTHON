## 🤔 What is a Function?

A **function** is a named, reusable block of code that does one job — you write it once and call it as many times as you want, with different inputs.

> 🧑‍🎓 **Analogy:** A function is like a recipe card — write it once, use it every time you want that dish. The ingredients you hand over (arguments) can change each time, but the steps (the function body) stay the same.

**Why do we need functions?**
- To avoid writing the same code over and over (define once, call many times)
- To break a big program into small, testable, understandable pieces
- To give a chunk of logic a clear name that says what it does

````python
#Defining Function
def addTwoNumber(a,b):
  c = a+b
  print(c)

#Calling Function
addTwoNumber(2,3)
````

````python
#defining a function
def my_function():
  print("Hello This is a function")

my_function()
````

````python
#passing arguments
def my_function(fname):
  print(fname + " Kumar")   #always call function with the require number of args;not less,not more


my_function("Pratham")
````

````python
#calling
my_function("Shubham")
````

**Parameter vs Argument:** `fname` in `def my_function(fname):` is the **parameter** (a placeholder name); `"Pratham"` or `"Shubham"` in the call is the **argument** (the actual value handed over). Same function, different arguments each time — that's the whole point of writing it once.

⏱️ **Time & Space:** every example above (`addTwoNumber`, the basic `my_function()` calls) runs in **O(1) time and O(1) space** — no loops, no recursion, and the amount of work never depends on any input size.

````python
ig = ["Worry", "Reast", "Rohan", "Tejas"]

#arbitrary arguments: if number of arguments is unknown, add a * before the arg
def my_function(*kids):
  print("The youngest child is " + kids[1])

my_function("Worry", "Reast")
````

**Note:** the `*` before `kids` means *arbitrary arguments* — Python collects **any number** of positional arguments into a single **tuple** called `kids`. `kids[1]` just means "the second item in that tuple"; it isn't smart enough to know who's actually youngest.

**Heads up:** the `ig` list defined above (`["Worry", "Reast", "Rohan", "Tejas"]`) is never actually used — the call below re-types two of those names by hand instead of unpacking `ig` with `my_function(*ig)`. Try it yourself below to see the tuple directly:

⏱️ **Time & Space:** packing the extra positional arguments into `kids` costs **O(k) time and O(k) space**, where `k` is how many arguments you pass in — not related to the values inside them.

````python
def show_args(*kids):
    print(kids, type(kids))   # a tuple, however many names you pass

show_args("Worry", "Reast", "Rohan", "Tejas")   # *ig would do the same, unpacked from the list above
````

````python
#keyword arguments:You can also send arguments with the key = value syntax.This way the order of the arguments does not matter.
def my_function(child3, child2, child1):
  print("The youngest child is " + child3)
  print("The youngest child is " + child2)

my_function(child1 = "Emil", child2 = "Tobias", child3 = "Linus")
````

**Note:** keyword arguments (`child1=..., child2=..., child3=...`) are matched **by name**, not by position — that's why you can call `my_function(child1="Emil", child2="Tobias", child3="Linus")` in a completely different order than the function's own parameter list (`child3, child2, child1`) and still get the right value in the right place.

⏱️ **Time & Space:** O(1) — a fixed, constant number of parameters, regardless of the order you supply them in.

````python
#arbitrary keyword arguments, **kwargs
def my_function(**kid):
  print("His last name is " + kid["lname"])

my_function(fname = "Tobias", lname = "Refsnes")
````

**Note:** `**kid` collects any number of *keyword* arguments into a **dict** — `kid["lname"]` looks it up by key, the same way you'd read any Python dictionary.

| `*args` | `**kwargs` |
|---------|------------|
| Extra **positional** arguments | Extra **keyword** arguments |
| Arrives as a **tuple** | Arrives as a **dict** |
| `f("A", "B", "C")` | `f(a="A", b="B", c="C")` |

> 📌 **Trick:** one star, one type of collection (positional → tuple); two stars, the "double" structure (keyword → key/value dict).

⏱️ **Time & Space:** same idea as `*args` — building the `kid` dict costs **O(k) time and O(k) space**, where `k` is the number of keyword arguments passed.

````python
#default parameters
def my_function(country = "Norway"):
  print("I am from " + country)

my_function()
````

**Note:** a default parameter (`country = "Norway"`) is only used when the caller doesn't supply that argument — call `my_function("India")` and the default is simply ignored in favor of what you passed.

⏱️ **Time & Space:** O(1) — using or skipping a default value doesn't add any extra work.

````python
#return values
def my_function(x):
  return 5 * x


my_function(4)
````

## Recursion Function

- The process in which a function calls itself directly or indirectly is called recursion and the corresponding function is called a recursive function.

## 🎯 Learning Goal (Recursion)

By the end of this section you should understand what recursion is, the two required pieces (base case + recursive case), how the call stack tracks recursive calls, and how to trace classic recursion problems.

> 🧑‍🎓 **Analogy:** Recursion is like a set of **Russian nesting dolls**. To open the big doll, you first have to open the slightly smaller doll inside it, which means opening the even smaller doll inside *that* one — until you reach the tiniest doll that doesn't open at all. That tiniest doll is the base case; unpacking each doll on the way back out is the recursive case unwinding.

**Why do we need recursion specifically?**
- Some problems are naturally defined "in terms of themselves" — e.g. `factorial(5) = 5 × factorial(4)`. Recursion lets the code mirror that definition directly instead of forcing it into a loop.
- It's the natural fit for anything tree-shaped or nested (folders inside folders, "this depends on a smaller version of this" problems), which is why it shows up constantly in DSA interviews.

### Loops vs Recursion


### 🧠 Key Idea

- Every correct recursive function needs exactly two parts: a **base case** (stops, returns directly, no further calls) and a **recursive case** (calls itself with a smaller/simpler input, moving toward the base case).
- Every recursive call is a real function call — Python remembers each one on a **call stack**, so recursion always uses more memory than an equivalent loop.
- Code placed *before* the recursive call runs on the way **in**; code placed *after* it runs on the way **out** (while unwinding) — where you put your logic changes the output order.
- A recursive relationship can build on more than one previous value (Fibonacci uses 2, Tribonacci uses 3) — the pattern generalizes: "looks back k steps → needs k base cases."

### 📚 Important Terms

| Term | Simple Meaning | Example |
|------|----------------|----------|
| Recursion | A function calling itself to solve a smaller version of the same problem | `factorialNum(n)` calling `factorialNum(n-1)` |
| Base case | The stopping condition — returns directly, no further recursive call | `if n == 0: return 1` |
| Recursive case | The step that calls the function again with a smaller input | `return n * factorialNum(n-1)` |
| Call stack | The stack of "paused" calls Python tracks while recursion is unwinding | `fun(3)` stacks `fun(3)→fun(2)→fun(1)→fun(0)` before anything prints |
| Unwinding | Calls returning back up the stack, one by one, after the base case is hit | The `print()` calls firing in `fun(3)` |

**How a normal function call works** — control jumps in, runs, and comes straight back:

```mermaid
flowchart LR
Caller -->|calls with args| Function
Function -->|runs body| Function
Function -->|return value| Caller
```

**How a recursive call works** — each call pauses and waits on a growing call stack until the base case is hit, then everything unwinds in reverse:

```mermaid
flowchart TB
A["factorialNum(5)"] --> B["factorialNum(4)"]
B --> C["factorialNum(3)"]
C --> D["factorialNum(2)"]
D --> E["factorialNum(1)"]
E --> F["factorialNum(0) → returns 1 (base case)"]
F -.unwind: 1×1=1.-> E
E -.unwind: 2×1=2.-> D
D -.unwind: 3×2=6.-> C
C -.unwind: 4×6=24.-> B
B -.unwind: 5×24=120.-> A
```

````python
n = 10
i = 1

while i<=n:
    print(i, end=" ")
    i+=1
````

**Loop vs. Recursion** (same job — print the numbers 1 through n):

| Loop (above) | Recursion (below) |
|------|-----------|
| `while`/`for` repeats manually with a counter | Function calls itself with a smaller `n` |
| **Time:** O(n) | **Time:** O(n) |
| **Space:** O(1) — just a counter | **Space:** O(n) — one stack frame per call |
| Usually faster for simple repetition | Usually cleaner for naturally "nested" problems |
| No risk of "stack overflow" | Missing/broken base case → stack overflow |

````python
def printNumbers(i, n):
    # base case
    if i > n:
        return
    
    # recursive case
    print(i, end=" ")
    printNumbers(i+1, n)
    
printNumbers(1, 10)
````

**Spot the two pieces in this exact code:** `if i > n: return` is the **base case** — it stops the recursion once `i` runs past `n`. `printNumbers(i+1, n)` is the **recursive case** — it calls itself with `i` one step closer to that stopping point.

> 🧑‍🎓 **Real-life version:** asking "how many people are in the queue behind me?" Instead of counting everyone yourself, you ask the person right behind you "how many are behind *you*?" — and so on, until the last person (base case: "nobody's behind me") answers, and the count adds up on the way back to you.

⏱️ **Time & Space:** O(n) time (n-i+1 calls) and **O(n) space** — unlike the loop above, every call sits on the stack until the base case returns, so memory grows with `n` too.

````python
def factorialNum(n):
    if n==0:
        return 1
    return n*factorialNum(n-1)

print(factorialNum(5))
    
````

**Trace it:** `factorialNum(5)` calls `factorialNum(4)`, which calls `factorialNum(3)`, and so on down to `factorialNum(0)` (the base case, returns `1`). Then each paused call multiplies its own `n` by the answer it gets back: `1×1=1 → 2×1=2 → 3×2=6 → 4×6=24 → 5×24=120`. See the Mermaid diagram above for the full picture.

⏱️ **Time & Space:** O(n) time (n calls deep) and O(n) space (n stack frames waiting to multiply on the way back out).

````python
## Recursive Stack
def fun(n):
    if n==0:
        return 
    fun(n - 1)
    print(n, end=" ")

fun(3)
    
````

**Why this prints `1 2 3`, not `3 2 1`:** the `print(n, ...)` line comes **after** the recursive call `fun(n - 1)`, so nothing gets printed until the recursion hits `fun(0)` (the base case) and starts unwinding:

```
fun(3)
 └─ fun(2)
     └─ fun(1)
         └─ fun(0) -- base case, returns immediately
         └─ print(1)   ← unwinding starts here
     └─ print(2)
 └─ print(3)
```

⏱️ **Time & Space:** O(n) time and O(n) space — same shape as `factorialNum` and `printNumbers`: one call per step, one stack frame per call.

````python
##Fibonacci Series
# - F(0) = 0, F(1) = 1
# F(n) = F(n - 1) + F(n - 2), for n > 1.

def fibo(n):
    # base case
    if n == 0 or n == 1:
        return n
    
    # recursive case
    return fibo(n-1) + fibo(n-2)
    
print(fibo(6))
````

**Note:** `F(0) = 0` and `F(1) = 1` are the two base cases — Fibonacci "looks back" 2 steps, so it needs 2 stopping points.

⚠ **Misconception:** this naive version is *not* an efficient way to compute Fibonacci — it recomputes the same smaller values over and over (`fibo(4)` gets recalculated many times inside `fibo(6)`), so it gets very slow for larger `n`. That repeated-work problem is exactly what memoization/dynamic programming fixes later on.

⏱️ **Time & Space:** naive `fibo(n)` costs **O(2ⁿ) time** — it roughly doubles the work for every +1 to `n`, because each call branches into 2 more calls. But it only needs **O(n) space** — at any single moment, only one chain of calls (n deep) is sitting on the stack; the exponential blow-up is in *how many times* the function gets called overall, not in how much memory is used at once.

````python
## Practice Problem | 1137. N-th Tribonacci Number

def trifibo(n):
    #base case
    if n == 0 or n == 1:
        return n
    if n == 2:
        return 1 # Or 2, depending on your problem's specific starting sequence
    #recursive case
    return trifibo(n-1) + trifibo(n-2) + trifibo(n-3)

print(trifibo(25))
    
````

**Generalizing the pattern:** Tribonacci looks back **3** steps (`T(n) = T(n-1)+T(n-2)+T(n-3)`), so — following the same rule as Fibonacci — it needs **3** base cases (for `n = 0, 1, 2`) instead of 2.

> 📌 **Trick:** count how many steps back the formula looks — that's exactly how many base cases you need.

⏱️ **Time & Space:** same shape as Fibonacci, one branch wider — **O(3ⁿ) time** (each call branches into 3 more calls) but still just **O(n) space** (call stack depth), for the same reason as above.

````python
## 231. Power of Two - LEETCODE

def isPowerOfTwo(n):
    
    if (n<=0): return False
    if (n==1):
        return True 
    while n%2==0:
        n//=2
        
    if n == 1:
        return True
    else:
        return False
    
print(isPowerOfTwo(78))


def isPowerOfTwoRecursive(n):
    
    if (n<=0): return False
    if (n==1):
        return True
    if (n%2!=0):
        return False
    
    return isPowerOfTwoRecursive(n//2)
    
print(isPowerOfTwoRecursive(32))
````

**Note:** `isPowerOfTwo` (loop) and `isPowerOfTwoRecursive` solve the *exact same problem* two ways — repeatedly halving `n` while it's even. The loop version keeps a `while`; the recursive version turns each halving into a fresh call, with `n == 1` and "still even? no" as its two base cases. Same idea, different shape.

⏱️ **Time & Space:** both versions halve `n` every step, so both run in **O(log n) time**. They differ in space: the loop uses **O(1) space** (just reassigns `n`), while the recursive version uses **O(log n) space** — one stack frame per halving.

````python
## GCD of Two Numbers - Euclidean Number
# The Euclidean algorithm is a method for efficiently calculating the greatest common divisor (GCD) of two integers

def gcd(a, b):
    if b==0:
        return a
    
    return gcd(b, a%b)

def lcm(a, b):
    return a*b//gcd(a, b) # / - float divison, // - Integer Divison

print(gcd(15, 50))
print(lcm(15, 50))
````

**Note:** the Euclidean algorithm (`gcd(a, b) = gcd(b, a % b)`, base case `b == 0`) is one of the oldest algorithms in existence and a great one to memorize outright — short, elegant, and it shows up constantly in interviews. `lcm` then reuses `gcd` rather than solving the problem from scratch.

⏱️ **Time & Space:** the Euclidean algorithm runs in **O(log(min(a, b))) time** — it's famously fast, shrinking quickly even for huge numbers — and uses the same **O(log(min(a, b))) space** for its call stack. `lcm` just adds one constant-time step on top of `gcd`.

````python

# 50. Pow(x, n) - LEETCODE

class Solution:
    def findPow(self, x,n):
        #base case
        if n==0:
            return 1
        a = self.findPow(x, n//2)
        if n%2==0:
            return a*a
        else:
            return a*a*x

    def myPow(self, x: float, n: int) -> float:
        if n>=0:
            return self.findPow(x, n)
        else:
            return 1/self.findPow(x, n*(-1))
        
print(Solution().myPow(2.0, 10))
````

**Fast exponentiation:** `findPow` computes `x^n` by **halving** `n` each call (`n // 2`) instead of subtracting 1 each time — that's why it only needs **~log₂(n)** calls instead of n calls. Negative exponents are handled by flipping to `1 / findPow(x, -n)` at the very end.

**Shrink-by-1 vs. halve-each-time recursion:**

| Shrink by 1 (factorial, Fibonacci) | Halve each time (`Pow(x, n)`) |
|---|---|
| `n` calls deep | ~`log₂(n)` calls deep |
| Simple to write, but slow for huge `n` | Far fewer calls — much faster for huge `n` |

⏱️ **Time & Space:** halving `n` every call gives **O(log n) time** — the whole reason it's called *fast* exponentiation, compared to the **O(n) time** you'd get multiplying `x` by itself `n` times in a loop. Space is **O(log n)** too, for the call stack.

## 🛑 What Happens When Recursion Never Stops

Python doesn't let a recursive function call itself forever — it protects itself with a built-in **recursion limit**.

- Every recursive call adds one more "frame" to the call stack, and Python keeps counting how deep that stack gets.
- By default, Python allows a maximum of **1000** stacked calls — check it yourself with `sys.getrecursionlimit()`.
- If a function has no base case (or a broken one that never triggers), it keeps calling itself, deeper and deeper, until it crosses that limit. Python then stops it with a `RecursionError`, instead of letting the program hang forever or silently eat up all your computer's memory.

> 🧑‍🎓 **Analogy:** the recursion limit is like a **circuit breaker** in your house. Draw too much power (recurse too deep) and it doesn't let the wiring melt down (crash the whole computer) — it trips cleanly and predictably, at a known threshold, so you get a clear signal that something's wrong.

````python
import sys
print(sys.getrecursionlimit())   # -> 1000 (the default, on most systems)

def infinite(n):
    return infinite(n + 1)   # no base case -- nothing ever stops this on its own

infinite(0)
# RecursionError: maximum recursion depth exceeded
````

- You *can* raise the ceiling with `sys.setrecursionlimit(5000)`, but that doesn't fix a missing base case — it just delays the crash and pushes closer to your computer's *real* hardware call stack limit. Push it too far past that, and instead of a clean, catchable `RecursionError`, Python can crash outright with a **segmentation fault** — no exception to catch, the interpreter itself dies.
- The correct fix is never "raise the limit" — it's always "make sure the base case is actually reachable" (e.g. check that `n` is genuinely shrinking toward it on every call).

⏱️ **Time & Space:** a runaway recursion like `infinite(n)` keeps calling itself until it crosses the recursion limit — so right before it crashes, it's holding roughly **O(limit) stack frames in space** (about 1000 by default), for effectively unbounded time (it never reaches a base case on its own).

## 📝 Quick Revision

- Recursion = a function calling itself on a smaller version of the same problem.
- Every recursive function needs a base case (stops) and a recursive case (calls itself, moving toward the base case).
- Work often happens on the way back **out** (unwinding), not just on the way in — see `fun(3)`.
- "Looks back k steps → needs k base cases" (Fibonacci: 2, Tribonacci: 3).
- Naive recursive Fibonacci/Tribonacci recomputes values repeatedly — slow without memoization.
- Fast exponentiation halves the exponent each call instead of subtracting 1 — ~log₂(n) calls instead of n.
- Python caps recursion depth at 1000 calls by default (`sys.getrecursionlimit()`) and raises `RecursionError` past that — raising the limit only delays a real problem, it doesn't fix it.
- Recursion trades memory/speed for code that mirrors the problem's own self-similar definition.

## 🎯 Key Takeaways

1. A function is a reusable, named block of code — define once, call as many times as you like. Python gives four ways to pass data in: positional args, `*args`, keyword args, and `**kwargs` — plus default parameter values.
2. Recursion is a function calling itself on a smaller version of the same problem — it always needs a base case and a recursive case.
3. Recursive calls stack up in memory before unwinding — code placed *after* the recursive call runs during the unwind, which is why `fun(3)` prints `1 2 3`, not `3 2 1`.
4. Looking back further in a sequence (Tribonacci looks back 3 vs. Fibonacci's 2) means you need that many base cases — the pattern generalizes cleanly.
5. Not all recursion shrinks by 1 each call — fast exponentiation halves the input instead, needing far fewer calls.
6. Python caps recursion at 1000 calls by default and raises `RecursionError` past that — a safety net, not a bug, and raising the limit is never the real fix for a missing base case.

```
 └─ func(count=1)
     └─ func(count=2)
         └─ func(count=3)
             └─ func(count=4) -- base case, returns 4
             └─ print("Pratham:", 4)   ← unwinding starts here
         └─ print("Pratham:", 3)
     └─ print("Pratham:", 2)
 └─ print("Pratham:", 1)
```

Each frame's `count` is incremented **before** the recursive call, then that same frame's own `count` is printed **after** it — so the deepest frame (`count=4`) prints first on the way back out, and the outermost frame (`count=1`) prints last. Because each frame owns its own `count`, the printed values shrink on the way out (`4 → 3 → 2 → 1`) instead of all showing the same final number, the way a single shared/global counter would.

```mermaid
flowchart TB
A["func(count=0)"] --> B["func(count=1)"]
B --> C["func(count=2)"]
C --> D["func(count=3)"]
D --> E["func(count=4) → base case, returns 4"]
E -.print 4, unwind.-> D
D -.print 3, unwind.-> C
C -.print 2, unwind.-> B
B -.print 1, unwind.-> A
```

⏱️ **Time & Space:** `func(count=0)` runs in **O(1) time and O(1) space** — the base case is a fixed constant (`count == 4`), not tied to any input size, so it's always exactly 5 stack frames (0 through 4) no matter what. Compare this with `factorialNum(n)`/`fibo(n)` above, where the number of calls scales with the input `n` — here there's no `n` at all, just a hardcoded stopping point.

````python
## Recursion Using Parameter : [print x = ?, N times] 
def printFunc(x, N):

    if N == 0:
        return
    print(x)
    printFunc(x, N-1)



printFunc(15, 4)
````

⏱️ **Time & Space:** `printFunc(x, N)` has **no base case at all** — `N` is never used to stop anything, so this calls itself forever until it crosses Python's recursion limit and crashes with `RecursionError` (see the "What Happens When Recursion Never Stops" section above). Right before crashing, time is unbounded and space is roughly **O(limit)** stack frames — the same failure mode as the `infinite(n)` demo, just with an unused parameter making it *look* like it should terminate.

````python
## Parametrized amd Functional Recursion

def sumFunc(sum, i, N):

    if i > N:
        print(sum)
        return

    sumFunc(sum+i, i+1, N)

sumFunc(0, 1, 10)
````

````python
## Question Reverse an Array using Recursion (left = 2, right = 5) | (left = 0, right = 7)
def reverse_array_recursion(nums: list[int], left, right):

    if left >= right:
        return nums

    nums[left], nums[right] = nums[right], nums[left]

    return reverse_array_recursion(nums, left + 1, right - 1)

num = [5, 7, 3, 2, 6, 1, 5, 9]
print(num)
print(reverse_array_recursion(num, 2, 5))
````

````python
# Question: check if a string is palindrome or not

def palin_check_string_recursion(s: str,left, right):

    if left >= right:
         return True

    if s[left] != s[right]:
                return False

    return palin_check_string_recursion(s, left+1, right-1)


def palin_check_string_loop(s: str) -> bool:

    n = len(s)
    left = 0
    right = n - 1

    while(left < right):
        if s[left] != s[right]:
            return False
        left += 1
        right -= 1
    return True
    

strr = "ANBCDUDCBNA"
print(palin_check_string_recursion(strr, 0, len(strr)-1))
print(palin_check_string_loop(strr))
````

````python
# Question : find the fibonacci series using recursion
class Solution:
    def func(self, N: int)-> int:
        if N == 0 or N == 1:
            return N
        return func(N-1) + func(N-2)

    def fibo(self, N: int) -> int:
        answer = self.func(N)
        return answer


num = 9
solution = Solution()
print(solution.fibo(num))
````
