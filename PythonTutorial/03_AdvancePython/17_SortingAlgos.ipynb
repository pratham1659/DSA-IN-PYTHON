**Where each sort in this notebook lands:**

| Algorithm | Best | Average | Worst | Space |
|-----------|------|---------|-------|-------|
| Insertion Sort | O(n) | O(n²) | O(n²) | O(1) |
| Bubble Sort (with `is_swap` flag) | O(n) | O(n²) | O(n²) | O(1) |
| Selection Sort | O(n²) | O(n²) | O(n²) | O(1) |

> 💡 All three sit at O(n²) worst-case -- the only real difference between them is how good their BEST case gets on a nearly-sorted array.


### Selection Sort

````python
## Selection Sort Algorithm

def selectionSort(nums: list[int]) -> list[int]:
    n = len(nums)

    for i in range(0, n):
        min_idx = i
        for j in range(i + 1, n):
            if nums[j] < nums[min_idx]:
                min_idx = j
        nums[i], nums[min_idx] = nums[min_idx], nums[i]   # swap once per outer pass, after the inner loop finds the minimum

    return nums


nums = [5, 7, 8, 4, 1, 6, 9, 2]
print(selectionSort(nums))
````

## ⏱ Time and Space Complexity -- Selection Sort

| Case | Time | Why |
|------|------|-----|
| Best | O(n²) | Still scans the remaining elements every pass to find the minimum, even if the array is already sorted -- no early exit |
| Average | O(n²) | Same nested-loop scan regardless of input order |
| Worst | O(n²) | Same nested-loop scan, even for a reverse-sorted array |

**Space: O(1)** -- sorts in place; only a few scalar variables (`i`, `j`, `min_idx`) are used, no extra array.

> 💡 Selection Sort's time is always O(n²) no matter how the input is arranged -- unlike some other sorts (e.g. Insertion Sort), it has no "already sorted" shortcut, because it always scans the full remaining unsorted portion to find the minimum before it can place anything.


### Bubble Sort in Python [Adjacent Swaps]


````python
## bubble sort

def bubbleSort(nums: list[int]) -> list[int]:
    n = len(nums)

    for i in range(n - 2, -1, -1):
        is_swap = False
        for j in range(0, i + 1):
            if nums[j] > nums[j + 1]:
                nums[j], nums[j + 1] = nums[j + 1], nums[j]
                is_swap = True
        if is_swap == False:
            return nums   
    return nums


nums = [5, 7, 8, 4, 1, 6, 9, 2]
print(bubbleSort(nums))
````

## ⏱ Time and Space Complexity -- Bubble Sort

| Case | Time | Why |
|------|------|-----|
| Best | O(n) | This version now has a `is_swap` flag -- if a full pass makes zero swaps, the array is already sorted, so it returns immediately instead of running the remaining passes |
| Average | O(n²) | A partially unsorted array still needs multiple passes, each scanning the unsorted portion |
| Worst | O(n²) | A reverse-sorted array needs a swap on every single comparison, every pass -- the flag never triggers an early exit |

**Space: O(1)** -- sorts in place; only a few scalar variables (`i`, `j`, `is_swap`) are used, no extra array.

> 💡 The `is_swap` flag is exactly the optimization mentioned earlier: it detects "no swaps this pass = already sorted" and exits early, dropping the best case from O(n²) down to O(n). Average and worst case are unchanged, since a genuinely unsorted array still needs the full nested-loop scan.


### Insertion Sort in Python


````python
def insertionSort(nums: list[int]) -> list[int]:
    n = len(nums)
    for i in range(1, n):
        key = nums[i]
        j = i - 1

        while j >= 0 and nums[j] > key:
            nums[j + 1] = nums[j]
            j -= 1
        nums[j + 1] = key

    return nums   # fixed: was missing entirely -- function returned None


nums = [5, 7, 8, 4, 1, 6, 9, 2]
print(insertionSort(nums))   # fixed: was calling bubbleSort(nums), a leftover from copy-paste
````

## ⏱ Time and Space Complexity -- Insertion Sort

| Case | Time | Why |
|------|------|-----|
| Best | O(n) | If the array is already sorted, `nums[j] > key` is never true, so the `while` loop exits immediately every time -- just one pass with no shifting |
| Average | O(n²) | Each element typically needs to shift back through roughly half the sorted portion to find its spot |
| Worst | O(n²) | A reverse-sorted array forces every element to shift all the way back to index 0 |

**Space: O(1)** -- sorts in place; only a few scalar variables (`i`, `j`, `key`) are used, no extra array.

> 💡 Insertion Sort naturally gets O(n) best case with NO extra flag needed (unlike Bubble Sort) -- the `while` condition itself is the early-exit check.
