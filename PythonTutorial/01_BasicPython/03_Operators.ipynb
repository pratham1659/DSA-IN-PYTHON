## 🤔 What is it?

An **operator** is a symbol that performs an operation on one or more values (called **operands**) — like `+` adding two numbers, or `==` comparing them.

> 🧸 Think of operators like verbs in a sentence. `a + b` reads almost like English: "a, plus, b." The operator is the action; the operands are what the action is performed on.

---

## ❓ Why do we need it?

Every calculation, comparison, and decision a program makes relies on operators. Knowing exactly what each one does — and which one runs FIRST when several are mixed together — is the difference between code that works and code that silently produces the wrong answer.

---

## 🧠 Key Idea

- Operators are grouped by what they do: **arithmetic** (math), **comparison** (relate two values), **logical** (combine booleans), **bitwise** (work on binary bits), **membership** (check inside a collection), and **identity** (check object sameness).
- **Precedence** decides which operator runs first when several appear in one expression (just like PEMDAS/BODMAS in math class).
- **Associativity** decides the order when operators of the SAME precedence appear together (most are left-to-right, but `**` is right-to-left).
- `/` always returns a `float`; `//` (floor division) always rounds DOWN toward negative infinity, not toward zero.

## ➕ Arithmetic Operators

| Operator | Meaning | Example | Result |
|----------|---------|---------|--------|
| `+` | Addition | `5 + 3` | `8` |
| `-` | Subtraction | `5 - 3` | `2` |
| `*` | Multiplication | `5 * 3` | `15` |
| `/` | Division (always returns a `float`) | `5 / 2` | `2.5` |
| `//` | Floor Division (rounds DOWN, toward negative infinity) | `5 // 2` | `2` |
| `%` | Modulo (the remainder) | `5 % 2` | `1` |
| `**` | Exponentiation (power) | `5 ** 2` | `25` |

````python
x = 5
print(x ** 3)   # exponentiation: 5 to the power of 3 -- NOT a square root!
````

> ❌ Misconception: `x ** 3` calculates a square root.
> ✅ Correct: `**` is exponentiation (raises to a power). A square root is `x ** 0.5` or `math.sqrt(x)`.

````python
import math

x = 25
print(x ** 0.5)     # square root via exponent 0.5
print(math.sqrt(x))  # the more explicit, readable way
````

### Floor Division and Modulo with Negative Numbers

This is a genuine gotcha: `//` rounds toward **negative infinity**, not toward zero — so it's not the same as "chop off the decimal part" when negative numbers are involved.

> ❌ Misconception: `-7 // 2` gives `-3` (like truncating `-3.5` toward zero).
> ✅ Correct: it gives `-4`, because floor division always rounds DOWN (toward negative infinity). And Python's `%` always returns a result with the SAME SIGN as the divisor.

````python
print(7 // 2, 7 % 2)     # 3, 1  -- straightforward with positive numbers
print(-7 // 2, -7 % 2)   # -4, 1 -- floor division rounds DOWN, not toward zero
````

````python
a = 10
b = 10
x = a + 3
y = b - 3
print(x, y)
````

````python
a = 10
b = 5
add = a + b       # adding int + int -> returns an int
mul = a * b
div = a / b       # / ALWAYS returns a float, even if it divides evenly
exp = a ** b      # a to the power of b
concat = str(a) + '!'   # + on strings concatenates, not adds
mul_string = b * 3        # * on a string repeats it -- NOT multiplication in the math sense

print(add, mul, div, exp, concat, mul_string)
````

## 🔁 Augmented Assignment Operators

`a += 3` is shorthand for `a = a + 3` — every arithmetic operator has an augmented assignment version.

| Operator | Same As |
|----------|---------|
| `a += b` | `a = a + b` |
| `a -= b` | `a = a - b` |
| `a *= b` | `a = a * b` |
| `a /= b` | `a = a / b` |
| `a //= b` | `a = a // b` |
| `a %= b` | `a = a % b` |
| `a **= b` | `a = a ** b` |

````python
a = 10
b = 10
a += 3
b -= 3
print(a, b)
````

````python
c = 2
c *= 4     # c = c * 4
print(c)

c //= 3    # c = c // 3
print(c)

c **= 3    # c = c ** 3
print(c)
````

## 🥇 Operator Precedence

When an expression mixes multiple operators, Python follows a fixed order — similar to PEMDAS/BODMAS from math class:

1. Parentheses `()`
2. Exponentiation `**`
3. Multiplication `*`, Division `/`, Floor Division `//`, Modulo `%`
4. Addition `+`, Subtraction `-`

> 💡 **Associativity matters too:** most operators evaluate left-to-right, but `**` is the odd one out — it's **right-to-left**. `2 ** 3 ** 2` means `2 ** (3 ** 2)` = `2 ** 9` = `512`, NOT `(2 ** 3) ** 2` = `64`.

````python
num1 = 10 + 3 * 2      # * runs before + -> 10 + 6 = 16
num2 = (2 + 3) * 10 - 3  # parentheses run first -> 5 * 10 - 3 = 47
print(num1)
print(num2)
````

````python
# ** is right-to-left, unlike most other operators
print(2 ** 3 ** 2)     # 2 ** (3 ** 2) = 2 ** 9 = 512
print((2 ** 3) ** 2)   # explicitly left-to-right instead -> 8 ** 2 = 64
````

## 🔢 Rounding: `round()`, `abs()`, `math.ceil()`, `math.floor()`

> ❌ Misconception: `round()` always rounds `.5` UP.
> ✅ Correct: Python's `round()` uses **"round half to even"** (banker's rounding) for exact `.5` values — `round(2.5)` gives `2`, not `3`, because `2` is the nearest EVEN number. This surprises almost everyone the first time they see it.

````python
print(round(2.6))   # 3 -- clearly closer to 3
print(round(2.9))   # 3 -- clearly closer to 3
print(round(2.5))   # 2 -- exact .5 rounds to the nearest EVEN number, not always up!
print(round(3.5))   # 4 -- 4 is the nearest even number here
print(abs(-2.4))    # abs() always returns the positive/magnitude value
````

`math.ceil()` always rounds UP to the next whole number; `math.floor()` always rounds DOWN — neither one looks at "which is closer," unlike `round()`.

````python
import math

print(math.ceil(2.9))    # 3 -- always rounds up
print(math.ceil(2.2))    # 3 -- always rounds up, even though 2.2 is closer to 2
print(math.floor(2.9))   # 2 -- always rounds down, even though 2.9 is closer to 3
print(math.floor(2.2))   # 2 -- always rounds down
````

## 🔧 Bitwise Operators

These operate on the **binary representation** of integers, bit by bit. Less commonly used in everyday code, but important for interviews, flags/permissions, and low-level performance tricks.

| Operator | Name | Meaning |
|----------|------|---------|
| `&` | AND | `1` only where BOTH bits are `1` |
| `\|` | OR | `1` where AT LEAST ONE bit is `1` |
| `^` | XOR | `1` where the bits are DIFFERENT |
| `~` | NOT | Flips every bit (and effectively computes `-x - 1`) |
| `<<` | Left Shift | Shifts bits left (multiplies by `2` per shift) |
| `>>` | Right Shift | Shifts bits right (divides by `2` per shift, rounding down) |

> 🧸 Think of `&`/`|`/`^` like comparing two rows of light switches: `&` is on only where BOTH rows have the switch on, `|` is on where EITHER row has it on, `^` is on only where they DISAGREE.

````python
a = 5   # binary: 0101
b = 3   # binary: 0011

print(a & b)    # 1  -- 0101 & 0011 = 0001
print(a | b)    # 7  -- 0101 | 0011 = 0111
print(a ^ b)    # 6  -- 0101 ^ 0011 = 0110 (bits that differ)
print(~a)       # -6 -- flips all bits; equivalent to -(a + 1)
print(a << 1)   # 10 -- shift left by 1 = multiply by 2
print(a >> 1)   # 2  -- shift right by 1 = divide by 2, rounding down
````

## 📋 Membership Operators: `in` and `not in`

Check whether a value exists inside a collection, without needing to write a search loop yourself.

````python
list1 = [1, 2, 3, 4, 5]
list2 = [6, 7, 8, 9]

for item in list1:
    if item in list2:
        print(item, "-> overlapping")
    else:
        print(item, "-> not overlapping")
````

The example above never actually shows the "overlapping" branch, since `list1` and `list2` share no values — here's a version where some values DO overlap, so both branches run.

````python
list1 = [1, 2, 3, 4, 5]
list3 = [4, 5, 6, 7]   # 4 and 5 overlap with list1

for item in list1:
    if item in list3:
        print(item, "-> overlapping")
    else:
        print(item, "-> not overlapping")
````

### `not in`

> ⚠ Naming tip: avoid naming a variable `list` — it shadows Python's built-in `list()` type, which could break other code later in the same scope that expects the real `list()` constructor.

````python
y = 20
numbers = [10, 20, 30, 40, 50]

if y not in numbers:
    print("y is NOT present in the given list")
else:
    print("y IS present in the given list")   # this is the branch that actually runs, since y = 20 is in the list
````

## 🆔 Identity Operators: `is` and `is not`

Fully explained in [04_Condition.ipynb](04_Condition.ipynb) (identity vs equality) — here's a common practical use: checking a value's type.

> 💡 For simple exact-type checks, `type(x) is int` works fine. But the more Pythonic and inheritance-aware choice is usually `isinstance(x, int)` — it also returns `True` for subclasses of `int`, which `type(x) is int` would miss.

````python
x = 5

if type(x) is int:
    print("true")
else:
    print("false")

print(isinstance(x, int))  # the more idiomatic way to check a type
````

````python
x = 5.2

if type(x) is not int:
    print("true")
else:
    print("false")
````

## 🦭 The Walrus Operator `:=` (Python 3.8+)

The walrus operator lets you **assign a value AND use it in the same expression** — handy for avoiding a redundant second call, especially inside a `while` loop or an `if` condition.

````python
# Without the walrus operator -- input() gets called, then compared, in two lines
data = input("Enter something (or 'quit' to stop): ")
if data != "quit":
    print(f"You entered: {data}")
````

````python
# With the walrus operator -- assign AND check in one line
numbers = [1, 2, 3, 4, 5]

# without walrus: you'd need to compute len(numbers) separately before the condition
if (n := len(numbers)) > 3:
    print(f"The list has {n} items, which is more than 3")
````

## ⚠ Common Misconceptions

❌ `x ** 3` calculates a square root.
✅ `**` is exponentiation (raises to a power); a square root is `x ** 0.5` or `math.sqrt(x)`.

❌ `-7 // 2` truncates toward zero, giving `-3`.
✅ Floor division always rounds DOWN toward negative infinity, giving `-4`.

❌ `round(2.5)` always rounds up to `3`.
✅ Python uses "round half to even" (banker's rounding) — `round(2.5)` gives `2`, the nearest even number.

❌ `2 ** 3 ** 2` is evaluated left-to-right, like most operators.
✅ `**` is right-to-left associative: it's `2 ** (3 ** 2)` = `512`, not `(2 ** 3) ** 2` = `64`.

❌ Naming a variable `list`, `sum`, or `type` is harmless.
✅ It shadows the built-in function/type of the same name for the rest of that scope — a classic source of confusing bugs later on.

## 📝 Quick Revision

- `/` always returns a float; `//` floors toward negative infinity (not toward zero).
- `%` (modulo) returns a result with the same sign as the divisor.
- Augmented assignment (`+=`, `-=`, ...) is shorthand for `a = a <op> b`.
- Precedence order: parentheses → exponentiation → multiplication/division → addition/subtraction.
- `**` is right-to-left associative — unusual among Python's operators.
- `round()` uses banker's rounding (round half to even); `math.ceil()`/`math.floor()` always round up/down.
- Bitwise operators (`& | ^ ~ << >>`) work on the binary representation of integers.
- `in`/`not in` check membership; `is`/`is not` check identity, not equality.
- The walrus operator `:=` assigns and returns a value in the same expression.
