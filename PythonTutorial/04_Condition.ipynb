---

## 🤔 What is it?

A **conditional statement** lets your program make a decision — run one block of code if something is `True`, and a different block if it's `False`.

> 🧸 Think of it like a fork in a hiking trail with a signpost: "If it's raining, take the shelter path. Otherwise, take the scenic path." The program checks the condition once and picks exactly one path.

---

## ❓ Why do we need it?

Without conditions, every program would run the exact same steps every single time, regardless of input. Real programs need to react differently depending on circumstances — a login system needs to check if a password is correct, a game needs to check if a player has won, a form needs to validate what the user typed.

---

## 🧠 Key Idea

- A condition is any expression that evaluates to `True` or `False` (a `bool`).
- `if` runs its block only when the condition is `True`; `elif` checks another condition only if all previous ones were `False`; `else` catches everything else.
- Python checks `elif` branches **in order** and stops at the first `True` one — so if two branches could both be true for the same value, only the first one ever runs, and the second becomes unreachable "dead code" for that overlap.
- Every value in Python has a "truthiness" — even without an explicit comparison, `if x:` checks whether `x` is truthy.

---

## 📚 Important Terms

| Term | Simple Meaning | Example |
|------|----------------|----------|
| Condition | An expression that evaluates to `True`/`False` | `age >= 18` |
| `if` / `elif` / `else` | Branch based on a condition | See below |

| Ternary / Shorthand if-else | A one-line if-else expression | `"A" if x > y else "B"` |

## ⚖ Comparison Operators

These compare two values and always produce a `bool` (`True`/`False`).

| Operator | Meaning | Example |
|----------|---------|---------|
| `==` | Equal to | `5 == 5` → `True` |
| `!=` | Not equal to | `5 != 3` → `True` |
| `>` | Greater than | `5 > 3` → `True` |
| `<` | Less than | `5 < 3` → `False` |
| `>=` | Greater than or equal to | `5 >= 5` → `True` |
| `<=` | Less than or equal to | `5 <= 3` → `False` |

> ❌ Misconception: `=` and `==` do the same thing.
> ✅ Correct: `=` **assigns** a value (`x = 5`), `==` **compares** two values (`x == 5`). Using `=` where you meant `==` is one of the most common beginner typos — and in an `if` statement, Python's syntax actually prevents it (`if x = 5:` is a `SyntaxError`), which is a helpful safety net.

````python
print(5 == 5)
print(5 != 3)
print(5 > 3)
print(5 < 3)
print(5 >= 5)
print(5 <= 3)
````

## 🔀 The `if` Statement

If the condition is `True`, the indented block runs. If it's `False`, the block is simply skipped — there's no `else` here, so nothing happens.

````python
a = 33
b = 200                    # be mindful of indentation -- it's how Python knows what's INSIDE the if-block
if b > a:
    print("b is greater than a")
````

---

## 🔀 `if` / `elif` / `else`

`elif` ("else if") lets you check additional conditions if the first one was `False`. `else` is the catch-all for when none of the above were `True`.

````python
is_hot = False
is_cold = False

if is_hot:
    print("It's very Hot")
    print("Drink plenty of water")
elif is_cold:
    print("It's a Cold Day")
    print("Wear warm clothes")
else:
    print("It's a Lovely Day")

print("Enjoy your Day")
````

### Comparing Against User Input

````python
var1 = 20
var2 = 30

var3 = int(input())

if var3 > var2:
    print("Greater")
elif var2 == var3:
    print("Equal")
else:
    print("Lesser")
````

## 🪆 Nested `if` Statements

An `if`/`elif`/`else` chain can contain ANOTHER `if`/`elif`/`else` chain inside one of its branches — useful when a decision depends on more than one condition, in stages.

````python
myvar = int(input("Enter the Age: "))

if myvar < 0:
    print("Number is Negative")
elif myvar > 0:
    if myvar <= 10:
        print("Number is Between 0 and 10")
    elif myvar > 10 and myvar <= 20:
        print("Number is Between 10 and 20")
    else:
        print("Number is Greater than 20")
else:
    print("Number is Zero")
````

## ⚠ A Real Bug: Overlapping `elif` Boundaries

Here's a time-of-day greeting checker. Look closely at the boundary numbers: `<= 12` in one branch and `>= 12` in the very next one — the value `12` satisfies BOTH.

> ❌ Misconception: overlapping boundaries like this cause wrong output.
> ✅ What actually happens: since Python checks `elif` branches in order and stops at the first `True` match, `time = 12` triggers the "Good Morning" branch (checked first) — the "Good Afternoon" branch's `>= 12` is **unreachable dead code** for that value. The output isn't wrong, but the boundary logic is confusing and fragile — the next section shows a cleaner way to write it.

````python
time = int(input("Enter your time: "))

if time < 0:
    print("Invalid time")
elif time > 0:
    if time <= 2:
        print("Good Night Sir")
    elif time >= 3 and time <= 12:
        print("Good Morning Sir")
    elif time >= 12 and time <= 17:   # 12 overlaps with the branch above -- unreachable for time == 12
        print("Good Afternoon Sir")
    elif time >= 17 and time <= 19:   # 17 overlaps with the branch above -- unreachable for time == 17
        print("Good Evening Sir")
    elif time >= 19 and time <= 24:   # 19 overlaps with the branch above -- unreachable for time == 19
        print("Good Night Sir")
    else:
        print("Invalid time")
else:
    print("Invalid time")   # NOTE: time == 0 falls here too, which is also debatable -- is midnight "invalid"?
````

### ✅ A Cleaner Version — No Overlapping Boundaries, No Dead Code

Once you sort boundaries from smallest to largest and only compare against the upper bound, each `elif` naturally only handles the range ABOVE the previous one — no overlap is even possible, and there's no need for the extra nested `if`.

````python
time = int(input("Enter your time (0-24): "))

if time < 0 or time > 24:
    print("Invalid time")
elif time <= 2:
    print("Good Night Sir")
elif time <= 12:
    print("Good Morning Sir")
elif time <= 17:
    print("Good Afternoon Sir")
elif time <= 19:
    print("Good Evening Sir")
else:
    print("Good Night Sir")
````

## 📋 Membership Operators: `in` and `not in`

These check whether a value exists inside a collection (list, string, tuple, ...) — returning a plain `bool`, without needing a loop.

````python
list1 = [3, 4, 5, 6]

print(5 in list1)        # True -- 5 IS in the list
print(10 in list1)       # False -- 10 is NOT in the list
print(10 not in list1)   # True -- the opposite check

if 3 in list1:
    print("Yes, it's in the list")
````

## ⚡ Shorthand if-else (Ternary Expression)

`value_if_true if condition else value_if_false` packs an entire if-else into ONE line — great for short, simple choices, but avoid nesting these deeply, since it hurts readability fast.

````python
a = 2
b = 330

print("A") if a > b else print("B")

# same idea, but assigning a value instead of directly calling a function
result = "A wins" if a > b else "B wins"
print(result)
````

## 🔗 Logical Operators: `and`, `or`, `not`

| Operator | Meaning | Example |
|----------|---------|---------|
| `and` | `True` only if BOTH sides are `True` | `a > b and c > a` |
| `or` | `True` if AT LEAST ONE side is `True` | `a > b or a > c` |
| `not` | Flips `True` to `False` and vice versa | `not is_raining` |

````python
a = 200
b = 33
c = 500

if a > b and c > a:
    print("Both conditions are True")

if a > b or a > c:
    print("At least one of the conditions is True")
````

`not` is the one missing from the original notes — it simply inverts a boolean.

````python
is_raining = False

if not is_raining:
    print("No umbrella needed today")

print(not True)    # False
print(not 0)       # True -- 0 is falsy, so "not 0" becomes True
````

## 🆔 Identity Operators: `is` and `is not`

`==` checks if two values are **equal**. `is` checks if two variables point to the exact **same object** in memory. These are NOT the same thing, and mixing them up is a classic bug source.

> ❌ Misconception: `is` and `==` are interchangeable.
> ✅ Correct: two separate lists with identical contents are `==` (equal) but NOT `is` (not the same object). Always use `is` only for comparing against `None`, `True`, or `False` — never for regular value comparisons.

````python
list_a = [1, 2, 3]
list_b = [1, 2, 3]      # a DIFFERENT list object, with the same contents
list_c = list_a          # list_c is just another name for the SAME object as list_a

print(list_a == list_b)  # True -- same contents
print(list_a is list_b)  # False -- different objects in memory
print(list_a is list_c)  # True -- literally the same object
````

````python
# The correct, idiomatic way to check for None
result = None

print(result is None)     # correct style
print(result == None)     # also technically works, but "is None" is the Python convention
````

## ✅❌ Truthy and Falsy Values

Every value in Python can be tested in a condition, even without an explicit comparison — `if x:` checks whether `x` is "truthy." A small, fixed set of values count as **falsy**; everything else is truthy.

**Falsy values:** `0`, `0.0`, `""` (empty string), `[]` (empty list), `{}` (empty dict), `()` (empty tuple), `None`, `False`.

> ❌ Misconception: `if x:` only works for actual booleans.
> ✅ Correct: `if x:` works for ANY value — it implicitly checks truthiness. An empty list, empty string, or `0` are all treated as `False`.

````python
values = [0, 1, -1, "", "hello", [], [1, 2], {}, {"a": 1}, None, False, True, 0.0]

for v in values:
    print(repr(v), "->", bool(v))
````

````python
# This is why "if my_list:" is a common, idiomatic way to check for an empty list
my_list = []

if my_list:
    print("The list has items")
else:
    print("The list is empty")   # this runs -- an empty list is falsy
````

## 🎯 The `match`-`case` Statement (Python 3.10+)

`match`-`case` is Python's version of a "switch statement" found in other languages — instead of a long `if`/`elif`/`elif`/... chain comparing one variable against many values, you list out the cases directly.

- `case _:` is the **wildcard** — matches anything, like a final `else`.
- `case _ if condition:` is a **guard clause** — matches anything, but only runs if `condition` is also `True`.
- Cases are checked top to bottom, and (just like `elif`) only the FIRST matching case runs.

````python
x = int(input("Enter the value of X: "))

match x:
    case 0:
        print("x is Zero")
    case 4:
        print("x matched the specific case 4")
    case _ if x < 10:
        print("x is < 10 (but not 0 or 4)")
    case _:
        print(x)
````

|--------|-----------|----------------------------|
| Comparison | `==`, `!=`, `>`, `<`, `>=`, `<=` | "How do these two VALUES relate?" |
| Logical | `and`, `or`, `not` | "How do multiple CONDITIONS combine?" |
| Membership | `in`, `not in` | "Does this value EXIST inside that collection?" |
| Identity | `is`, `is not` | "Are these the exact SAME OBJECT in memory?" |

## ⚠ Common Misconceptions

❌ `=` and `==` do the same thing.
✅ `=` assigns; `==` compares. Python's syntax blocks the most dangerous mix-up (`if x = 5:` is a `SyntaxError`).

❌ `is` and `==` are interchangeable.
✅ `==` compares values; `is` compares object identity. Two equal-looking lists are usually NOT the same object — use `is` only for `None`/`True`/`False` checks.

❌ Overlapping `elif` boundaries (like `<= 12` then `>= 12`) cause incorrect output.
✅ They don't cause wrong output — `elif` stops at the first match — but the overlapping branch becomes unreachable dead code, which is confusing and worth cleaning up (see the time-of-day example above).

❌ `if x:` only works when `x` is already a `True`/`False` value.
✅ `if x:` works on any value by checking its truthiness — `0`, `""`, `[]`, `{}`, `()`, and `None` are all falsy; everything else is truthy.

❌ `match`-`case` works in any Python version, just like `if`/`elif`.
✅ `match`-`case` requires Python 3.10 or later — trying to use it on an older version raises a `SyntaxError`.

## 📝 Quick Revision

- `if`/`elif`/`else` picks exactly ONE branch to run, checked top to bottom.
- Comparison operators (`==`, `!=`, `>`, `<`, `>=`, `<=`) compare values; logical operators (`and`, `or`, `not`) combine conditions.
- `in`/`not in` check membership in a collection; `is`/`is not` check object identity, not value equality.
- Overlapping `elif` boundaries don't break your code, but they do create confusing, unreachable dead branches — sort boundaries cleanly to avoid this.
- Every value has a truthiness — `0`, `""`, `[]`, `{}`, `()`, `None`, and `False` are falsy; everything else is truthy.
- The ternary expression (`x if cond else y`) is a one-line if-else — great for simple choices, avoid nesting it deeply.
- `match`-`case` (Python 3.10+) is a cleaner alternative to long `elif` chains that compare one variable against many fixed values.
