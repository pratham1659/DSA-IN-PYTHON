## 🔄 Type Conversion (Typecasting)

### 🤔 What is it?

**Type Conversion** (or typecasting) means converting a value from one data type to another — like turning a number into text, or text into a number — using functions like `str()`, `int()`, `float()`.

> 🧸 Think of it like currency exchange. A number and a string are different "currencies" — Python won't let you mix them directly (`10 + "Rohan"` fails), but once you convert one to match the other, they work together fine.

### ❓ Why do we need it?

`input()` always gives you a string, but often you need a number to do math. Without converting it first, Python throws a `TypeError` because it doesn't know how to combine incompatible types.

### 💻 Technical Example

Below, `x + y` fails because `x` is an integer and `y` is a string — Python can't automatically combine a number and text. Converting `x` to a string with `str(x)` fixes it, because now both sides are text and `+` just joins them together.

```python
y = "Rohan"
x = 10

x + y
```

```python
str(x) + y
```

### 🌍 Real-Life Example

A classic use of typecasting: asking for someone's birth year (text from `input()`) and subtracting it from the current year (a number) to calculate their age. This only works once the birth year is converted to an integer.

```python
# This is Typecasting
birth_year = input("write your birth Year : ")
age = 2022 - int(birth_year)
print(age)
```

More typecasting examples: `int()` → `float()`, `list()` → `set()` (which also removes duplicates!), and converting strings to numbers to do math before turning the result back into a string.

```python
a=2
f=float(a)
print("Integer a =", a )
print("float a =", f )
l = [1,3,3,4,5]
s= set(l)
print(s)
print(l)
var1 = "20" #String integer Typecasting
var2 = "30" #String Typecasting
print(10 * str(int(var1) + int(var2)), " ")
```

```python
weight_lbs = input("weight in LBS ")
weight_kg = int(weight_lbs) * 0.45
print(weight_kg)
```

> 💡 **Quotes trick:** Python lets you use single `'...'` or double `"..."` quotes for strings. If your text contains an apostrophe (like `Python's`), wrap it in double quotes. If it contains double quotes, wrap it in single quotes. This avoids "escaping" characters.

```python
course1 = "Python's for Beginners"
course2 = 'Python for "Beginners"'
print(course1, "\n", course2)
```

## ⌨ Escape Sequence Characters

### 🤔 What is it?

An **escape sequence** is a special character combo starting with a backslash `\` that represents something you can't type directly in a string — like a new line or a tab space.

> 🧸 Think of `\` as a "special instruction" flag. `\n` doesn't print a literal `n` — it tells Python "insert a line break here instead."

### 📚 Important Terms

| Escape Sequence | Simple Meaning |
|------------------|-----------------|
| `\n` | New line (move to the next line) |
| `\t` | Tab (adds horizontal spacing) |
| `\\` | A literal backslash character |
| `\'` | A literal single quote inside a `'...'` string |
| `\"` | A literal double quote inside a `"..."` string |

> ⚠ Note: a lone `\` followed by a space or letter that isn't a recognized escape sequence (like `\ P`) is usually just a mistake — Python will print it close to literally, which is confusing. Use `\\` if you actually want a backslash character.

```python
print("hello \n pratham")   # \n -> new line
print("hello \t Pratham Good Boy")  # \t -> tab space
print("This is a backslash: \\")   # \\ -> literal backslash
```

## 🐍 Indentation in Python

### 🤔 What is it?

Python uses **indentation** (spaces at the start of a line) instead of curly braces `{}` to show which lines belong together in a block of code.

> 🧸 Think of indentation like paragraph structure in an essay — it visually shows what belongs together. Python is strict about this: mixing it up breaks the "sentence."

### ❓ Why do we need it?

Other languages (like Java or C++) use `{ }` to group code. Python instead relies on **consistent spacing** to know where a block starts and ends. This keeps code clean and readable, but it also means incorrect spacing causes real errors — not just messy formatting.

### ⚠ What is `IndentationError`?

If a line has spacing that doesn't match the structure Python expects (e.g., an unexpected extra space with no reason), Python throws an `IndentationError` and refuses to run the code.

## 💬 Comments in Python

### 🤔 What is it?

A **comment** is a note in your code that Python ignores when running — it's there purely for humans to read.

> 🧸 Think of comments like sticky notes on a recipe card — they don't change the dish, but they help the cook (or a future you) remember why a step is done a certain way.

### 📚 Important Terms

| Type | Simple Meaning | Example |
|------|----------------|----------|
| Single-line comment | Starts with `#`, ends at the end of that line | `# this is a comment` |
| Multi-line comment | Multiple `#` lines, OR a triple-quoted string used as a comment | See below |
| Docstring | A triple-quoted string used to document a function/class — NOT ignored like a comment, it's actually stored as real data | `"""This function adds two numbers"""` |

### Single-Line Comment

```python
# Adding 2 variables and printing the result in a new variable
a = 2
b = 3
c = a + b
print(c)
```

### Multi-Line Comment

Python doesn't have a dedicated multi-line comment symbol — most people either stack multiple `#` lines, or use a triple-quoted string (`""" ... """`) as a throwaway value that Python evaluates but discards.

```python
var1 = "Hello print"
var2 = "43"
var22 = "34"
var3 = 34.6
var4 = 34424

print(type(var3))               # float
print(var3 + var4)              # float + int works fine -> 34458.6
print(100 * int(var22) + var3)  # typecasting var22 to int before math -> 3434.6
print(100 * str(int(var2) + int(var22)))  # int math first, then cast to str and repeat it 100 times
```

## 📄 Docstring Comment

### 🤔 What is it?

A **docstring** looks like a comment (triple-quoted text), but it's actually different: Python does NOT throw it away. It's stored as real data — usually as the very first line inside a function, class, or module, describing what that piece of code does.

> 🧸 Think of a regular comment as a sticky note the chef throws away after reading, and a docstring as the description printed on the menu — visible and accessible to anyone using the dish (or function).

> ❌ Misconception: "Docstrings are just comments with fancy quotes."
> ✅ Correct: Docstrings are real string objects Python keeps around — you can access them later with `help(function_name)` or `function_name.__doc__`.

## 🏗 Classes and Objects

### 🤔 What is it?

A **class** is a blueprint for creating objects. An **object** is a specific "thing" built from that blueprint, with its own data.

> 🧸 Think of a class like a cookie cutter, and objects like the cookies made from it. The cutter (class) defines the *shape* — every cookie (object) made from it shares that shape, but each cookie can still be decorated differently.

### ❓ Why do we need it?

Classes let us group related data (**attributes**) and behavior (**methods**) together, so we can create many similar objects without repeating code. This is the foundation of **Object-Oriented Programming (OOP)**.

### 📚 Important Terms

| Term | Simple Meaning | Example |
|------|----------------|----------|
| Class | The blueprint/template | `class Car:` |
| Object | A specific instance made from the class | `BMW = Car()` |
| Attribute | A piece of data stored on the object | `attr1 = "Petrol"` |
| Method | A function defined inside a class | `def start(self):` |
| `self` | Refers to "this specific object" inside a method | `self.attr1` |
| `__init__` | A special method that runs automatically when an object is created (the "constructor") | `def __init__(self, name):` |

### 🌍 Real-Life Example

Think of `Car` as the blueprint for "what a car is" (has fuel type, horsepower, can start). `BMW = Car()` builds one actual car from that blueprint. You could make another car, `Tesla = Car()`, and it would have the same *structure*, but be a completely separate object.

```python
# Python3 program to
# demonstrate instantiating
# a class
class Car:
    # A simple class
    # attribute
    attr1 = "Petrol"
    attr2 = "750 HP"
    # A sample method 
    def start(self):
        print("Engine Started : Engine Type ", self.attr1)
        print("Ready to GO : Horse Power ", self.attr2)
# Driver code
# Object instantiation
BMW = Car()
# Accessing class attributes
# and method through objects
print(BMW.attr1)
BMW.start()
```

### The `__init__` Constructor

Most real classes use `__init__` to set up an object's data automatically when it's created, instead of hardcoding attributes like `Car` did above. `self.name = name` stores the value passed in on the object itself, so every `Person` object can have a different name.

> 📌 **Easy trick:** `self` always refers to "the object being worked on right now." When you write `self.name`, think "this specific person's name," not some global name.

```python
class Person:
    # init method or constructor 
    def __init__(self, name):
        self.name = name
    # Sample Method 
    def say_hi(self):
        print('Hello, my name is', self.name)
p = Person('Robert')
p.say_hi()
```

| Ignored completely by Python | Stored as real data (`__doc__`) |
| Only for human readers | Can be accessed programmatically (`help()`) |
| Any position in code | Conventionally the first line inside a function/class/module |
