<a id="what-is-it"></a>
## 🤔 What is it?

**File handling** is how a Python program reads data FROM files on disk, and writes data TO files on disk — so information survives after the program stops running. Everything you've worked with so far (variables, lists, dictionaries) lives only in memory, and disappears the instant the program ends. A file is memory's permanent cousin.

> 🧸 Think of your computer's RAM (where variables live while a program runs) like a whiteboard — great for working things out, but wiped clean the moment you leave the room. A file on disk is like writing it down in a notebook instead — it's still there tomorrow, next week, even after the computer restarts.

---

<a id="why-do-we-need-it"></a>
## ❓ Why do we need it?

- Programs need to **remember things between runs** — user settings, saved progress, logs, a list of customers — none of that should vanish just because the program closed.
- Real data usually comes FROM somewhere else, too — a CSV of sales figures, a JSON config file, a text log from another system — and Python needs a standard way to read it in.
- Sharing data between programs (or between a program and a human) almost always happens through files — one program writes a report, another reads it, a person opens it in a text editor.

---

<a id="key-idea"></a>
## 🧠 Key Idea

- `open(path, mode)` returns a **file object** you can read from or write to — but you must eventually `close()` it, or use `with`, which closes it for you automatically.
- The **mode** (`"r"`, `"w"`, `"a"`, ...) controls what you're allowed to do (read-only? write? both?) and what happens to any EXISTING content in the file.
- Reading and writing both have a "current position" inside the file — you can move it manually with `seek()`, or just let it advance naturally as you read/write.
- `with open(...) as f:` is the idiomatic, safe way to work with files — the file gets closed automatically even if an exception happens partway through.
- The `csv` and `json` modules build on top of plain file handling to read/write STRUCTURED data (rows and columns, or nested Python objects) without you having to parse the raw text yourself.

<a id="important-terms"></a>
## 📚 Important Terms

| Term | Simple Meaning | Example |
|------|----------------|----------|
| File object | What `open()` returns — your handle to read/write with | `f = open("data.txt", "r")` |
| Mode | Controls read/write permission and truncate-vs-append behavior | `"r"`, `"w"`, `"a"`, `"r+"` |
| Context Manager | An object usable with `with`, which auto-cleans-up (like closing a file) | `with open(...) as f:` |
| File Pointer | The current read/write position inside the file | Moved by `seek()`, reported by `tell()` |
| Buffering | Python temporarily holds written data before it's actually saved to disk | Flushed on `close()` |
| Encoding | The character set used to convert bytes ↔ text (e.g. UTF-8) | `open(path, encoding="utf-8")` |
| Binary Mode | Reads/writes raw bytes (`bytes`), not text (`str`) | `open(path, "rb")` |
| Serialization | Converting a Python object into a saveable format (text, bytes) | `json.dump(data, f)` |

<a id="how-it-works"></a>
## 🔄 How it Works

```mermaid
flowchart TD
Open["open(path, mode)"] --> Object["File object returned"]
Object -->|read mode| Read["f.read() / f.readline() / for line in f"]
Object -->|write mode| Write["f.write() / f.writelines()"]
Read --> Close["f.close() -- or automatic, via 'with'"]
Write --> Close
Close --> Done["Data safely saved / connection released"]
```

Read it like this: `open()` gives you a file object tied to a specific mode. What you can DO with it (read or write) depends on that mode. Either way, the file MUST eventually be closed — manually, or automatically via `with` — to make sure everything you wrote is actually saved and the underlying system resource is released.

## 🌍 Real-Life Example

Think of a **library book** 📖. `open()` is like checking the book out — now you have access to it. Reading is flipping through its pages; writing is like using a checkout slip to add your own notes (if it's the kind of book that allows that). `close()` is returning the book to the library — until you do, the library can't be sure it's safe to lend it to someone else, or that your notes are properly filed. The `with` statement is like an automatic return system: even if you get distracted (an exception happens) while reading, the book still gets returned — you can't accidentally walk off with it forever.

<a id="opening-closing-files"></a>
## 📂 Opening and Closing Files (the Manual Way)

`open(path, mode)` returns a **file object**. Once you're done, `close()` releases it — this matters more than it looks: until a file is closed, Python may not have actually flushed everything you wrote to disk yet, and the operating system may treat it as still "in use."

All examples in this notebook read and write files inside a `Data/` folder next to this notebook.

````python
f = open("Data/hello.txt", "w")
f.write("Hello, this is a test file.\n")
f.close()   # without this, the write above might not actually be saved to disk yet

f = open("Data/hello.txt", "r")
content = f.read()
f.close()
print(content)
````

````python
f = open("Data/hello.txt", "r")
print("Is file closed?", f.closed)   # False -- still open
f.close()
print("Is file closed?", f.closed)   # True -- now it is
````

**Note:** `f.closed` is `True`/`False`, telling you whether a file object is still usable. The real problem with the manual approach above: if an exception happens BETWEEN `open()` and `close()`, `close()` never runs — the file is left open, and any data you were writing might never actually get saved. That's exactly the problem `with` solves next.

---

<a id="with-statement"></a>
## 🛡️ The `with` Statement — Automatic, Exception-Safe Closing

`with open(path, mode) as f:` opens the file, hands you `f` to work with inside the indented block, and **guarantees** `close()` runs the moment that block ends — success OR failure.

````python
with open("Data/hello.txt", "r") as f:
    content = f.read()

print(content)
print("File closed automatically?", f.closed)   # True -- even though we never called f.close() ourselves
````

````python
# Proving it's exception-safe -- close() still runs even when something goes wrong INSIDE the block
try:
    with open("Data/hello.txt", "r") as f:
        data = f.read()
        raise ValueError("Something went wrong while processing")
except ValueError as e:
    print(f"Caught: {e}")

print("File closed even though an exception happened:", f.closed)
````

**Note:** The `ValueError` happens INSIDE the `with` block, has nothing to do with the file itself, and is caught by an OUTER `try`/`except` — yet `f.closed` is still `True` afterward. `with` is built on the same idea as `try`/`finally` ([see the Exception Handling notebook](16_ExceptionHandling.ipynb)): `f.close()` is guaranteed to run when the block exits, no matter how it exits. This is why `with` is the standard, idiomatic way to work with files in Python — manual `open()`/`close()` is shown above mainly to explain what's happening under the hood.

> ❌ Misconception: `f.close()` is optional if you're "careful" and remember to call it.
> ✅ Correct: forgetting to close a file (or having an exception skip your `close()` call) can leave data unsaved or the file locked by the OS. `with` removes the need to remember at all.

<a id="file-modes"></a>
## 🔑 File Modes — `r`, `w`, `a`, `x`, `r+`, `w+`, `a+`

The **mode** you pass to `open()` decides what you're allowed to do, and — critically — what happens to any content the file already has.

| Mode | Meaning | If file exists | If file doesn't exist |
|---|---|---|---|
| `"r"` | Read only | Opens it | `FileNotFoundError` |
| `"w"` | Write only | **Erases everything**, starts blank | Creates it |
| `"a"` | Append only | Keeps content, writes AFTER it | Creates it |
| `"x"` | Exclusive create | `FileExistsError` | Creates it |
| `"r+"` | Read + write | Opens it, no erasing | `FileNotFoundError` |
| `"w+"` | Write + read | **Erases everything**, starts blank | Creates it |
| `"a+"` | Append + read | Keeps content, writes AFTER it | Creates it |

### `"w"` — Truncates Existing Content

````python
with open("Data/modes.txt", "w") as f:
    f.write("First write\n")

with open("Data/modes.txt", "w") as f:   # 'w' AGAIN -- wipes out "First write" completely
    f.write("Second write\n")

with open("Data/modes.txt", "r") as f:
    print(repr(f.read()))   # only "Second write" survives
````

**Bug to watch for:** opening a file with `"w"` when you actually meant to keep old data is one of the most common file-handling mistakes — "First write" is gone the instant the second `open(..., "w")` runs, even before anything new is written.

### `"a"` — Append, Never Truncates

````python
with open("Data/modes.txt", "a") as f:
    f.write("Appended line\n")   # ADDS to the end -- "Second write" is untouched

with open("Data/modes.txt", "r") as f:
    print(f.read())
````

### `"x"` — Exclusive Creation (Fails if the File Already Exists)

````python
import os

if os.path.exists("Data/brand_new.txt"):
    os.remove("Data/brand_new.txt")   # start clean for this demo

with open("Data/brand_new.txt", "x") as f:
    f.write("Created fresh\n")

try:
    with open("Data/brand_new.txt", "x") as f:   # 'x' AGAIN -- the file now already exists
        f.write("Trying again")
except FileExistsError as e:
    print(f"FileExistsError: {e}")
````

**Note:** `"x"` is the safest mode when you genuinely never want to overwrite an existing file by accident — unlike `"w"`, which would happily erase it, `"x"` refuses outright and raises `FileExistsError`.

### `"r+"`, `"w+"`, `"a+"` — Combined Read + Write Modes

These three all allow BOTH reading and writing, but differ in the same way `r`/`w`/`a` do about existing content.

````python
# 'r+' -- read AND write, but does NOT truncate; starts at position 0
with open("Data/modes.txt", "r+") as f:
    first_line = f.readline()
    print("First line before overwrite:", repr(first_line))
    f.seek(0)          # jump back to the start
    f.write("XXXX")    # overwrite the first 4 characters only

with open("Data/modes.txt", "r") as f:
    print(f.read())
````

````python
# 'w+' -- write AND read, but TRUNCATES first, just like plain 'w'
with open("Data/modes.txt", "w+") as f:
    f.write("Brand new content\n")
    f.seek(0)   # must seek back to 0 before reading what you just wrote
    print(f.read())
````

````python
# 'a+' -- append AND read; does NOT truncate, new writes go at the END
with open("Data/modes.txt", "a+") as f:
    f.write("More appended text\n")
    f.seek(0)   # move back to the start so we can read everything, old AND new
    print(f.read())
````

**Note:** All three read+write modes need `f.seek(0)` before reading back what you just wrote — writing (or appending) moves the file's position FORWARD, so without seeking back, `f.read()` would just find nothing left to read (you'd already be at the end). `"r+"` vs `"w+"` vs `"a+"` is really just the SAME `r`/`w`/`a` truncate-vs-append rule from before, now combined with read access too.

<a id="binary-mode"></a>
## 💾 Binary Mode

Adding `"b"` to any mode (`"rb"`, `"wb"`, `"ab"`, ...) switches from TEXT (`str`) to raw BYTES (`bytes`) — needed for non-text files (images, audio, any binary format), or whenever you want exact byte-level control.

````python
with open("Data/data.bin", "wb") as f:
    f.write(bytes([72, 101, 108, 108, 111]))   # raw byte values -- spells out "Hello" in ASCII

with open("Data/data.bin", "rb") as f:
    raw = f.read()

print(raw)                     # b'Hello' -- a bytes object, not a str
print(raw.decode("utf-8"))     # decode it back into readable text
````

**Note:** `write(bytes([...]))` writes RAW bytes with no text encoding applied at all — that's the point of binary mode. Reading it back with `"rb"` gives you a `bytes` object (`b'Hello'`), and `.decode("utf-8")` is what turns those bytes back into a readable Python string. Trying to `write()` a plain `str` in binary mode (or a `bytes` object in text mode) raises a `TypeError` — the two modes are not interchangeable.

<a id="reading-files"></a>
## 📖 Reading Files — `read()`, `readline()`, `readlines()`, Iteration

Python gives you four different ways to read a file, each suited to a different situation.

````python
with open("Data/poem.txt", "w") as f:
    f.write("Roses are red\nViolets are blue\nPython is fun\nAnd so are you\n")

# read() -- the ENTIRE file, as one string
with open("Data/poem.txt", "r") as f:
    whole = f.read()
print(repr(whole))
````

````python
# read(n) -- only the first n characters, not the whole file
with open("Data/poem.txt", "r") as f:
    partial = f.read(13)
print(repr(partial))
````

````python
# readline() -- ONE line at a time; each call continues where the last one left off
with open("Data/poem.txt", "r") as f:
    line1 = f.readline()
    line2 = f.readline()

print(repr(line1))
print(repr(line2))
````

````python
# readlines() -- ALL lines at once, as a LIST of strings
with open("Data/poem.txt", "r") as f:
    lines = f.readlines()
print(lines)
````

````python
# Iterating directly over the file object -- the most Pythonic, most memory-efficient option
with open("Data/poem.txt", "r") as f:
    for line in f:
        print(line.strip())   # .strip() removes the trailing \n each line already has
````

**Note:** `read()` and `readlines()` both load the ENTIRE file into memory at once — fine for a small file, but potentially a problem for a multi-gigabyte one. Iterating directly with `for line in f:` reads and processes ONE line at a time, never holding the whole file in memory — this is the recommended approach for large files. `readline()` gives you the same one-line-at-a-time behavior, but manually, which is handy when you need to read just a couple of specific lines rather than looping through all of them.

> ❌ Misconception: `readlines()` and iterating with `for line in f:` are basically the same thing.
> ✅ Correct: `readlines()` builds the FULL list in memory before you get anything back; `for line in f:` never builds a list at all — it hands you one line, then the next, without ever holding the whole file at once. For huge files, this difference genuinely matters.

<a id="writing-files"></a>
## ✍️ Writing Files — `write()`, `writelines()`

`write()` writes a string. `writelines()` writes a LIST of strings. Neither one adds a newline for you automatically — that's a common early mistake.

````python
# The bug: no \n between writes -- they run together into one line
with open("Data/notes.txt", "w") as f:
    f.write("Line one")
    f.write("Line two")

with open("Data/notes.txt", "r") as f:
    print(repr(f.read()))
````

````python
# Fixed -- add \n explicitly at the end of each write
with open("Data/notes.txt", "w") as f:
    f.write("Line one\n")
    f.write("Line two\n")

with open("Data/notes.txt", "r") as f:
    print(f.read())
````

````python
# writelines() -- writes a LIST of strings, but STILL doesn't add newlines for you
lines = ["Apple\n", "Banana\n", "Cherry\n"]   # each string includes its own \n

with open("Data/fruits.txt", "w") as f:
    f.writelines(lines)

with open("Data/fruits.txt", "r") as f:
    print(f.read())
````

**Note:** `writelines(lines)` is really just a shortcut for looping and calling `write()` on each item — it's NOT the write-equivalent of `readlines()` adding structure back in. Each string in `lines` above already ends in `\n`; if it didn't, `writelines()` would join them together with nothing in between, exactly like the `write()` bug above.

> ❌ Misconception: `write()`/`writelines()` automatically add a newline after each piece of text, like `print()` does.
> ✅ Correct: `print()` adds `\n` for you by default — `write()` and `writelines()` do NOT. You must include `\n` yourself wherever you want a line break.

<a id="pointer-control"></a>
## 🧭 `seek()` and `tell()`

Every open file has a **current position** — where the NEXT read or write will happen. `tell()` reports it; `seek(position)` jumps directly to it.

````python
with open("Data/pointer.txt", "w") as f:
    f.write("ABCDEFGHIJ")

with open("Data/pointer.txt", "r") as f:
    print("Position at start:", f.tell())          # 0 -- haven't read anything yet

    chunk = f.read(3)
    print("Read:", chunk)
    print("Position after reading 3 chars:", f.tell())   # 3 -- moved forward by exactly what was read

    f.seek(0)   # jump back to the very beginning
    print("Position after seek(0):", f.tell())
    print("Read again from start:", f.read(5))

    f.seek(5)   # jump DIRECTLY to position 5, skipping everything before it
    print("Read from position 5:", f.read())
````

**Note:** Every `read()` call moves the position forward by exactly how much it read — `tell()` is just checking "where am I right now?" `seek(n)` jumps DIRECTLY to byte/character position `n`, without reading anything in between — this is what made `f.seek(0)` necessary earlier in the `"r+"`/`"w+"`/`"a+"` examples, and it's also how you'd implement something like "read only the last 100 bytes of a huge log file" without reading the whole thing first.

<a id="paths-and-existence"></a>
## 🗺️ Checking Paths — `os.path` and `pathlib`

Before opening a file, it's often useful to check whether it exists, or inspect its path — Python offers two ways to do this: the older `os.path` module, and the more modern, object-oriented `pathlib`.

````python
import os

print(os.path.exists("Data/poem.txt"))     # True -- we created this earlier
print(os.path.exists("Data/ghost.txt"))     # False -- never created
print(os.path.isfile("Data/poem.txt"))       # True -- it's a file
print(os.path.isdir("Data/poem.txt"))         # False -- it's not a directory
print(os.path.join("Data", "poem.txt"))         # builds a path correctly for any OS
````

````python
from pathlib import Path

p = Path("Data/poem.txt")
print(p.exists())    # True
print(p.name)         # "poem.txt" -- just the filename
print(p.suffix)        # ".txt" -- just the extension
print(p.stem)            # "poem" -- filename WITHOUT the extension
````

**Note:** `os.path.join("Data", "poem.txt")` matters more than it looks — it correctly uses `/` on Mac/Linux and `\` on Windows, so your code doesn't break when it runs on a different operating system than the one you wrote it on. `pathlib.Path` offers the same information as convenient PROPERTIES (`p.name`, `p.suffix`, `p.stem`) instead of separate function calls — most new Python code prefers `pathlib` over `os.path` for exactly this reason.

### More `os.path` Examples — Splitting, Sizes, and Cleaning Up Messy Paths

`os.path` has quite a few more functions beyond `exists()`/`isfile()`/`join()` — for pulling a path apart into pieces, checking file metadata, and normalizing paths that have redundant bits in them.

````python
path = "Data/poem.txt"

print(os.path.dirname(path))     # "Data" -- just the directory part
print(os.path.basename(path))    # "poem.txt" -- just the filename part
print(os.path.split(path))       # ("Data", "poem.txt") -- both at once, as a tuple
print(os.path.splitext(path))    # ("Data/poem", ".txt") -- path WITHOUT extension, and the extension
````

````python
print(os.path.getsize(path))    # size of the file, in bytes
print(os.path.isfile(path))      # True -- it's a file
print(os.path.isdir(path))        # False -- it's not a directory
````

````python
import datetime

mtime = os.path.getmtime(path)   # last-modified time, as a raw timestamp (float)
print(mtime)
print(datetime.datetime.fromtimestamp(mtime).strftime("%Y-%m-%d %H:%M:%S"))   # converted to something readable
````

````python
# normpath -- cleans up "." (current dir), ".." (parent dir), and redundant slashes
messy = "Data/./subfolder/../poem.txt"
print(os.path.normpath(messy))    # simplifies down to "Data/poem.txt"
````

````python
# relpath -- the OPPOSITE of abspath: express a path RELATIVE TO some other folder
full_path = os.path.abspath(path)
project_root = os.getcwd()
print(os.path.relpath(full_path, project_root))   # back to "Data/poem.txt", relative to project_root
````

**Note:** `dirname`/`basename`/`split`/`splitext` all just do STRING manipulation on the path — none of them touch the actual filesystem or check whether the file exists. `getsize()`/`getmtime()`, by contrast, DO touch the filesystem — they raise `FileNotFoundError` if the path doesn't actually exist, since there's no size or modification time to report for a file that isn't there. `normpath()` is purely textual cleanup (it doesn't check existence either), while `relpath()` is the mirror image of `abspath()` — turning a full path back into a short one, relative to whatever base folder you choose.

### Absolute vs Relative Paths

Every path is either **relative** (interpreted starting from wherever the program is currently running) or **absolute** (the full path from the filesystem's root — unambiguous no matter where the program runs from).

````python
# Relative path -- interpreted relative to the CURRENT WORKING DIRECTORY
relative_path = "Data/hello.txt"
print("Relative path:", relative_path)
print("Current working directory:", os.getcwd())

# Absolute path -- the FULL path from the filesystem root, unambiguous
absolute_path = os.path.abspath(relative_path)
print("Absolute path:", absolute_path)

print("Is relative path absolute?", os.path.isabs(relative_path))    # False
print("Is absolute path absolute?", os.path.isabs(absolute_path))    # True
````

````python
# Both point to the SAME file -- opening either one reads identical content
with open(relative_path, "r") as f:
    print(f.read())

with open(absolute_path, "r") as f:
    print(f.read())
````

````python
# pathlib equivalent
from pathlib import Path

p = Path("Data/hello.txt")
print(p.resolve())                  # the absolute path, computed via pathlib
print(p.is_absolute())              # False -- p itself is still a relative Path
print(p.resolve().is_absolute())    # True -- the RESOLVED version is absolute
````

**Note:** `os.getcwd()` reports the **current working directory** — the folder Python considers "here" right now, which is what every relative path is measured FROM. `os.path.abspath()` (or `Path.resolve()`) converts a relative path into its absolute equivalent by combining it with the current working directory. Both `relative_path` and `absolute_path` point to the exact same file here — the difference only matters when the CURRENT WORKING DIRECTORY changes, shown next.

### Why It Matters — Relative Paths Can Break, Absolute Paths Can't

````python
original_cwd = os.getcwd()
absolute_path = os.path.abspath("Data/hello.txt")   # computed BEFORE changing directory

os.chdir("/tmp")   # simulate running this code from a completely different directory

try:
    with open("Data/hello.txt", "r") as f:   # relative path -- now looks for /tmp/Data/hello.txt, which doesn't exist!
        print(f.read())
except FileNotFoundError as e:
    print(f"FileNotFoundError (relative path broke): {e}")

with open(absolute_path, "r") as f:   # absolute path -- still finds the SAME file, no matter the current directory
    print("Absolute path still works:", f.read())

os.chdir(original_cwd)   # restore -- so every later cell in this notebook keeps working correctly
````

**Note:** `os.chdir("/tmp")` changes the current working directory — the exact same relative path (`"Data/hello.txt"`) that worked perfectly before now points somewhere completely different (`/tmp/Data/hello.txt`), which doesn't exist, so it raises `FileNotFoundError`. The absolute path, computed BEFORE the directory change, is unaffected — it always points to the one specific file, regardless of what "here" currently means. `os.chdir(original_cwd)` at the end restores things — this matters because every earlier example in this notebook uses relative `"Data/..."` paths, and they'd all break for the rest of the notebook if the working directory were left changed.

> ❌ Misconception: a relative path and an absolute path to the same file are basically interchangeable — just two ways of writing the same thing.
> ✅ Correct: they behave identically ONLY as long as the current working directory stays the same. A relative path is really shorthand for "wherever we currently are, plus this" — change "wherever we currently are," and the relative path now means something else entirely. This is why long-running programs, scripts run from cron/schedulers, or packaged applications often convert paths to absolute early on, to avoid depending on whatever directory happened to launch them.

<a id="missing-file-handling"></a>
## 🚫 Handling a Missing File — EAFP vs LBYL

There are two competing philosophies for dealing with a file that might not exist: check first ("Look Before You Leap"), or just try it and handle the failure ("Easier to Ask Forgiveness than Permission").

````python
# EAFP -- "Easier to Ask Forgiveness than Permission" -- try first, handle the exception (the Pythonic way)
try:
    with open("Data/ghost.txt", "r") as f:
        content = f.read()
except FileNotFoundError:
    print("File doesn't exist -- using default content instead")
    content = ""
````

````python
# LBYL -- "Look Before You Leap" -- check first, then act
if os.path.exists("Data/ghost.txt"):
    with open("Data/ghost.txt", "r") as f:
        content = f.read()
else:
    print("File doesn't exist -- using default content instead")
    content = ""
````

**Note:** Both produce the same result here, but EAFP is generally preferred in Python for one subtle reason: **the race condition**. Between `os.path.exists(...)` returning `True` and the very next line actually opening the file, something else (another program, another thread) could delete it — so `open()` could STILL raise `FileNotFoundError` even after the check said the file was there. `try`/`except` doesn't have this gap at all, because it just attempts the real operation directly. This connects straight back to the [Exception Handling notebook](16_ExceptionHandling.ipynb)'s misconception about `try`/`except` — catching `FileNotFoundError` here isn't a fallback for lazy code, it's the genuinely safer approach.

<a id="file-exceptions"></a>
## 🧯 More File-Related Exceptions

`FileNotFoundError` isn't the only thing that can go wrong when working with files. A few others come up often enough to be worth knowing by name: `IsADirectoryError`, `PermissionError`, and `UnicodeDecodeError`.

### `IsADirectoryError` — Opening a Folder as if It Were a File

````python
os.makedirs("Data/a_folder", exist_ok=True)

try:
    with open("Data/a_folder", "r") as f:   # "a_folder" is a DIRECTORY, not a file
        f.read()
except IsADirectoryError as e:
    print(f"IsADirectoryError: {e}")
````

**Note:** `open()` on a directory path fails with `IsADirectoryError`, not `FileNotFoundError` — the path DOES exist, it's just the wrong KIND of thing. This is a common bug when a path comes from user input or a config file and might point to a folder by mistake.

### `PermissionError` — No Permission to Read or Write

````python
with open("Data/locked.txt", "w") as f:
    f.write("secret")

os.chmod("Data/locked.txt", 0o444)   # make it READ-ONLY (Mac/Linux permission bits)

try:
    with open("Data/locked.txt", "w") as f:   # trying to WRITE to a read-only file
        f.write("trying to overwrite")
except PermissionError as e:
    print(f"PermissionError: {e}")

os.chmod("Data/locked.txt", 0o644)   # restore normal permissions so later cleanup works
````

**Note:** `os.chmod(path, 0o444)` sets read-only permissions (on Mac/Linux) — writing to it afterward raises `PermissionError` before Python even gets a chance to touch the content. This can also happen for files owned by another user, files inside a protected system folder, or a file another program currently has locked. `os.chmod(..., 0o644)` afterward restores normal permissions — this line matters because leaving a file read-only could break later cells (or later runs) that try to write to it again.

### `UnicodeDecodeError` — Reading Binary Data as if It Were Text

````python
with open("Data/binarydata.bin", "wb") as f:
    f.write(bytes([0xff, 0xfe, 0x00, 0x80]))   # NOT valid UTF-8 text -- just raw bytes

try:
    with open("Data/binarydata.bin", "r", encoding="utf-8") as f:   # opened in TEXT mode, wrong choice
        content = f.read()
except UnicodeDecodeError as e:
    print(f"UnicodeDecodeError: {e}")
````

**Note:** Text mode (`"r"`) always tries to DECODE the raw bytes into a string using an encoding (UTF-8 by default) — `0xff` isn't a valid starting byte in UTF-8, so decoding fails with `UnicodeDecodeError`. This is exactly why genuinely binary files (images, the `.bin` file from the Binary Mode section above) must be opened with `"rb"`, not `"r"` — text mode assumes the content IS text, and breaks the moment that assumption is wrong.

### Combining Them — One Function, Several File-Specific `except` Blocks

This mirrors the [Exception Handling notebook](16_ExceptionHandling.ipynb)'s "one `try`, several stacked `except` blocks" pattern — but applied specifically to file operations.

````python
def safe_read(path):
    try:
        with open(path, "r") as f:
            return f.read()
    except FileNotFoundError:
        print(f"'{path}' does not exist")
    except IsADirectoryError:
        print(f"'{path}' is a directory, not a file")
    except PermissionError:
        print(f"No permission to read '{path}'")
    except UnicodeDecodeError:
        print(f"'{path}' isn't valid text (maybe it's a binary file?)")
    return None


print(safe_read("Data/ghost.txt"))         # FileNotFoundError
print(safe_read("Data/a_folder"))           # IsADirectoryError
print(safe_read("Data/binarydata.bin"))      # UnicodeDecodeError
print(safe_read("Data/hello.txt"))            # succeeds normally
````

**Note:** `safe_read()` gives every file-related problem its OWN clear message, instead of one generic "something went wrong" — exactly the benefit of catching specific exception types instead of a bare `except:`. `return None` at the end (not inside any `except`) is the fallback for every failure case; the successful case returns straight from inside the `try`, so it never reaches that final line at all.

### `try` / `except` / `else` / `finally` — a Full File-Reading Pattern

````python
def read_config(path):
    try:
        with open(path, "r") as f:
            data = f.read()
    except FileNotFoundError:
        print(f"Config file {path} missing -- using defaults")
        data = "{}"
    else:
        print(f"Loaded config from {path}")
    finally:
        print("Finished attempting to load config")   # ALWAYS runs, success or failure
    return data


print(read_config("Data/settings.json"))   # doesn't exist yet -- falls back to defaults
print("---")

with open("Data/settings.json", "w") as f:
    f.write('{"theme": "dark"}')

print(read_config("Data/settings.json"))   # exists now -- loads it for real
````

**Note:** `else` here runs ONLY when the file was actually found and read successfully — printing "Loaded config from..." directly inside `try` instead would ALSO run if something later in that block failed for an unrelated reason, which `else` avoids (same lesson as the [Exception Handling notebook](16_ExceptionHandling.ipynb)'s `else` section). `finally` prints "Finished attempting to load config" both times — whether the file was missing or not — making it the right place for something like closing a database connection or writing an audit log entry, regardless of outcome.

<a id="csv-files"></a>
## 📊 Working with CSV Files

CSV ("Comma-Separated Values") is a plain-text format for tabular data — rows and columns, like a simple spreadsheet. The `csv` module handles the comma-splitting, quoting, and edge cases for you, instead of you writing `line.split(",")` yourself (which breaks the moment a value contains a comma).

````python
import csv

# Writing a CSV file with csv.writer
with open("Data/students.csv", "w", newline="") as f:   # newline="" avoids extra blank lines on Windows
    writer = csv.writer(f)
    writer.writerow(["name", "age", "grade"])   # header row
    writer.writerow(["Pratham", 23, "A"])
    writer.writerow(["Isha", 21, "B"])

with open("Data/students.csv", "r") as f:
    print(f.read())
````

````python
# Reading a CSV file with csv.reader -- each row comes back as a LIST of strings
with open("Data/students.csv", "r") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
````

````python
# DictWriter / DictReader -- work with DICTIONARIES instead of plain lists (by column name)
with open("Data/students2.csv", "w", newline="") as f:
    fieldnames = ["name", "age", "grade"]
    writer = csv.DictWriter(f, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerow({"name": "Pratham", "age": 23, "grade": "A"})
    writer.writerow({"name": "Isha", "age": 21, "grade": "B"})

with open("Data/students2.csv", "r") as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(dict(row))
````

**Note:** `csv.reader` gives you `["Pratham", "23", "A"]` — notice `23` came back as the STRING `"23"`, not the number `23`. CSV files are plain text; the `csv` module doesn't know or guess that a column is numeric, so converting types back (`int(row[1])`) is on you. `DictReader` is often more convenient than `reader` for wider tables, since you access columns by NAME (`row["age"]`) instead of remembering they're the 2nd item in a list.

<a id="json-files"></a>
## 🗂️ Working with JSON Files

JSON ("JavaScript Object Notation") maps almost perfectly onto Python's own dicts, lists, strings, numbers, and booleans — making it the standard format for structured data (configs, API responses, saved app state). The `json` module converts between a Python object and its JSON text representation.

````python
import json

data = {
    "name": "Pratham",
    "age": 23,
    "skills": ["Python", "SQL", "AI"],
    "is_student": False
}

# json.dump() -- writes a Python object directly to a FILE, as JSON text
with open("Data/profile.json", "w") as f:
    json.dump(data, f, indent=2)   # indent=2 makes it human-readable ("pretty-printed")

with open("Data/profile.json", "r") as f:
    print(f.read())
````

````python
# json.load() -- reads a JSON file back into a real Python object (dict, list, ...)
with open("Data/profile.json", "r") as f:
    loaded = json.load(f)

print(loaded)
print(type(loaded))        # <class 'dict'> -- a genuine Python dict, not a string
print(loaded["skills"])    # you can use it exactly like any other dict/list
````

````python
# json.dumps() / json.loads() -- same conversion, but with a STRING, not a file
json_string = json.dumps(data)   # "dumps" = dump to a STRING (note the trailing 's')
print(json_string)

back_to_dict = json.loads(json_string)   # "loads" = load from a STRING
print(back_to_dict["name"])
````

**Note:** The naming is consistent and worth memorizing: **`dump`** = write to a FILE, **`dumps`** = write to a STRING ("dump-string"). Same for reading: **`load`** = read from a FILE, **`loads`** = read from a STRING ("load-string"). Unlike CSV, JSON round-trips types correctly — `23` came back as the actual number `23`, and `False` came back as the actual boolean `False`, not text versions of them.

> ❌ Misconception: `json.dump()` and `json.dumps()` do fundamentally different conversions.
> ✅ Correct: they do the EXACT same conversion (Python object → JSON) — the only difference is the destination: `dump()` writes straight into a file object you give it, `dumps()` just returns the result as a string for you to use however you want (print it, send it over a network, etc.).

<a id="deleting-files"></a>
## 🗑️ Deleting Files and Directories

`open()` and friends only READ/WRITE content — actually removing a file or folder from disk uses the `os` (and `shutil`) modules instead.

````python
with open("Data/temp_delete_me.txt", "w") as f:
    f.write("This file is about to be deleted")

print(os.path.exists("Data/temp_delete_me.txt"))   # True

os.remove("Data/temp_delete_me.txt")   # deletes the file
print(os.path.exists("Data/temp_delete_me.txt"))   # False
````

````python
# Removing a file that doesn't exist -- raises FileNotFoundError, exactly like open() would
try:
    os.remove("Data/already_gone.txt")
except FileNotFoundError as e:
    print(f"FileNotFoundError: {e}")
````

````python
import shutil

# Creating a nested directory structure in one call
os.makedirs("Data/reports/2024", exist_ok=True)   # exist_ok=True -- don't crash if it already exists
print(os.path.exists("Data/reports/2024"))

# Removing a directory AND everything inside it
shutil.rmtree("Data/reports")
print(os.path.exists("Data/reports"))
````

**Note:** `os.remove()` only deletes FILES — trying it on a directory raises an error. `os.makedirs(path, exist_ok=True)` creates every missing folder along the path in one call (`Data/reports/2024` creates BOTH `reports` and `2024` if neither existed) — without `exist_ok=True`, calling it a second time would raise `FileExistsError`. `shutil.rmtree()` is the directory equivalent of `os.remove()`, but it's genuinely destructive — it deletes the folder and EVERYTHING inside it, with no confirmation prompt and no undo.

> ❌ Misconception: `os.remove()` works on both files and directories.
> ✅ Correct: `os.remove()` (and `os.unlink()`, its alias) only removes files. An empty directory needs `os.rmdir()`; a directory with contents needs `shutil.rmtree()`.

<a id="create-delete-rename"></a>
## 🔧 Create, Delete, and Rename — `os` vs `pathlib` Side-by-Side

Both `os` and `pathlib` can create, delete, and rename files/directories — `pathlib` just does it as METHODS on a `Path` object instead of separate functions that take a path string.

### Creating Files and Directories

````python
# os -- create a file
with open("Data/os_created.txt", "w") as f:
    f.write("Created via os / open()\n")
print(os.path.exists("Data/os_created.txt"))

# os -- create a nested directory in one call
os.makedirs("Data/os_folder/nested", exist_ok=True)
print(os.path.exists("Data/os_folder/nested"))
````

````python
# pathlib -- create a file
p = Path("Data/pathlib_created.txt")
p.touch()                              # creates an EMPTY file (like the Unix "touch" command)
p.write_text("Created via pathlib\n")   # write content directly -- no explicit open()/close() needed
print(p.exists())
print(p.read_text())                    # the read equivalent -- one call, no open()/close() either
````

````python
# pathlib -- create a nested directory
folder = Path("Data/pathlib_folder/nested")
folder.mkdir(parents=True, exist_ok=True)   # parents=True -- like os.makedirs; exist_ok=True -- like before
print(folder.exists())
````

**Note:** `p.write_text(...)`/`p.read_text()` are `pathlib` shortcuts that open the file, do the read/write, and close it again — all in one call, with no `with` block needed for simple cases. `Path.mkdir(parents=True, exist_ok=True)` is the direct equivalent of `os.makedirs(path, exist_ok=True)` — `parents=True` is what lets it create every missing folder along the way, not just the final one.

### Deleting Files and Directories

````python
# os -- delete a file
os.remove("Data/os_created.txt")
print(os.path.exists("Data/os_created.txt"))

# os -- delete an EMPTY directory (os.rmdir fails on a non-empty one)
try:
    os.rmdir("Data/os_folder")   # fails -- still contains "nested"
except OSError as e:
    print(f"OSError: {e}")

os.rmdir("Data/os_folder/nested")   # remove the inner (empty) folder first
os.rmdir("Data/os_folder")            # NOW the outer folder is empty too, so this works
print(os.path.exists("Data/os_folder"))
````

````python
# pathlib -- delete a file
p = Path("Data/pathlib_created.txt")
p.unlink()               # pathlib's equivalent of os.remove() -- deletes the file
print(p.exists())

# pathlib -- delete an EMPTY directory (Path.rmdir has the same non-empty restriction)
nested = Path("Data/pathlib_folder/nested")
nested.rmdir()
outer = nested.parent
outer.rmdir()   # NOW the outer folder is empty too, so this works -- same two-step pattern as os.rmdir() above
print(outer.exists())
````

**Note:** `Path.unlink()` is named after the low-level Unix system call, but it does exactly what `os.remove()` does — delete a file. Both `os.rmdir()` and `Path.rmdir()` refuse to remove a NON-empty directory (that's `shutil.rmtree()`'s job, shown earlier) — here, the inner `nested` folder has to be removed first, emptying the outer folder, before the outer one can be removed too.

### Renaming and Moving Files

````python
with open("Data/draft.txt", "w") as f:
    f.write("Some notes\n")

# os -- rename a file
os.rename("Data/draft.txt", "Data/final.txt")
print(os.path.exists("Data/draft.txt"))    # False -- the old name is gone
print(os.path.exists("Data/final.txt"))    # True -- it's the SAME file, just renamed
````

````python
# os.rename() also works as a MOVE, if the destination is in a DIFFERENT folder
os.makedirs("Data/archive", exist_ok=True)
os.rename("Data/final.txt", "Data/archive/final.txt")
print(os.path.exists("Data/final.txt"))              # False -- moved away from here
print(os.path.exists("Data/archive/final.txt"))       # True -- now lives here instead
````

````python
# pathlib -- rename a file
p = Path("Data/archive/final.txt")
new_path = p.rename("Data/archive/completed.txt")   # returns a NEW Path pointing to the renamed file

print(p.exists())            # False -- the OLD Path no longer points to a real file
print(new_path.exists())     # True -- rename() hands back a fresh Path for the new location
print(new_path)
````

**Note:** `os.rename(src, dst)` and `Path.rename(dst)` both do double duty as a MOVE operation, not just a same-folder rename — if `dst` is in a different directory, the file ends up there instead. The key difference in style: `os.rename()` always returns `None`, so you'd check the destination path yourself afterward; `Path.rename()` returns a brand-new `Path` object pointing at the new location, which is often more convenient to keep working with immediately. Neither one can rename/move ACROSS different drives/filesystems on some platforms — for that edge case, `shutil.move()` is the safer, more general choice.

> ❌ Misconception: `p.rename(...)` updates the ORIGINAL `Path` object (`p`) in place to reflect the new name.
> ✅ Correct: `p` still refers to the OLD path, which no longer exists — `rename()` returns a SEPARATE, new `Path` object; you have to capture it (`new_path = p.rename(...)`) to keep using it.

<a id="multiple-files-at-once"></a>
## 🔀 Multiple Files Open at Once

A single `with` statement can open more than one file at a time, separated by commas — genuinely common when copying or transforming data from one file into another.

````python
with open("Data/source.txt", "w") as f:
    f.write("Line A\nLine B\nLine C\n")

# Reading from ONE file and writing to ANOTHER, both managed by the same with statement
with open("Data/source.txt", "r") as infile, open("Data/destination.txt", "w") as outfile:
    for line in infile:
        outfile.write(line.upper())

with open("Data/destination.txt", "r") as f:
    print(f.read())
````

**Note:** `with open(a) as infile, open(b) as outfile:` opens BOTH files, and both are guaranteed to be closed properly when the block ends — including if an exception happens partway through, exactly like the single-file case earlier. This is the standard pattern for line-by-line file transformation: read one line from the source, process it, write it to the destination, without ever loading either file fully into memory.

| Pythonic idiom? | ✅ Generally preferred | ⚠️ Fine for simple scripts, but not bulletproof |

<a id="misconceptions"></a>
## ⚠ Common Misconceptions

❌ `f.close()` is optional if you're careful.
✅ An exception between `open()` and `close()` skips `close()` entirely — `with` is the safe way to guarantee it always runs.

❌ `"w"` mode preserves existing content and just adds to it.
✅ `"w"` erases the ENTIRE file the instant it's opened — use `"a"` if you want to preserve and add to existing content.

❌ `write()` and `writelines()` add newlines for you, like `print()` does.
✅ Neither one adds `\n` automatically — you must include it yourself wherever you want a line break.

❌ `readlines()` and iterating with `for line in f:` are the same thing.
✅ `readlines()` loads the WHOLE file into a list in memory first; `for line in f:` never does — it's the better choice for large files.

❌ Checking `os.path.exists()` before opening a file makes it impossible for `open()` to fail with `FileNotFoundError`.
✅ There's a gap between the check and the `open()` call — the file could be deleted in between. `try`/`except` (EAFP) doesn't have this gap.

❌ `FileNotFoundError` is the only exception you need to worry about when opening a file.
✅ A path might exist but be a DIRECTORY (`IsADirectoryError`), be unreadable due to permissions (`PermissionError`), or contain data that isn't valid text (`UnicodeDecodeError`) — each needs its own specific handling, not a one-size-fits-all catch.

❌ CSV files preserve the original Python types (numbers, booleans) of the data you wrote.
✅ CSV is plain text — everything read back from a CSV is a STRING, even values that started as numbers. JSON, by contrast, does preserve types correctly.

❌ `json.dump()` and `json.dumps()` (or `load()`/`loads()`) do different conversions.
✅ Same conversion either way — the only difference is FILE vs STRING as the source/destination (the trailing `s` means "string").

❌ `os.remove()` can delete a directory, since it "removes" things.
✅ `os.remove()` only works on files. Deleting a directory needs `os.rmdir()` (empty only) or `shutil.rmtree()` (with contents).

❌ `p.rename(...)` updates the original `Path` object in place.
✅ It returns a brand-new `Path` pointing at the new location — the old `Path` object still refers to a path that no longer exists.

<a id="quick-revision"></a>
## 📝 Quick Revision

- `open(path, mode)` returns a file object; it must be closed (manually, or via `with`) to guarantee data is actually saved.
- `with open(...) as f:` is the standard, exception-safe way to work with files — `close()` runs automatically no matter how the block exits.
- `"r"` = read only, `"w"` = write (ERASES existing content), `"a"` = append (keeps existing content), `"x"` = create-only (fails if it exists).
- `"r+"`, `"w+"`, `"a+"` combine read + write, following the same truncate/append rules as `r`/`w`/`a`.
- Add `"b"` for binary mode (`"rb"`, `"wb"`) — works with `bytes` instead of `str`, for non-text files.
- `read()`/`readlines()` load everything into memory; iterating with `for line in f:` (or `readline()`) processes one line at a time — better for large files.
- `write()`/`writelines()` never add `\n` automatically — you must include it yourself.
- `seek(n)` jumps to position `n` in the file; `tell()` reports the current position.
- `os.path`/`pathlib` check whether a path exists, is a file/directory, or inspect its name/extension.
- EAFP (`try`/`except FileNotFoundError`) is generally preferred over LBYL (`if os.path.exists(...)`) — no race-condition gap between checking and acting.
- Beyond `FileNotFoundError`: `IsADirectoryError` (path is a folder), `PermissionError` (no access rights), and `UnicodeDecodeError` (binary data opened in text mode) are common file-related exceptions worth catching by name.
- The `csv` module reads/writes tabular data; everything comes back as strings. The `json` module reads/writes nested Python objects and preserves their real types.
- `os.remove()` deletes files; `os.makedirs(path, exist_ok=True)` creates nested folders; `shutil.rmtree()` deletes a folder and everything inside it.
- `os` and `pathlib` can both create, delete, and rename — `pathlib`'s versions are METHODS on a `Path` (`p.touch()`, `p.unlink()`, `p.rename()`) instead of standalone functions taking a path string.

<a id="key-takeaways"></a>
## ✅ Key Takeaways

1. `open(path, mode)` gives you a file object; it must be closed to guarantee data is actually saved — `with` does this automatically, even when an exception happens.
2. `"w"` erases existing content, `"a"` preserves it and adds to the end, `"x"` refuses to overwrite anything — picking the wrong mode is one of the most common file-handling bugs.
3. `read()`/`readlines()` load the whole file into memory; iterating with `for line in f:` processes one line at a time — the better choice for large files.
4. `write()`/`writelines()` never add `\n` automatically, unlike `print()` — you have to include it yourself.
5. `seek()`/`tell()` give you direct control over the file's current read/write position — essential for the `r+`/`w+`/`a+` combined modes.
6. EAFP (`try`/`except FileNotFoundError`) is generally safer than checking `os.path.exists()` first, because there's no gap where the file could disappear between the check and the actual `open()`.
7. File operations can fail in more ways than just "missing" — `IsADirectoryError`, `PermissionError`, and `UnicodeDecodeError` each signal a genuinely different problem, and catching them by name (rather than a bare `except:`) lets your code respond appropriately to each one.
8. The `csv` module handles tabular data but returns everything as strings; the `json` module handles nested Python objects and preserves their real types (numbers, booleans, `None`).
9. Deleting is a separate concern from reading/writing — `os.remove()` for files, `shutil.rmtree()` for directories with contents — and both are permanent, with no built-in undo.
10. `os` and `pathlib` offer the same create/delete/rename operations in two different styles — standalone functions taking a path string, vs. methods on a `Path` object — and `Path.rename()` returns a new `Path` rather than modifying the original in place.
