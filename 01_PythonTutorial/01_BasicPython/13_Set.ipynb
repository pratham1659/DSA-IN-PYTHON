## 1. 🧺 What is a Set?

A set is a collection of unique data, meaning that elements within a set cannot be duplicated.

For instance, if we need to store information about student IDs, a set is suitable since student IDs cannot have duplicates.

## 🎯 Learning Goal

By the end of this notebook you should understand:
- What a set is, and why it can never contain duplicates
- The difference between an **alias** (`set2 = set1`) and an actual **copy** (`set1.copy()`)
- The core set operations — union, intersection, difference, symmetric difference — and their `|`/`&`/`-`/`^` shortcuts
- Why membership checks (`x in my_set`) are so much faster on a set than on a list

> 🧑‍🎓 **Analogy:** a set is like a **guest list at the door** — a name is either on it or it isn't, there's no such thing as "on the list twice," and nobody's assigned a numbered seat (no indexing).

**Why do we need sets specifically?**
- To automatically remove duplicates from a collection (`set(some_list)` is the classic one-liner for this)
- To answer "is this here?" questions very fast, no matter how large the collection gets
- To do math-style set operations (union, intersection, etc.) directly, without writing loops yourself

### 🧠 Key Idea

- A set stores **unique, unordered** items — adding a duplicate is silently ignored, and there's no guaranteed order to rely on.
- Sets are **mutable** (you can add/remove items after creation) but every *item inside* a set must be **hashable** (immutable) — numbers, strings, and tuples work; lists and dicts don't.
- Membership testing (`x in my_set`) and `add`/`remove` are **O(1) on average** — a set is backed by a hash table, the same structure behind a dict, not a plain array like a list.
- `set1.copy()` makes a real, independent copy; plain assignment (`set2 = set1`) just gives you a second name for the *same* set.

### 📚 Important Terms

| Term | Simple Meaning | Example |
|------|----------------|----------|
| Set | An unordered collection of unique items | `{1, 2, 3}` |
| Hashable | A value that can't change after creation, so Python can index it in a hash table | numbers, strings, tuples — not lists/dicts |
| Alias | A second variable name pointing at the *same* object | `set2 = set1` |
| Copy | A brand-new, independent object with the same contents | `set1.copy()` |
| Union (`\|`) | Everything in either set | `{1,2} \| {2,3}` → `{1,2,3}` |
| Intersection (`&`) | Only what's in both sets | `{1,2} & {2,3}` → `{2}` |
| Difference (`-`) | In the first set, but not the second | `{1,2} - {2,3}` → `{1}` |
| Symmetric difference (`^`) | In exactly one of the two sets, not both | `{1,2} ^ {2,3}` → `{1,3}` |

**⚖ Set vs. List vs. Dict:**

| | Set | List | Dict |
|---|---|---|---|
| Ordered? | No | Yes | Insertion-ordered (3.7+) |
| Duplicates? | Never | Allowed | Keys never, values can repeat |
| Indexable (`x[0]`)? | No | Yes | By key, not position |
| `x in ...` speed | O(1) average | O(n) | O(1) average (by key) |

```python
list1 = [2, 3, 3, 4, 2, "hello", "world", "live", 55, 66, "hello", "live"]
print(list1)
print(len(list1))

## count number of unique elements in the list
set1 = set(list1)
print(set1)
print(type(set1))
print(len(set1))
```

```python
list1 = [2, 3, 3, 4, 2, "hello", "world", "live", 55, 66, "hello", "live"]
print(list1)

set1 = set(list1)
print(set1)
print(len(set1))

set1.add(1)
set1.add(2)
print(set1)
print(len(set1))
```

**Note:** `.add()` only inserts a value if it isn't already there — adding `1` and `2` here grows the set by exactly 2, but adding a value that's already present (like re-adding `"hello"`) would silently do nothing and leave the size unchanged.

## 2. ➕ Set Operations (add, discard, remove, membership)

```python
list1 = [2, 3, 3, 4, 2, "hello", "world", "live", 55, 66, "hello", "live"]
print(list1)

set1 = set(list1)
print(set1)
print(len(set1))

set1.add(1) ## add operation
set1.add(2)
print(set1)
print(len(set1))

set1.discard("hello") ## discard operation
set1.remove(2) ## Remove operation

set1.discard("hiii") ## element which is not availble will not give error in discard
# set1.remove("hii") ## element which is not available will give error in remove
print(set1)
print(len(set1))
```

```python
set1 = {2, 3, 3, 4, 2, "hello", "world", "live", 55, 66, "hello", "live"}

if 2 in set1:
    print(True)
else:
    print(False)
    
print(set1)  
set1.clear()

print(set1)

## Note : Indexing not available in set because elements were not stored in order
```

**Note:** `set1.clear()` empties the set **in place** — `set1` still exists afterward, just as `set()` (empty), which is why the second `print(set1)` shows `set()` instead of an error.

⏱️ **Time & Space:** `x in set1` is **O(1) on average** (hash lookup) — the whole reason sets are the right tool for "have I seen this before?" checks, versus `x in list1` which is **O(n)** (has to scan every element).

```python
## Copy in set
set1 = {2, 3, 4}
set2 = set1 ## This is an ALIAS -- same object, not a copy at all

set3 = set1.copy() ## This IS a (shallow) copy -- a new, independent set

set1.add(5)
print(set1)
print(set2)
print(set3)
```

**Gotcha — alias vs. copy (fixed above):** `set2 = set1` does **not** create a new set — it just gives `set1` a second name. Both variables point at the exact same object in memory, so `set1.add(5)` changes what `set2` shows too. `set1.copy()` is the one that actually makes an independent set — that's why `set3` is unaffected by the `.add(5)` below.

> 📌 **Trick:** if changing one variable silently changes another, you have an alias, not a copy — the fix is almost always `.copy()`.

```python
set1 = {1, 8, 5, 2, 4, 6}
set2 = {8, 2, 9, 6}
set3 = { 3, 6, 7, 8, 3, 2, 9}

## print union
print(set1 | set2)
print(set1.union(set3))


## print Intersection
print(set1 & set2)
print(set1.intersection(set3))


## print difference
print(set1 - set2)
print(set1.difference(set3))

## print symemetric difference
print(set1 ^ set3)
print(set1.symmetric_difference(set3))
```

**Note:** every set operation has two equivalent spellings — an operator (`|`, `&`, `-`, `^`) and a method (`.union()`, `.intersection()`, `.difference()`, `.symmetric_difference()`). They do the exact same thing; the methods are handy when you want to pass a plain list/tuple instead of another set.

⏱️ **Time & Space:** these all run in roughly **O(len(set1) + len(set2)) time** — Python still has to look at every element at least once, but hashing keeps each individual lookup O(1).

## 6. 📚 Set Methods Reference

* add()	Adds an element to the set
* clear()	Removes all the elements from the set
* copy()	Returns a copy of the set
* difference()	Returns a set containing the difference between two or more sets
* difference_update()	Removes the items in this set that are also included in another, specified set
* discard()	Remove the specified item
* intersection()	Returns a set, that is the intersection of two other sets
* intersection_update()	Removes the items in this set that are not present in other, specified set(s)
* isdisjoint()	Returns whether two sets have a intersection or not
* issubset()	Returns whether another set contains this set or not
* issuperset()	Returns whether this set contains another set or not
* pop()	Removes an element from the set
* remove()	Removes the specified element
* symmetric_difference()	Returns a set with the symmetric differences of two sets
* symmetric_difference_update()	inserts the symmetric differences from this set and another
* union()	Return a set containing the union of sets
* update()	Update the set with the union of this set and others

## 9. 🏋️ Practice Problems


**Problem 1 — remove duplicates while keeping the original order.** A plain `set(list1)` removes duplicates but scrambles the order — here's how to keep the first-seen order:

```python
def dedupe_keep_order(items):
    seen = set()
    result = []
    for item in items:
        if item not in seen:
            seen.add(item)
            result.append(item)
    return result

dedupe_keep_order([3, 1, 3, 2, 1, 4])   # -> [3, 1, 2, 4]
```

**Problem 2 — find common elements between two lists, fast.** Converting to sets turns an O(n×m) nested-loop search into an O(n+m) one:

```python
list_a = [1, 2, 3, 4, 5]
list_b = [4, 5, 6, 7]

common = set(list_a) & set(list_b)
print(common)   # -> {4, 5}
```

**Problem 3 — check whether one list's items are a subset of another's:

```python
required_skills = {"python", "sql"}
candidate_skills = {"python", "sql", "excel", "communication"}

print(required_skills.issubset(candidate_skills))   # -> True: candidate has everything required
```

## 🎓 Cheat Sheet

| Operation | Syntax | Notes |
|---|---|---|
| Create | `{1, 2, 3}` or `set(some_list)` | `{}` alone is a dict, not a set |
| Add one | `s.add(x)` | Ignored if `x` already present |
| Add many | `s.update([x, y, z])` | |
| Remove (strict) | `s.remove(x)` | Raises `KeyError` if missing |
| Remove (safe) | `s.discard(x)` | No error if missing |
| Remove arbitrary | `s.pop()` | Order is not guaranteed |
| Empty it | `s.clear()` | Set still exists, just empty |
| Delete entirely | `del s` | Variable no longer exists |
| Copy | `s.copy()` | `s2 = s` is an alias, NOT a copy |
| Union | `s1 \| s2` / `s1.union(s2)` | Everything in either |
| Intersection | `s1 & s2` / `s1.intersection(s2)` | Only what's in both |
| Difference | `s1 - s2` / `s1.difference(s2)` | In s1, not s2 |
| Symmetric diff. | `s1 ^ s2` | In exactly one, not both |
| Membership | `x in s` | O(1) average — much faster than `x in list` |
| Immutable set | `frozenset(...)` | Can't `.add()`/`.remove()` afterward |

## ⚠ Common Misconceptions

❌ Sets are immutable.
✅ The **set itself** is mutable (`.add()`, `.remove()`, etc. all work) — only the *items inside* it must be immutable/hashable.

❌ `set2 = set1` makes a copy.
✅ It makes an **alias** — both names point at the same object. Use `.copy()` for a real, independent copy.

❌ `.pop()` removes the last item you added.
✅ It removes an **arbitrary** item — sets have no order, so there's no "last."

❌ `{}` creates an empty set.
✅ `{}` is always an empty **dict**. Use `set()` for an empty set.

## 🎯 Key Takeaways

1. A set stores unique, unordered, hashable items — duplicates are silently dropped, and there's no indexing.
2. The set itself is mutable; only the items inside it must be immutable. `frozenset` is the fully-immutable version.
3. `set2 = set1` is an alias (same object); `set1.copy()` is a real, independent copy — mixing these up is the #1 set bug.
4. `.remove()` raises `KeyError` if missing, `.discard()` doesn't, and `.pop()` removes an arbitrary (not "last") item.
5. Membership checks and add/remove are O(1) on average thanks to hashing — that's the whole reason to reach for a set over a list.
6. `{}` is always an empty dict, never an empty set — that one keeps tripping people up because of the shared `{}` syntax.
