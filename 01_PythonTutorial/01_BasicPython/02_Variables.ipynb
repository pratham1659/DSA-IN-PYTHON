<a id="what-is-it"></a>
## 🤔 What is it?

A **variable** is a name that points to a value stored in memory. A **data type** describes what *kind* of value that is (a whole number, text, true/false, etc.), which determines what you can do with it.

> 🧸 Think of a variable like a labeled sticky note on a box in storage. The label (`age`) is the variable name; what's inside the box (`25`) is the value; and the *kind* of thing inside (a number vs. a photo vs. a document) is the data type.

---

<a id="why-do-we-need-it"></a>
## ❓ Why do we need it?

- Without variables, you'd have to retype the same value everywhere it's used, and update every copy by hand if it changes.
- Data types matter because operations only make sense for certain types — you can add two numbers, but adding a number to a list doesn't mean anything, so Python blocks it.
- Knowing a value's type tells you what tools (methods, operators) are available on it.

---

<a id="key-idea"></a>
## 🧠 Key Idea

- Python is **dynamically typed** — you never declare a variable's type up front; Python figures it out from the value you assign.
- A variable can be **reassigned to a completely different type** later (`x = 10` then `x = "hello"` is legal — `x` just now points to a string).
- Every value has a type, checkable with `type(x)`.
- Some type combinations require explicit conversion (**typecasting**) before they can be combined — e.g., you can't add a string and an integer directly.

<a id="variable-naming-rules"></a>
## 📛 Variable Naming Rules

- Must start with a **letter or underscore** — never a digit.
- Can contain letters, digits, and underscores after that.
- **Case-sensitive** — `age` and `Age` are two different variables.
- Cannot be a Python **reserved keyword** (like `class`, `for`, `if`).
- Convention: use `snake_case` (lowercase words separated by underscores) for variable names.

````python
# Valid variable names
age = 25
_age = 25
age2 = 25
my_age = 25

# Invalid variable names (uncommenting any of these raises a SyntaxError):
# 2age = 25       -> can't start with a digit
# my-age = 25     -> hyphens aren't allowed in names
# class = 25      -> "class" is a reserved keyword

print(age, _age, age2, my_age)
````

<a id="local-and-global-variables"></a>
## 🌐 Local and Global Variables

A variable's **scope** is where in your code it can be seen and used. Python cares about two scopes here:

- **Local variable** — created inside a function. It only exists while that function is running, and disappears the moment the function finishes. Nothing outside the function can see it.
- **Global variable** — created at the top level of your script (outside any function). Every function can freely READ it — but changing it from inside a function needs one extra step (see below).

> 🧸 Think of a local variable like something you whisper inside a closed room — only people in that room hear it, and once you leave, it's gone. A global variable is like an announcement over a building's loudspeaker — everyone, everywhere, can hear it.

````python
def greet():
    message = "Hello from inside the function"  # local variable -- only exists here
    print(message)

greet()
print(message)  # NameError: 'message' doesn't exist outside greet()
````

````python
counter = 0  # global variable -- created at the top level

def show_counter():
    print(counter)  # reading a global variable from inside a function is fine, no keyword needed

show_counter()
print(counter)
````

<a id="modifying-a-global-variable"></a>
### Modifying a Global Variable Inside a Function

Reading a global variable inside a function works with no special syntax. But if you try to **reassign** it inside a function, Python assumes you're creating a brand-new local variable with the same name — it does NOT touch the outer/global one.

> ❌ Misconception: `counter += 1` inside a function updates the global `counter`.
> ✅ Correct: Python sees the assignment and decides `counter` is local to that function. Since a local `counter` doesn't have a value yet when `counter += 1` tries to read it, this raises `UnboundLocalError` — not the same thing as a plain `NameError`.

````python
counter = 0

def increment_broken():
    counter += 1  # UnboundLocalError -- Python treats this "counter" as a new local variable
    print(counter)

increment_broken()
````

The fix: the **`global`** keyword tells Python "don't create a local variable — use the outer/global one."

````python
counter = 0

def increment_fixed():
    global counter   # use the outer global counter, don't create a local one
    counter += 1
    print(counter)

increment_fixed()
increment_fixed()
print("Final value:", counter)
````

| Scope | Created where | Visible from | Lifetime |
|-------|---------------|---------------|----------|
| Local | Inside a function | Only inside that same function | Destroyed as soon as the function returns |
| Global | Top level of the file (outside any function) | Anywhere, including inside functions (read freely; needs `global` to reassign) | Lives as long as the program runs |

````python
x = 10
print(x, type(x))
````

Less common in everyday code, but Python supports complex numbers natively — the `j` suffix marks the imaginary part.

<a id="bool-type"></a>
### `bool` — True or False

A boolean is the result of any comparison (`>`, `<`, `==`, ...). Under the hood, `True` behaves like `1` and `False` like `0`.

````python
num = 10 > 9
print(num, type(num))

is_active = True
is_deleted = False
print(is_active, is_deleted, type(is_active))
````

<a id="nonetype"></a>
### `NoneType` — Representing "Nothing"

`None` is Python's way of saying "no value here (yet)." It's commonly used as a placeholder before a variable has a real value, or as a function's default return when nothing is explicitly returned.

> ❌ Misconception: `None` is the same as `0`, `False`, or an empty string.
> ✅ Correct: `None` is its own distinct type (`NoneType`), meaning "no value at all" — not even zero.

````python
result = None
print(result, type(result))
print(result == 0)      # False -- None is not 0
print(result is None)   # True -- this is the correct way to check for None
````

<a id="str-type"></a>
### `str` — Text

Strings support `len()` to count characters, and indexing to access individual characters (0-based, just like lists).

````python
name = "edureka"
#       0123456
print(len(name))
print(type(name))
````

````python
print(name[2])  # 3rd character (index 2) -> "u"
````

<a id="fstrings"></a>
## 🎨 Formatted Output with f-strings

An f-string (`f"..."`) lets you embed variables and formatting directly inside a string. The tricky part is the **format spec** after the colon — `:2f` and `:.2f` look almost identical but mean very different things.

> ❌ Misconception: `f"{x:2f}"` rounds `x` to 2 decimal places.
> ✅ Correct: `2f` sets a minimum **width** of 2 characters (still 6 decimal places by default). The **dot** is what means "decimal places" — `.2f` rounds to exactly 2 decimals.

````python
x = 3.14159

print(f"{x:2f}")    # no dot -> just a minimum width of 2, still 6 decimals by default
print(f"{x:.2f}")   # the dot means "2 decimal places" -> rounds to 3.14
print(f"{x:10.2f}")  # combine both: minimum width 10, 2 decimal places
````

## 💻 Mixed Data Types in Practice

````python
var1 = "Hello print"          # string
var2 = "43"                     # string (looks like a number, but it's text!)
var3 = 34                       # integer
var4 = 34.6                     # float
var5 = 34424232646363646        # integer (Python has no fixed integer size limit)

print(type(var1))
print(var1 + var2)   # string + string -> concatenation, NOT addition
print(type(var2))
print(type(var3))
print(type(var4))
print(type(var5))
````

### Mixing Incompatible Types Raises `TypeError`

Python won't silently guess what you meant when combining unrelated types — it stops you with an error, forcing you to be explicit about the conversion (typecasting) you want.

````python
a = [1, 2]
a + "hi"   # TypeError: can only concatenate list (not "str") to list
````

<a id="typecasting"></a>
## 🔄 TypeCasting in Python

**Typecasting** means converting a value from one data type to another — using functions like `int()`, `float()`, `str()`. `var2` above is the string `"43"`, not the number `43` — so `var2 + var3` would fail unless we explicitly cast `var2` to an int first.

````python
var1 = "Hello print"
var2 = "43"
var3 = 34

print(int(var2) + var3)  # cast the string "43" to an int, THEN add
````

<a id="types-of-typecasting"></a>
#### Types of Typecasting

1. **Implicit Typecasting** — performed automatically BY Python (e.g. `int + float` → `float`, no conversion function needed).
2. **Explicit Typecasting** — performed BY the programmer, using `int()`, `float()`, `str()`, etc.

<a id="explicit-typecasting"></a>
##### Example: Explicit Typecasting

Converting `string` to an `int` is something YOU choose to do — Python won't do it for you. If the string isn't a valid number, this raises a `ValueError`.

````python
string = "15"
number = 7

str_num = int(string)  # explicit -- raises ValueError if the string isn't a valid number
total = number + str_num
print("The sum of both values is:", total)
````

<a id="implicit-typecasting"></a>
##### Example: Implicit Typecasting

Adding an `int` and a `float` — Python automatically promotes the result to a `float`, with no conversion function needed.

````python
var1 = 7
var2 = 3.0

ans = var1 + var2
print(ans)          # int + float -> float
print(type(ans))
````

<a id="common-misconceptions"></a>
## ⚠ Common Misconceptions

❌ `input()` returns a number if the user types digits.
✅ `input()` ALWAYS returns a string — you must explicitly convert it with `int()` or `float()` before doing math.

❌ `None`, `0`, `False`, and `""` are all "the same as nothing."
✅ They're different types with different meanings. `None` specifically means "no value at all" — check for it with `is None`, not `== 0`.

❌ `f"{x:2f}"` and `f"{x:.2f}"` do the same thing.
✅ `2f` sets a minimum width; `.2f` sets the number of decimal places. The dot is what controls rounding.

❌ Once you assign a type to a variable, it's stuck with that type.
✅ Python is dynamically typed — `x = 10` then `x = "ten"` is completely legal; `x` just now refers to a different value of a different type.

❌ Naming a variable `sum`, `list`, `type`, or `str` is harmless.
✅ These are built-in function/type names in Python. Reassigning them (like `sum = 5`) shadows the built-in for the rest of that scope, which can cause confusing bugs later if you try to use the real `sum()`.

❌ Assigning to a variable inside a function updates the global variable with the same name.
✅ Without the `global` keyword, that assignment creates a brand-new LOCAL variable instead — the outer global variable is left untouched, and reading it beforehand in the same function raises `UnboundLocalError`, not `NameError`.

## 📝 Quick Revision

- A variable is a name pointing to a value; Python infers its type automatically (dynamic typing).
- Variable names must start with a letter/underscore, can't be a reserved keyword, and are case-sensitive.
- Local variables live only inside the function that creates them; global variables live at the top level and are visible everywhere.
- Functions can READ a global variable freely, but need the `global` keyword to REASSIGN it — otherwise Python creates a new local variable instead.
- Core data types: `int`, `float`, `complex`, `bool`, `str`, and `NoneType`.
- `type(x)` reveals a value's data type.
- Typecasting converts between types — implicit (Python does it automatically, e.g. int + float) or explicit (you call `int()`/`float()`/`str()` yourself).
- `input()` always returns a string — cast it before doing math.
- f-string format specs: `.2f` means "2 decimal places"; `2f` alone just sets a minimum width.
- Avoid naming variables after built-ins (`sum`, `list`, `type`) — it shadows the real function.
