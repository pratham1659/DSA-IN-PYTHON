## 🔄 Type Conversion (Typecasting)

### 🤔 What is it?

**Type Conversion** (or typecasting) means converting a value from one data type to another — like turning a number into text, or text into a number — using functions like `str()`, `int()`, `float()`.

> 🧸 Think of it like currency exchange. A number and a string are different "currencies" — Python won't let you mix them directly (`10 + "Rohan"` fails), but once you convert one to match the other, they work together fine.

### ❓ Why do we need it?

`input()` always gives you a string, but often you need a number to do math. Without converting it first, Python throws a `TypeError` because it doesn't know how to combine incompatible types.

### 💻 Technical Example

Below, `x + y` fails because `x` is an integer and `y` is a string — Python can't automatically combine a number and text. Converting `x` to a string with `str(x)` fixes it, because now both sides are text and `+` just joins them together.

````python
y = "Rohan"
x = 10

x + y
````

````python
str(x) + y
````

### 🌍 Real-Life Example

A classic use of typecasting: asking for someone's birth year (text from `input()`) and subtracting it from the current year (a number) to calculate their age. This only works once the birth year is converted to an integer.

````python
# This is Typecasting
birth_year = input("write your birth Year : ")
age = 2022 - int(birth_year)
print(age)
````

More typecasting examples: `int()` → `float()`, `list()` → `set()` (which also removes duplicates!), and converting strings to numbers to do math before turning the result back into a string.

````python
a=2
f=float(a)
print("Integer a =", a )
print("float a =", f )
l = [1,3,3,4,5]
s= set(l)
print(s)
print(l)
var1 = "20" #String integer Typecasting
var2 = "30" #String Typecasting
print(10 * str(int(var1) + int(var2)), " ")
````

````python
weight_lbs = input("weight in LBS ")
weight_kg = int(weight_lbs) * 0.45
print(weight_kg)
````

> 💡 **Quotes trick:** Python lets you use single `'...'` or double `"..."` quotes for strings. If your text contains an apostrophe (like `Python's`), wrap it in double quotes. If it contains double quotes, wrap it in single quotes. This avoids "escaping" characters.

````python
course1 = "Python's for Beginners"
course2 = 'Python for "Beginners"'
print(course1, "\n", course2)
````

## ⌨ Escape Sequence Characters

### 🤔 What is it?

An **escape sequence** is a special character combo starting with a backslash `\` that represents something you can't type directly in a string — like a new line or a tab space.

> 🧸 Think of `\` as a "special instruction" flag. `\n` doesn't print a literal `n` — it tells Python "insert a line break here instead."

### 📚 Important Terms

| Escape Sequence | Simple Meaning |
|------------------|-----------------|
| `\n` | New line (move to the next line) |
| `\t` | Tab (adds horizontal spacing) |
| `\\` | A literal backslash character |
| `\'` | A literal single quote inside a `'...'` string |
| `\"` | A literal double quote inside a `"..."` string |

> ⚠ Note: a lone `\` followed by a space or letter that isn't a recognized escape sequence (like `\ P`) is usually just a mistake — Python will print it close to literally, which is confusing. Use `\\` if you actually want a backslash character.

````python
print("hello \n pratham")   # \n -> new line
print("hello \t Pratham Good Boy")  # \t -> tab space
print("This is a backslash: \\")   # \\ -> literal backslash
````

## 🐍 Indentation in Python

### 🤔 What is it?

Python uses **indentation** (spaces at the start of a line) instead of curly braces `{}` to show which lines belong together in a block of code.

> 🧸 Think of indentation like paragraph structure in an essay — it visually shows what belongs together. Python is strict about this: mixing it up breaks the "sentence."

### ❓ Why do we need it?

Other languages (like Java or C++) use `{ }` to group code. Python instead relies on **consistent spacing** to know where a block starts and ends. This keeps code clean and readable, but it also means incorrect spacing causes real errors — not just messy formatting.

### ⚠ What is `IndentationError`?

If a line has spacing that doesn't match the structure Python expects (e.g., an unexpected extra space with no reason), Python throws an `IndentationError` and refuses to run the code.

## 💬 Comments in Python

### 🤔 What is it?

A **comment** is a note in your code that Python ignores when running — it's there purely for humans to read.

> 🧸 Think of comments like sticky notes on a recipe card — they don't change the dish, but they help the cook (or a future you) remember why a step is done a certain way.

### 📚 Important Terms

| Type | Simple Meaning | Example |
|------|----------------|----------|
| Single-line comment | Starts with `#`, ends at the end of that line | `# this is a comment` |
| Multi-line comment | Multiple `#` lines, OR a triple-quoted string used as a comment | See below |
| Docstring | A triple-quoted string used to document a function/class — NOT ignored like a comment, it's actually stored as real data | `"""This function adds two numbers"""` |

### Single-Line Comment

````python
# Adding 2 variables and printing the result in a new variable
a = 2
b = 3
c = a + b
print(c)
````

### Multi-Line Comment

Python doesn't have a dedicated multi-line comment symbol — most people either stack multiple `#` lines, or use a triple-quoted string (`""" ... """`) as a throwaway value that Python evaluates but discards.

````python
var1 = "Hello print"
var2 = "43"
var22 = "34"
var3 = 34.6
var4 = 34424

print(type(var3))               # float
print(var3 + var4)              # float + int works fine -> 34458.6
print(100 * int(var22) + var3)  # typecasting var22 to int before math -> 3434.6
print(100 * str(int(var2) + int(var22)))  # int math first, then cast to str and repeat it 100 times
````

## 🧮 Built-in Modules in Daily Life — `math` and `random`

### 🤔 What is it?

Python ships with dozens of **built-in modules** — ready to `import`, no installation needed — that cover common, everyday needs. Two of the most-used ones are `math` (mathematical functions and constants) and `random` (generating randomness).

> 🧸 Think of built-in modules like the toolbox that comes free with a new house — a hammer and screwdriver are already there; you don't need to go buy them separately, unlike specialty tools (third-party packages) you'd order online.

### 📚 Important Terms

| Term | Simple Meaning | Example |
|------|----------------|----------|
| Built-in module | Ships with Python itself, no install needed | `math`, `random`, `os` |
| `math.sqrt()` | Square root of a number | `math.sqrt(64)` → `8.0` |
| `math.ceil()` / `math.floor()` | Round UP / round DOWN to the nearest whole number | `math.ceil(4.3)` → `5` |
| `random.randint(a, b)` | A random WHOLE number between `a` and `b`, inclusive | `random.randint(1, 6)` |
| `random.choice(seq)` | Picks ONE random item from a list | `random.choice(["a","b"])` |

````python
import math

print(math.sqrt(64))        # square root -- e.g. finding the side of a square room from its floor area
print(math.pow(2, 10))       # 2 raised to the power 10 -- pow() always returns a float
print(math.pi)                 # the constant pi -- useful for anything involving circles
print(math.ceil(4.3))           # rounds UP -- e.g. you need 5 pizza boxes even if you only filled 4.3
print(math.floor(4.7))           # rounds DOWN
print(math.factorial(5))          # 5! = 5*4*3*2*1 -- used in combinatorics/probability
````

````python
import random

print(random.random())                 # a random float between 0.0 and 1.0
print(random.randint(1, 6))              # simulates rolling a single six-sided die
print(random.choice(["rock", "paper", "scissors"]))   # picks ONE item at random -- like rock-paper-scissors

cards = ["Ace", "King", "Queen", "Jack"]
random.shuffle(cards)   # shuffles the list IN PLACE -- like shuffling a deck of cards
print(cards)
````

**Note:** `math.sqrt()`/`math.pow()`/`math.ceil()`/`math.floor()`/`math.factorial()` all show up constantly outside the classroom — splitting a bill evenly and rounding UP so nobody underpays, figuring out how many boxes you need, computing compound interest. `random.random()` gives a float between 0 and 1; `random.randint(a, b)` gives a WHOLE number, which is why it's the one used for dice, lottery numbers, and OTP-style codes. `random.choice()` picks one item (a raffle winner, "spin the wheel"), and `random.shuffle()` rearranges a list in place (an actual card shuffle, or randomizing a playlist).

## 🖩 Building a Calculator

### 🤔 What is it?

A simple calculator function takes two numbers and an operator, and returns the result — the exact same `if`/`elif`/`else` pattern you already used for typecasting decisions, now applied to arithmetic.

````python
def calculator(a, b, operator):
    if operator == "+":
        return a + b
    elif operator == "-":
        return a - b
    elif operator == "*":
        return a * b
    elif operator == "/":
        if b == 0:
            print("Error: cannot divide by zero")
            return None
        return a / b
    else:
        print(f"Unknown operator: {operator}")
        return None


print(calculator(10, 5, "+"))
print(calculator(10, 5, "-"))
print(calculator(10, 5, "*"))
print(calculator(10, 5, "/"))
print(calculator(10, 0, "/"))    # the divide-by-zero guard kicks in
print(calculator(10, 5, "%"))    # an operator we never defined
````

**Note:** The divide-by-zero check runs BEFORE the actual division — without it, `a / b` would raise `ZeroDivisionError` and crash the whole program. The final `else` branch is what makes `calculator(10, 5, "%")` print a helpful message instead of silently returning nothing or crashing with a confusing error.

### Extending It — a Scientific Calculator Using `math`

````python
def scientific_calculator(value, operation):
    if operation == "sqrt":
        return math.sqrt(value)
    elif operation == "square":
        return math.pow(value, 2)
    elif operation == "factorial":
        return math.factorial(value)
    else:
        print(f"Unknown operation: {operation}")
        return None


print(scientific_calculator(81, "sqrt"))
print(scientific_calculator(7, "square"))
print(scientific_calculator(5, "factorial"))
````

**Note:** `scientific_calculator()` follows the exact same "dictionary of operations, chosen by a string key" idea as the first `calculator()` — the only difference is which module's functions it calls underneath (`math` instead of plain `+`/`-`/`*`/`/`).

### A Daily-Life Use of `random` — Rolling Dice

````python
def roll_dice(num_rolls=1):
    return [random.randint(1, 6) for _ in range(num_rolls)]


rolls = roll_dice(3)
print("You rolled:", rolls)
print("Total:", sum(rolls))
````

**Note:** `roll_dice(3)` uses a list comprehension to call `random.randint(1, 6)` three separate times, collecting each result — this is the same "simulate a physical dice roll" logic behind any board-game app or tabletop-gaming website that offers a "roll for me" button.

## ▶️ `__name__` and `if __name__ == "__main__":`

### 🤔 What is it?

Every Python file (and every running notebook) has a hidden variable called `__name__`. When you run a file **directly** (as the main program), `__name__` is set to the string `"__main__"`. But when that SAME file is **imported** as a module from somewhere else, `__name__` is set to the module's own name instead.

> 🧸 Think of `__name__` like a name tag at a party. If you're the one throwing the party, your tag says "HOST" (`"__main__"`). If you're a guest who was invited by someone else's party (imported by another file), your tag says your actual name instead — Python can tell the difference, and your code can check the tag and behave differently depending on which one it is.

### ❓ Why do we need it?

Without this check, importing a file to reuse one of its functions would ALSO re-run every "run this now" line in that file — printing things, opening windows, kicking off a whole program — none of which the importer asked for. `if __name__ == "__main__":` lets a file say: "run THIS part only when I'm the one being executed directly, not when someone else just wants to borrow a function from me."

````python
print(__name__)   # "__main__" -- because THIS notebook/script is the one being run directly
````

**Note:** This prints `"__main__"` because this code is running as the top-level program right now — nothing else imported it. To actually SEE `__name__` change, we need a second file that gets imported rather than run directly. Let's create one on disk and prove it.

### 💻 Technical Example — Proving It With a Real Module File

This writes a small `.py` file to disk (using the file-writing skills from earlier), then `import`s it — which lets us see `__name__` switch from `"__main__"` to the module's own name.

````python
with open("greeting_module.py", "w") as f:
    f.write('''print("Module is being loaded / executed")

def greet():
    print("Hello from greeting_module!")

if __name__ == "__main__":
    print("Running directly! __name__ =", __name__)
    greet()
else:
    print("Imported as a module! __name__ =", __name__)
''')
````

````python
import greeting_module   # the module-level print() runs immediately, and __name__ inside it is NOT "__main__"

greeting_module.greet()          # we can still call its function normally, after importing it
print(greeting_module.__name__)  # "greeting_module" -- its OWN name, not "__main__"
````

**Note:** Notice the `if __name__ == "__main__":` block inside `greeting_module.py` never ran — it printed "Imported as a module!" instead, because `import` sets `__name__` to `"greeting_module"` there, not `"__main__"`. If you instead ran that same file directly (e.g. `python greeting_module.py` from a terminal), `__name__` inside IT would be `"__main__"`, and the `if` branch would run instead. This is exactly why real Python scripts almost always end with:
```python
if __name__ == "__main__":
    main()
```
— it lets the same file work both as a standalone script AND as an importable module, without accidentally re-running its "main program" logic every time someone just wants to reuse one of its functions.

> ❌ Misconception: `__name__` always equals the file's own name.
> ✅ Correct: `__name__` equals `"__main__"` ONLY for whichever file was run directly — every OTHER file that gets `import`ed gets its own module name instead, even though it's the exact same kind of `.py` file.

## 🏗 Classes and Objects

### 🤔 What is it?

A **class** is a blueprint for creating objects. An **object** is a specific "thing" built from that blueprint, with its own data.

> 🧸 Think of a class like a cookie cutter, and objects like the cookies made from it. The cutter (class) defines the *shape* — every cookie (object) made from it shares that shape, but each cookie can still be decorated differently.

### ❓ Why do we need it?

Classes let us group related data (**attributes**) and behavior (**methods**) together, so we can create many similar objects without repeating code. This is the foundation of **Object-Oriented Programming (OOP)**.

### 📚 Important Terms

| Term | Simple Meaning | Example |
|------|----------------|----------|
| Class | The blueprint/template | `class Car:` |
| Object | A specific instance made from the class | `BMW = Car()` |
| Attribute | A piece of data stored on the object | `attr1 = "Petrol"` |
| Method | A function defined inside a class | `def start(self):` |
| `self` | Refers to "this specific object" inside a method | `self.attr1` |
| `__init__` | A special method that runs automatically when an object is created (the "constructor") | `def __init__(self, name):` |

### 🌍 Real-Life Example

Think of `Car` as the blueprint for "what a car is" (has fuel type, horsepower, can start). `BMW = Car()` builds one actual car from that blueprint. You could make another car, `Tesla = Car()`, and it would have the same *structure*, but be a completely separate object.

````python
# Python3 program to
# demonstrate instantiating
# a class
class Car:
    # A simple class
    # attribute
    attr1 = "Petrol"
    attr2 = "750 HP"
    # A sample method 
    def start(self):
        print("Engine Started : Engine Type ", self.attr1)
        print("Ready to GO : Horse Power ", self.attr2)
# Driver code
# Object instantiation
BMW = Car()
# Accessing class attributes
# and method through objects
print(BMW.attr1)
BMW.start()
````

### The `__init__` Constructor

Most real classes use `__init__` to set up an object's data automatically when it's created, instead of hardcoding attributes like `Car` did above. `self.name = name` stores the value passed in on the object itself, so every `Person` object can have a different name.

> 📌 **Easy trick:** `self` always refers to "the object being worked on right now." When you write `self.name`, think "this specific person's name," not some global name.

````python
class Person:
    # init method or constructor 
    def __init__(self, name):
        self.name = name
    # Sample Method 
    def say_hi(self):
        print('Hello, my name is', self.name)
p = Person('Robert')
p.say_hi()
````

- `#` starts a single-line comment; multiple `#` lines or a throwaway `"""..."""` can serve as a multi-line comment.
- A docstring looks like a comment but is actually stored as real data, unlike a comment which is discarded.
- `math` (sqrt, pow, ceil, floor, factorial, pi) and `random` (random, randint, choice, shuffle) cover everyday needs from rounding bills to rolling dice.
- A simple calculator is just an `if`/`elif`/`else` chain mapping an operator string to the matching arithmetic operation.
- `__name__` is `"__main__"` only in the file that's run directly; an imported file's `__name__` is its own module name instead — `if __name__ == "__main__":` uses this to separate "always run" code from "only run when this file is the main program" code.
- A class is a blueprint; an object is a specific instance built from it.
- `self` refers to the current object inside a class method; `__init__` runs automatically when an object is created.

<a id="key-takeaways"></a>
## ✅ Key Takeaways

1. `input()` always returns a string — convert it with `int()`/`float()` before doing math (typecasting).
2. Escape sequences like `\n` and `\t` let you represent special characters inside strings.
3. Python relies on indentation to define code blocks — get the spacing wrong and you get an `IndentationError`.
4. Comments (`#`) are discarded by Python, but docstrings (`"""..."""` at the start of a function/class) are stored as real, retrievable data.
5. Built-in modules like `math` and `random` need no installation at all, unlike third-party packages (`numpy`, `pandas`) — and they cover genuinely everyday tasks: rounding, roots, dice rolls, shuffling.
6. A calculator function is just a decision tree (`if`/`elif`/`else`) mapping an operator to an operation — always guard against invalid input (like dividing by zero) BEFORE performing it.
7. `__name__` equals `"__main__"` only in the file being run directly; `if __name__ == "__main__":` lets a file work both as a standalone script and as a safely importable module.
8. A class is a blueprint for objects — `self` refers to the specific object a method is currently running on, and `__init__` sets up a new object's data automatically.
